## Install MATLAB to be used on interactive Python Jupyter Notebooks

### Step 1: Install MATLAB API for Python

You have to install the [MATLAB API for Python](https://mathworks.com/help/matlab/matlab-engine-for-python.html) first. The installation process is described [here](https://mathworks.com/help/matlab/matlab_external/install-the-matlab-engine-for-python.html), but the following should give a summary. To do so, 

* first identify where MATLAB is installed on the hard disk. This could be a folder like `C:\Program Files\MATLAB\R2021b`.
* Open a Anaconda Prompt 
  * If you are using Windows, the Anaconda Prompt has to be opened with Administrator rights.
    Run this in a Anaconda Prompt. Replace `<MATLABROOT>` with the path you just identified.

    ```bash
    cd <MATLABROOT>\extern\engines\python\
    python setup.py install
    ```


## Step 2: Install MATLAB Kernel for Python

Run this in a terminal window. 
```bash
conda install -c conda-forge matlab_kernel
```
or 
```bash
pip install matlab_kernel
```

Launch either `jupyter notebook` or jupyter jab` and select `New > Matlab`.
