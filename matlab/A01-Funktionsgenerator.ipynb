## Funktionsgenerator - Kapitel 2.1

In dieser Übung soll ein Funktionsgenerator programmiert werden, der ein Signal mit der Frequenz $f_0$  
bei einer Abtastfrequenz von $f_\mathrm{A}$ erzeugt.  
Dieser Generator soll in der Lage sein, ein Sinus- und ein Sägezahnsignal zu generieren.

Für ein zeitkontinuierliches System wird üblicherweise mit der Kreisfrequenz  
$\omega_0 = 2 \pi f_0$ gerechnet.  
Damit ergibt sich ein zeitkontinuierliches Sinussignal zu

$$
x_\mathrm{sin,K}(t) = \sin(\omega_0 \cdot t)
$$

In einem zeitdiskreten System wird das Sinussignal mit Abtastperioden der Länge  
$T = 1 / f_\mathrm{A}$ abgetastet, sodass die Zeit  
$t = T \cdot k$ durch die Indizes $k = 0, 1, 2, ...$ dargestellt werden kann.  
Daraus ergibt sich

$$
x_\mathrm{sin}(k) = x_\mathrm{sin,K}(t)|_{t = kT} = \sin(\omega_0 \cdot T \cdot k)
$$

Für zeitdiskrete Systeme ist es sinnvoll, die **normierte Kreisfrequenz** $\Omega_0$ einzuführen,  
die keine Einheit besitzt:

$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T \\
         & =  & 2 \pi f_0 \cdot T \\
         & =  & 2 \pi f_0 / f_\mathrm{A}
\end{array}
$$

Diese Definition wird in den folgenden Aufgaben verwendet, da sie die Behandlung digitaler Systeme  
ohne Einschränkung der Abtastfrequenz ermöglicht.  
Damit gilt:

$$
x_\mathrm{sin}(k) = \sin(\Omega_0 \cdot k)
$$
