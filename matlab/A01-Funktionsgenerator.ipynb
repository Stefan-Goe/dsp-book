## Theorie zur Erzeugung eines diskreten Sinus Signals 

In dieser Aufgabe soll ein Funktionsgenerator programmiert werden,  der  ein Nutzsignal der Frequenz $f_0$ erzeugt und bei einer festen Abtastfrequenz $f_\mathrm{A}$ arbeitet. Dabei soll ein Sinus- und ein Sägezahnsignal erzeugt werden.

Bei zeitkontinuierlichen Systemen ist es üblich, mit der Kreisfrequenz $\omega_0 = 2 \pi f_0$ zu rechnen.
Damit ergibt sich das zeitkontinuierliche Sinussignal zu

$$
x_\mathrm{sin,K}(t)=\sin \left( \omega_0 \cdot t \right)\,.
$$

In einem diskreten System wird das Sinussignal in Perioden der Dauer $T = 1/f_\mathrm{A}$  abgetastet, so daß die diskreten Zeitpunkte $t=T_\mathrm{A} \cdot k$ jetzt mit dem Zeitindex $k =0, 1 , 2,...$  beschrieben werden können. Damit ergibt sich

$$
x_\mathrm{sin}[k]= \left.x_\mathrm{sin,K}(t)\right|_{t=k T_\mathrm{A}} 
= \sin \left( \omega_0 \cdot T_\mathrm{A} \cdot k \right)\,.
$$

In dem abgetasteten System ist es somit sinnvoll, eine normierte Kreisfrequenz

$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T_\mathrm{A} \\
         & =  & 2 \pi f_0 \cdot T_\mathrm{A} \\
         & =  & 2 \pi f_0 / f_\mathrm{A} \\
\end{array}
$$

einzuführen. Diese normierte Frequenz besitzt keine [Einheit](https://de.wikipedia.org/wiki/Internationales_Einheitensystem "Weitere Informationen zu Einheiten auf Wikipedia"). Diese Definition wird im weiteren immer verwendet, da sie die  Behandlung digitaler Systeme ohne konkrete Festlegung der Abtastfrequenz erlaubt.
Es gilt somit

$$
x_\mathrm{sin}[k] = \sin \left( \Omega_0 \cdot k \right)\,.
$$

Erzeugen Sie dazu einen Koeffizientenvektor <code>k=[0, 1, ..., 59]</code>der den Zeitindex des abgetasteten Signals $x(k)$ darstellt. Nutzen Sie die <code>sin()</code>-Funktion von <code>Matlab</code>.

```matlab
fa=8000;        % Festsetzen der Abtastrate

% # Berechnung des Signalvektors # %
k=0:59;             % Erstellen eines Vektors der Laenge 60 
                    % mit den Eintraegen 0 bis 59
f0=400;             % Signalfrequenz
Omega0=2*pi*f0/fa;  % Auf Abtastrate normierte Frequenz
xsin=sin(Omega0*k); % Sinussignalvektor

% # Grafische Ausgabe # %
figure;                 % Neue Figure öffnen
stem(k,xsin);           % Diskretes 'plotten' des Signals xsin; 
                        % k ist der Vektor mit den Werten auf der x-Achse
xlabel('k');            % Achsenbeschriftung an der x-Achse
ylabel('x_{sin}[k]');   % Achsenbeschriftung an der y-Achse
```

##  Teilaufgabe b:  Sinussignal, veraenderte Periode 



*  Ver"andern Sie die Frequenz des abgetasteten Sinussignals auf
  $f_0=960\,\mathrm{Hz}$
  und stellen Sie das entstehende Signal erneut mit {\tt stem()} dar.
  
  Wie gro"s ist jetzt die Periode des digitalen Signals?
   
	  
* Erzeugen Sie nun ein S"agezahnsignal mit der Frequenz $f_0=960\,\mathrm{Hz}$ bei einer Abtastfrequenz von $f_{A}=8\,\mathrm{kHz}$.
   
   Das S"agezahnsignal soll im Wertebereich $0... 1$ liegen.
   Verwenden Sie zur Berechnung die {\sc Matlab}-Funktion
   {\tt mod()}.
   Die Normierung der Signalfrequenz $f_0$ auf die Abtastfrequenz $f_{A}$
   k"onnen sie analog zu den Berechnungen bei dem Sinussignal vornehmen.
   Beachten Sie jedoch, da"s Sie die Periodendauer der {\tt mod()}
   Funktion im Gegensatz zu der $2\pi$-Periodiztit"at der {\tt sin()}-Funktion selbst
   festlegen.
   
   Stellen Sie das S"agezahnsignal mit {\tt stem()} dar und  vergleichen Sie es
   mit der Sinussignal. Stimmt die Signalfrequenz?
   
* Erzeugen Sie jeweils ein Vektor mit einem Sinus- und einem S"agezahnsignal 
   mit $f_0=1\,\mathrm{kHz}$ bei einer Abtastfrequenz von $f_{A}=8\,\mathrm{kHz}$.
   
   Die Signale sollen "uber eine Soundkarte ausgegeben werden und
   daher eine Abspieldauer von $d=2~\mathrm{s}$ haben.\\
   Wieviele Abtastwerte sind daf"ur notwendig?
   
   Geben Sie die Signale jeweils mit Hilfe des {\sc Matlab}-Befehls 
   {\tt sound(x,fa)} über die Soundkarte ihres Rechners aus.
\end{itemize}   

```matlab
% # Berechnung des Signalvektors # %
f0=960;                 % Signalfrequenz
Omega0=2*pi*f0/fa;      % Normierte Frequenz
xsin=sin(Omega0*k);     % Sinussignalvektor 

% # Grafische Ausgabe # %
figure;         
stem(k,xsin);   
xlabel('k');
ylabel('x_{sin}[k]');
```

##  Teilaufgabe c:  Saegezahnsignal

```matlab
xsaw=mod(f0/fa*k,1);    % Saegezahnsignalvektor

% # Grafische Ausgabe # %
figure; 
stem(k,xsaw); 
xlabel('k'); 
ylabel('x_{saw}[k]');
```

##  Teilaufgabe d:  entnormierte Signale, Tonerzeugung

```matlab
f0=1000;                % Signalfrequenz
d=2;                    % Abspieldauer in Sekunden
w=d*fa;                 % Anzahl der Abtastwerte
k=0:w;                  % Koeffizientenvektor des Zeitindexes
xsin=sin(2*pi*f0/fa*k); % Sinussignalvektor
xsaw=mod(f0/fa*k,1);    % Sägezahnsignalvektor
```

```matlab
% # Wiedergabe der Signale # % 
disp('Sie hoeren: 2s Sinussignal');     
sound(xsin,fa);
```

```matlab
disp('Sie hoeren: 2s Saegezahnsignal'); 
sound(xsaw,fa);
```
