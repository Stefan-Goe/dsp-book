# Entwurf von nichtrekursiven Filtern mit Fensterverfahren

Das vollständige Filterdesign durch Fensterung einer zeitverschobenen Impulsantwort eines idealen Tiefpassfilters wird in **MATLAB** mit der Funktion `fir1()` umgesetzt.

Aufruf:

```matlab
b = fir1(m, Wn)
```

erzeugt $m+1$ Filterkoeffizienten eines Tiefpassfilters der Ordnung `m` mit der Grenzfrequenz `Wn`. Der Aufruf

```matlab
b = fir1(m, Wn, 'high')
```

liefert den entsprechenden Hochpassfilter mit Grenzfrequenz `Wn`.

```matlab
b = fir1(m, [Wn1 Wn2])
```

liefert ein Bandpassfilter mit den Grenzfrequenzen `Wn1` und `Wn2`.

Als **Standardfenster** wird das Hamming-Fenster verwendet. Zusätzlich kann der letzte Parameter ein Vektor mit Abtastwerten eines beliebigen Fensterfunktion sein:

```matlab
b = fir1(m, Wn, window)
```

Dieser Vektor muss $m+1$ Abtastwerte enthalten.
