# Sine Generator

In this exercise a sine generator for a fixed frequency shall be synthesized, which is realized without using any tables or polynomial approximations. The basis of this generator is a 2nd order system which, once excited, produces the values of a sine signal.

For the synthesis process the following correspondence of the Z-transform shall be used:

$$
h(k)=\sin!\big(\Omega_0 k\big)
$$

$$
H(z) = Z{h(k)} = \frac{z,\sin(\Omega_0)}{z^2 - 2z\cdot\cos(\Omega_0) + 1}
$$

This correspondence can be interpreted as the impulse response $h(k)$ of a system $H(z)=Y(z)/X(z)$.
