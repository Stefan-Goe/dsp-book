# Design of Non-Recursive Filters Using Windowing Techniques

The complete filter design by windowing an impulse response of a time-shifted, ideal low-pass filter is implemented in MATLAB using the function `fir1()`.

Calling

```matlab
b = fir1(m, Wn)
```

generates $m+1$ filter coefficients of a low-pass filter with order `m` and cut-off frequency `Wn`. The function call

```matlab
b = fir1(m, Wn, 'high')
```

yields the corresponding high-pass filter with cut-off frequency `Wn`.

```matlab
b = fir1(m, [Wn1 Wn2])
```

yields a pass-band filter with cut-off frequencies `Wn1` and `Wn2`.

The Hamming window is used as **default**. In addition, the last parameter may be a vector with samples of an arbitrary window function:

```matlab
b = fir1(m, Wn, window)
```

This vector must cover $m+1$ samples.
