## Funktionsgenerator - Kapitel 2.1

In dieser Übung soll ein Funktionsgenerator programmiert werden, der ein Signal mit der Frequenz $f_0$   bei einer Abtastfrequenz von $f_\mathrm{A}$ erzeugt.  

Dieser Generator soll in der Lage sein, ein Sinus- und ein Sägezahnsignal zu generieren.

Für ein zeitkontinuierliches System wird üblicherweise mit der Kreisfrequenz $\omega_0 = 2 \pi f_0$ gerechnet.  

Damit ergibt sich ein zeitkontinuierliches Sinussignal zu

$$
x_\mathrm{sin,K}(t) = \sin(\omega_0 \cdot t)
$$

In einem zeitdiskreten System wird das Sinussignal mit Abtastperioden der Länge $T = 1 / f_\mathrm{A}$ abgetastet, sodass die Zeit $t = T \cdot k$ durch die Indizes $k = 0, 1, 2, ...$ dargestellt werden kann.  
Daraus ergibt sich

$$
x_\mathrm{sin}(k) = x_\mathrm{sin,K}(t)|_{t = kT} = \sin(\omega_0 \cdot T \cdot k)
$$

Für zeitdiskrete Systeme ist es sinnvoll, die **normierte Kreisfrequenz** $\Omega_0$ einzuführen, die keine Einheit besitzt:

$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T \\
         & =  & 2 \pi f_0 \cdot T \\
         & =  & 2 \pi f_0 / f_\mathrm{A}
\end{array}
$$

Diese Definition wird in den folgenden Aufgaben verwendet, da sie die Behandlung digitaler Systeme  ohne Einschränkung der Abtastfrequenz ermöglicht.  

Damit gilt:

$$
x_\mathrm{sin}(k) = \sin(\Omega_0 \cdot k)
$$

### a)

Erstellen Sie einen Vektor `x` mit 60 Abtastwerten eines Sinussignals mit der Frequenz $f_0 = 400\,\text{Hz}$ bei einer Abtastfrequenz von $8\,\text{kHz}$.

Initialisiere dazu einen Koeffizientenvektor `k = [0, 1, ..., 59]`, der den Zeitindex des abgetasteten Signals $x(k)$ darstellt. Verwenden Sie dazu die MATLAB-Funktion `sin()`.

Stelle das Signal mit der Funktion `stem()` dar.  Wie groß ist die Periode des abgetasteten Signals?

### b)

Ändere die Frequenz des abgetasteten Sinussignals auf  
$f_0 = 960\,\text{Hz}$  
und stelle das resultierende Signal erneut mit `stem()` dar.

Wie groß ist nun die Periode des abgetasteten Signals?

### c)

Erzeuge nun ein Sägezahnsignal mit der Frequenz  
$f_0 = 960\,\text{Hz}$ bei einer Abtastfrequenz von $f_\mathrm{A} = 8\,\text{kHz}$.

Das Sägezahnsignal soll einen Amplitudenbereich von $0 \ldots 1$ haben.  
Zur Berechnung verwende die MATLAB-Funktion `mod()`.  
Die Normierung der Signalfrequenz $f_0$ durch die Abtastfrequenz $f_\mathrm{A}$  
kann analog zu oben erfolgen.  
Beachte dabei, dass du die Periode der `mod()`-Funktion selbst festlegst – im Gegensatz zur $2\pi$-Periodizität der `sin()`-Funktion.

Stelle das Sägezahnsignal mit `stem()` dar und vergleiche es mit dem Sinussignal.  
Ist die Signalfrequenz korrekt?

### d)

Erzeuge einen Vektor sowohl für ein Sinus- als auch für ein Sägezahnsignal mit  
$f_0 = 1\,\text{kHz}$ bei einer Abtastfrequenz von $f_\mathrm{A} = 8\,\text{kHz}$.

Die Signale sollen über eine Soundkarte wiedergegeben werden und eine Dauer von $d = 2\,\text{s}$ haben.  
Wie viele Abtastwerte sind dafür erforderlich?

Spiele die Signale mit dem MATLAB-Befehl  
`sound(x, fa)` ab.

```python
# BEGIN SOLUTION
close all;      % alle Figures schliessen
clear all;      % alle Variablen löschen

fa=8000;        % Festsetzen der Abtastrate


% #####  Teilaufgabe a:  Sinussignal  #####
% # Berechnung des Signalvektors # %
k=0:59;             % Erstellen eines Vektors der Laenge 60 mit den Eintraegen 
                    % 0 bis 59
f0=400;             % Signalfrequenz
Omega0=2*pi*f0/fa;  % Auf Abtastrate normierte Frequenz
xsin=sin(Omega0*k); % Sinussignalvektor

% # Grafische Ausgabe # %
figure;                 % Neue Figure öffnen
stem(k,xsin);           % Diskretes 'plotten' des Signals xsin; k ist der Vektor
                        % mit den Werten auf der x-Achse
xlabel('k');            % Achsenbeschriftung an der x-Achse
ylabel('x_{sin}(k)');   % Achsenbeschriftung an der y-Achse


% #####  Teilaufgabe b:  Sinussignal, veraenderte Periode  #####
% # Berechnung des Signalvektors # %
f0=960;                 % Signalfrequenz
Omega0=2*pi*f0/fa;      % Normierte Frequenz
xsin=sin(Omega0*k);     % Sinussignalvektor 

% # Grafische Ausgabe # %
figure;         
stem(k,xsin);   
xlabel('k');
ylabel('x_{sin}(k)');
                
                
% #####  Teilaufgabe c:  Saegezahnsignal  #####
xsaw=mod(f0/fa*k,1);    % Saegezahnsignalvektor

% # Grafische Ausgabe # %
figure; 
stem(k,xsaw); 
xlabel('k'); 
ylabel('x_{saw}(k)');

% #####  Teilaufgabe d:  entnormierte Signale, Tonerzeugung  #####
f0=1000;                % Signalfrequenz
d=2;                    % Abspieldauer in Sekunden
w=d*fa;                 % Anzahl der Abtastwerte
k=0:w;                  % Koeffizientenvektor des Zeitindexes
xsin=sin(2*pi*f0/fa*k); % Sinussignalvektor
xsaw=mod(f0/fa*k,1);    % S�gezahnsignalvektor

% # Wiedergabe der Signale # % 
disp('Sie hoeren: 2s Sinussignal');     sound(xsin,fa);
pause(length(xsin) / fa)
disp('Sie hoeren: 2s Saegezahnsignal'); sound(xsaw,fa);
pause(length(xsaw) / fa)
# END SOLUTION
```

## Lösung — Funktionsgenerator (Kapitel 2.1)

**a)**  
Das Sinussignal hat den folgenden Verlauf:

<img src="./pngs/ldsv2a20_a1.png" width="250">

Die Periode des zeitkontinuierlichen Signals $x_\text{sin,K}(t) = \sin(2\pi f_0 \cdot t)$ beträgt $T_0 = 1 / f_0$. Da die Anzahl der Abtastwerte pro Periode des zeitkontinuierlichen Signals eine ganze Zahl ist, muss das zeitdiskrete Signal eine Periodenlänge von

$$
\begin{array}{rcl}
k_p &=& T_0 / T \\
    &=& f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 400\,\text{Hz} \\
    &=& 20
\end{array}
$$

Werten haben.

**b)**  
Nach Erhöhung der Signalfrequenz auf $f_0 = 960\,\text{Hz}$  
stimmen die Perioden des zeitdiskreten und des zeitkontinuierlichen Signals nicht mehr überein.

<img src="./pngs/ldsv2a20_b1.png" width="250">

Die Periode des zeitdiskreten Signals tritt nach

$$
\begin{array}{rcl}
k_p &=& L \cdot T_0 / T \\
    &=& L \cdot f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 960\,\text{Hz} \\
    &=& L \cdot 8.\overline{3} \\
    &\Downarrow& L = 3 \\
    &=& 25
\end{array}
$$

Abtastwerten auf.  
Dabei ist $L = 3$ die Anzahl der Perioden des kontinuierlichen Signals, nach denen sich die Periodizität des abgetasteten Signals einstellt.

**c)**  
Analog zum Sinussignal kann die Normierung durch die Abtastfrequenz auf die gleiche Weise erfolgen:

$$
x_\text{saw}(k) = \mathrm{mod}\!\left( \frac{f_0}{f_A} \cdot k , 1 \right)
$$

Die Abbildung zeigt das Sägezahnsignal, dessen Periode — ähnlich wie beim Sinussignal in b) —  
$k_p = 25$ Abtastwerte umfasst.

<img src="./pngs/ldsv2a20_c1.png" width="250">

**d)**  
Für eine Signaldauer von $d = 2\,\text{s}$ gilt:

$$
w = \frac{d}{T} = d \cdot f_A = 16000
$$

Es werden also **16 000 Abtastwerte** benötigt.
