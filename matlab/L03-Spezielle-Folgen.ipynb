## Übung: Spezielle Folgen

Diese Übung demonstriert einige grundlegende Reihen für die digitale Signalverarbeitung.

* **a)** Erzeuge eine Impulsreihe
  $$
  x(k) = 1.2\cdot \delta(k-4), \qquad -5 \leq k \leq 10
  $$
  mit
  $$
  \delta(k):=\begin{cases}
  1 & k=0\
  0 & k\neq 0
  \end{cases}
  $$
  und stelle diese Reihe mit der Funktion `stem()` dar.

* **b)** Stelle die folgenden gaußschen und laplaceschen Reihen mit der Funktion `stem()` dar.
  Warum ist es ungünstig, den Befehl `plot()` zur Darstellung der Folgen zu verwenden?

 | Nr. | Signaldefinition | Wertebereich | Parameter | Beschreibung |
|:---:|:-----------------|:--------------|:-----------|:--------------|
| 1 | $x_1(k) = e^{-\alpha k^2}$ | $-20 \leq k \leq 20$ | $\alpha = 0.01$ | Gaussian series |
| 2 | $x_2(k) = e^{-\beta\,\|k\|}$ | $-20 \leq k \leq 20$ | $\beta = 0.2$ | Laplace series |
| 3 | $x_3(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{4})$ | $-30 \leq k \leq 30$ | $\beta = 0.1$ | mod. Laplacian series |
| 4 | $x_4(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{80})$ | $-600 \leq k \leq 600$ | $\beta = 0.005$ | — |


```python
# BEGIN SOLUTION
% #####  Teilaufgabe a:  Impulsfolge  #####
k_a = -5:10;    % initialisieren des Koeffizientenvektors k mit Werten von 
                % -5 .. 10;
x = zeros(1,16);% x wird als ZEILEN-Vektor mit 16 Nullen initialisiert;
x(5+1+4) = 1;   % Der Diracimpuls '1' soll beim zeitlichen Index k=4 erscheinen;
                % Der zeitliche Index k=-5 entspricht dem Matlab-Index '1',
                % daher muss man f�r k=4 den Matlabe-Index 10 w�hlen;
                 
x = 1.2 * x;    % Multiplikation der Folge mit dem Faktor 1.2 ;

% # grafiche Ausgabe # %
figure; 
stem(k_a,x);
set(gca,'XTick',[-4:2:10]); 
title('x(k)=1.2 \delta(k-4)'); 
xlabel('Abtastwerte k'); 
ylabel('Amplitude');

% #####  Teilaufgabe b:  Gauss- und Laplacefolgen  #####
% Koeffizientenvektoren
k_b1 = -20:20;
k_b2 = -20:20;
k_b3 = -30:30;
k_b4 = -600:600;

% Alpha-Faktor f�r Gau�folge
alpha = 0.01;

% Beta-Faktoren f�r Laplacefolgen
beta_2 = 0.2;
beta_3 = 0.1;
beta_4 = 0.005;

% Berechnung der Folgen
x1 = exp(-alpha *k_b1.^2);
x2 = exp(-beta_2*abs(k_b2));
x3 = exp(-beta_3*abs(k_b3)) .* cos(2*pi/4*k_b3);
x4 = exp(-beta_4*abs(k_b4)) .* cos(2*pi/80*k_b4);

% # grafische Ausgabe # %
% Gau�folge
figure;
plot(k_b1,x1);
hold on; 
stem(k_b1,x1,'k'); 
hold off;
title('Gaussfolge x_1(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% Laplacefolge
figure;
plot(k_b2,x2);
hold on;
stem(k_b2,x2,'k'); 
hold off;
title('Laplacefolge x_2(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% modifizierte Laplacefolge beta = 0.1
figure;
plot(k_b3,x3);
hold on;
stem(k_b3,x3,'k');  
hold off;
grid on; 
title('Modulierte Laplacefolge x_3(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% modifizierte Laplacefolge beta = 0.005
figure;
plot(k_b4,x4,'k');
grid on;
title('Modulierte Laplacefolge  x_4(k)'); 
xlabel('k'); 
ylabel('Amplitude');
# END SOLUTION
```
