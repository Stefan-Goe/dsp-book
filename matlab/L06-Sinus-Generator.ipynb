# Sinusgenerator

In dieser Aufgabe soll ein Sinusgenerator für eine feste Frequenz realisiert werden, der **ohne Tabellen oder Polynomapproximationen** arbeitet. Die Grundlage dieses Generators ist ein **System 2. Ordnung**, das nach einmaliger Anregung die Werte eines Sinussignals liefert.

Für den Syntheseprozess soll folgende Z-Transformationskorrespondenz verwendet werden:

$$
h(k) = \sin!\big(\Omega_0 k\big)
$$

$$
H(z) = Z{h(k)} = \frac{z ,\sin(\Omega_0)}{z^2 - 2 z \cdot \cos(\Omega_0) + 1}
$$

Diese Korrespondenz kann als **Impulsantwort** $h(k)$ eines Systems $H(z) = Y(z)/X(z)$ interpretiert werden.

---

### a)

Zeichnen Sie ein **Blockschaltbild** des resultierenden Sinusgenerators.

**Hinweis:** Formen Sie die obige Gleichung zunächst in eine Darstellung um, die nur Verzögerungen enthält (Terme der Form $z^{-d}$).

---

### b)

Realisieren Sie das System, indem Sie seine **Koeffizientenvektoren** bestimmen:

```text
b = [b_0, b_1, b_2]  
a = [a_0, a_1, a_2]
```

für die Frequenz $\Omega_0 = 1/32$.

Zeichnen Sie das **Nullstellen-Pol-Plot** mit der Funktion `zplane()`.

Bestimmen Sie die **Impulsantwort** des Systems mit der Funktion `impz()`. Vergleichen Sie das Ergebnis mit einem gesampelten Sinussignal der Frequenz $\Omega_0 = 1/32$.

---

### c)

Welche **Auswirkung** hat der Vorwärtsweg dieses Systems?

Was passiert, wenn das Verzögerungsglied und das Multiplikationselement im Vorwärtsweg entfernt werden?

Überprüfen Sie Ihre Annahme, indem Sie die Koeffizientenvektoren `a` und `b` entsprechend ändern und die Impulsantwort neu berechnen.

### d)

Untersuchen Sie nun die Effekte der **Quantisierung der Systemkoeffizienten**.

Ersetzen Sie dazu den Ausdruck $2 \cdot \cos(\Omega_0)$ im Koeffizientenvektor `a` durch $2.1 \cdot \cos(\Omega_0)$.

Zeigen Sie das **Nullstellen-Pol-Plot** und die Impulsantwort des Systems.

````python
# BEGIN SOLUTION
% normierte dimensionslose Kreisfrequenz
Omega0 = 2*pi*1/32;

% #####  Teilaufgabe b:  Sinus-Generator  #####
b = [0 sin(Omega0) 0]
a = [1 -2*cos(Omega0) 1]

% abgetastetes Sinussignal
k = 0:99; 
x = sin(Omega0*k);

% # grafische Ausgabe # %
% Pol- und Nullstellen in der z-Ebene
figure; 
zplane(b,a);
xlabel('Realteil'); 
ylabel('Imagin�rteil');
title('z-Ebene');

% Impulsantwort
figure; 
impz(b,a,100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulsantwort');
hold on; 
plot(k,x,'r-'); 
hold off;

% #####  Teilaufgabe c:  Vorwaertszweig  #####
b = [1 0 0]
a = [1 -2*cos(Omega0) 1]

% # grafische Ausgabe # %
figure; 
impz(b,a,100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulsantwort');

% #####  Teilaufgabe d:  Instabilitaet  #####
b = [0 sin(Omega0) 0]
a = [1 -2.1*cos(Omega0) 1]

% # grafische Ausgabe # %
% Pol- und Nullstellen in der z-Ebene
figure; 
zplane(b,a);
xlabel('Realteil'); 
ylabel('Imagin�rteil');
title('z-Ebene');

% Impulsantwort
figure; 
impz(b,a,100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulsantwort');
# END SOLUTION
````
