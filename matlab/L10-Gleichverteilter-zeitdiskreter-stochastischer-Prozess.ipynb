### Teilaufgabe b: Histogramm

```matlab
n = rand(N,1);
[H,x] = hist(n,30);
H = H/N;      % Damit H eine Schaetzung der Wahrscheinlichkeitsdichte
              % ist, muss die Flaeche eins sein

% # grafische Ausgabe # %              
figure; 
bar(x,H);
grid; 
xlabel('x'); 
ylabel('f_X(x)'); 
title(['Gleichvert. Mittelw.=',num2str(mean(n)),' Var.=',num2str(std(n).^2)]);
```
