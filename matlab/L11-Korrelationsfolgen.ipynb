# Exercise 2.14: Korrelationsreihe

Neben den Momenten aus Aufgabe~\ref{dsv2a11e} gibt es noch weitere charakteristische Parameter für stochastische Prozesse, d. h. *die Autokorrelation, Kreuzkorrelation sowie die Autokovarianz* bzw. *Kreuzkovarianz*.

Die gemeinsame Eigenschaft dieser Funktionen ist: Sie sind Maßzahlen
für die Abhängigkeit der beobachteten Prozesse und ihrer
Zufallsvariablen. In dieser Aufgabe wird beispielhaft die
Autokorrelationsreihe betrachtet. Die folgenden stochastischen
Prozesse werden als *ergodisch* angenommen, siehe Abschnitt (2.6).  
Ein Hauptproblem ist die Schätzung der theoretischen
Autokorrelationsreihe (2.6.6):

$$
r_{XX}(\kappa) = E\{X^\ast(k)\, X(k+\kappa)\}
$$

basierend auf einer endlichen Anzahl $N$ von Stichproben.

Die MATLAB-Funktion `xcorr()` schätzt eine
Auto-/Kreuzkorrelationsreihe eines stochastischen
Prozesses.  
Der Aufruf `r = xcorr(x)` erstellt eine nicht
normierte Autokorrelationsreihe aus dem Vektor `x` unter Verwendung
der Formel  

$$
\hat{r}_{XX}(\kappa) = \sum_{k=0}^{N-|\kappa|-1} x(k)\,x^{\ast}(k+\kappa)
$$

Im Gegensatz zu Gl. (2.6.6) ist die Autokorrelationsreihe selbst bei Verwendung von MATLAB konjugiert definiert. Daher sind die Ergebnisse von `xcorr()` immer konjugiert.

Der Aufruf `r = xcorr(x, 'option')` ermöglicht es, die Korrelationsreihe zu normieren. Als `'option'` stehen die folgenden Möglichkeiten zur Verfügung:

- biased:
$$
\hat{r}_{XX}(\kappa) = \frac{1}{N} \sum_{k=0}^{N-|\kappa|-1} x(k) \, x^*(k+\kappa)
$$

- unbiased
$$
\hat{r}'_{XX}(\kappa) = \frac{1}{N - |\kappa|} \sum_{k=0}^{N - |\kappa| - 1} x(k) \, x^*(k+\kappa)
$$

- coeff  
normiert die Korrelationsreihe $\hat{r}''_{XX}(\kappa)$ so, dass sie bei $\kappa=0$ den Wert eins hat


**a)**  
  Erzeuge eine weiße gaußsche Rauschsequenz der Länge **N = 8000**.  
  Schätze die Autokorrelationsreihe mit der Funktion `xcorr()`  
  und plote die Ergebnisse.  
  Verwende alle drei verschiedenen `'option'`-Einstellungen.

```python
# BEGIN SOLUTION
n = randn(800,1);
kappa = -799:799;
s_biased   = xcorr(n,'biased').';
s_unbiased = xcorr(n,'unbiased').';
s_coeff    = xcorr(n,'coeff').';

figure; stem(kappa,s_biased);   grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autokorrelationsfolge: Nicht erwartungstreu');
figure; stem(kappa,s_unbiased); grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autokorrelationsfolge: erwartungstreu');
figure; stem(kappa,s_coeff);    grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autokorrelationsfolge: auf time-lag Null normiert');
# END SOLUTION
```

**b)**  
  Die Fourier-Transformation der Autokorrelationsreihe ist das  
  (periodische) Leistungsdichtespektrum.  
  Verwende die Funktion `ldtft()` und plotte die Schätzung  
  des Leistungsdichtespektrums.

```python
# BEGIN SOLUTION
n = randn(200,1);
s = xcorr(n,'biased');
[S W] = ldtft(s,512);

figure; semilogy(W/pi, abs(S) ); % normierte Frequenz
grid; xlabel('Omega/pi'); ylabel('$|S_{xx}(e^{j\Omega})|$', 'Interpreter', 'latex');
title('Leistungsdichtespektrum');
# END SOLUTION
```
