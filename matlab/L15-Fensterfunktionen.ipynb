# Fensterfunktionen

Im Gegensatz zu einem rekursiven Filter, dessen Design auf der Transformation von kontinuierlichen IIR-Systemen in diskrete IIR-Systeme basiert, beruht das Design nicht-rekursiver (FIR-)Systeme auf der Approximation der gewünschten Frequenzantwort des diskreten Systems. Auf Basis einer konstanten Gruppengeschwindigkeit werden meist **linearphasige** Filter angewendet, deren Design im Folgenden behandelt wird.

FIR-Systeme mit linearer Phase basieren auf einer Fensterung mit idealen Tiefpassfiltern. Um die Linearphaseneigenschaft zu erhalten, müssen die Fensterfunktionen symmetrisch sein.

Die gebräuchlichsten Fensterfunktionen sind die sogenannten Cosinus-Fenster, die folgende Form haben:

$$
f(k) =
\begin{cases}
a - b \cdot \cos\!\left( \frac{2\pi k}{m} \right) + c \cdot \cos\!\left( \frac{4\pi k}{m} \right), & 0 \leq k \leq m \\
0, & \text{sonst.}
\end{cases}
$$

| Fenster      | MATLAB-Funktion | a    | b    | c    |
|--------------|-----------------|------|------|------|
| Rechteck     | `boxcar`        | 1.0  | 0.0  | 0.0  |
| von Hann     | `hanning`       | 0.5  | 0.5  | 0.0  |
| Hamming      | `hamming`       | 0.54 | 0.46 | 0.0  |
| Blackman     | `blackman`      | 0.42 | 0.5  | 0.08 |

Ein weiteres Fenster ist das Dreiecks- oder Bartlett-Fenster (`bartlett`).

## Aufgaben

### a)

Berechne die oben genannten Fenstertypen für $m=21$ und $m=42$. Stelle die Signale im Zeitbereich sowie deren zeitdiskrete Fourier-Transformation dar. Vergleiche die verschiedenen Fenstertypen hinsichtlich ihrer Durchlassbanddämpfung, der Dämpfung des ersten Überschwingers und der Sperrbanddämpfung.

```python
# BEGIN SOLUTION
window_length = 21;%f�r andere Fensterl�ngen hier die Variable �ndern

% Rechteckfenster   
tltstr = 'Rechteck-Fenster';
rw = boxcar(window_length);      % Berechnung des Rechteckfenster der L�nge 
                                 % window_length
[Rw W] = ldtft(rw,512);          % Berechnung der DTFT des Rechteckfensters

i = find(abs(Rw)==0);            % Vermeidung von Werten==0
Rw(i) = 1e-100*ones(size(i));      

% # grafische Ausgabe # %
% normierter Betragsfrequenzgang des Rechteckfensters
figure;                        
plot(W/pi,20*log10(abs(Rw/max(Rw))));
axis([-1 1 -80 0]);
grid;
xlabel('Omega/pi');
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB');
title([tltstr,' m: ', num2str(window_length)]); 

% Rechteckfenster
figure; 
stem(0:window_length-1,rw);
grid; 
xlabel('k'); 
ylabel('f(k)'); 
title(tltstr);

% von Hann
tltstr = 'von Hann-Fenster';
rh = hanning(window_length);      % Berechnung des vonHann-Fenster der L�nge
                                  % window_length

[Rh W] = ldtft(rh,512);           % Berechnung der DTFT des vonHann-Fensters

i = find(abs(Rh)==0);             % Vermeidung von Werten==0
Rh(i) = 1e-100*ones(size(i));

% # grafische Ausgabe # %
% normierter Betragsfrequenzgang des vonHann-Fensters
figure;
plot(W/pi,20*log10(abs(Rh/max(Rh))));
axis([-1 1 -80 0]);
grid;
xlabel('Omega/pi');
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB');
title([tltstr,' m: ', num2str(window_length)]);

% vonHann-Fenster
figure; 
stem(0:window_length-1,rh);
grid; xlabel('k'); 
ylabel('f(k)'); 
title(tltstr);

% Hamming
tltstr  = 'Hamming-Fenster';
rhm = hamming(window_length);
[Rhm W] = ldtft(rhm,512);

i = find(abs(Rhm)==0);
Rhm(i) = 1e-100*ones(size(i)); 

% # grafische Ausgabe # %
% normierter Betragsfrequenzgang des Hamming-Fensters
figure;
plot(W/pi,20*log10(abs(Rhm/max(Rhm))));
axis([-1 1 -80 0]);
grid;
xlabel('Omega/pi');
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB');
title([tltstr,' m: ', num2str(window_length)]);

% Hamming-Fenster
figure; 
stem(0:window_length-1,rhm);
grid; xlabel('k'); 
ylabel('f(k)'); 
title(tltstr);

% Blackman
tltstr  = 'Blackman-Fenster';
rhb = blackman(window_length);
[Rhb W] = ldtft(rhb,512);

i = find(abs(Rhb)==0);                  
Rhb(i) = 1e-100*ones(size(i)); 

figure;
plot(W/pi,20*log10(abs(Rhb/max(Rhb))));
axis([-1 1 -80 0]);
grid;
xlabel('Omega/pi');
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB');
title([tltstr,' m: ', num2str(window_length)]);

figure; 
stem(0:window_length-1,rhb);
grid;
xlabel('k');
ylabel('f(k)');
title(tltstr);

% Bartlett
tltstr  = 'Bartlett-Fenster';
rhbt = bartlett(window_length);
[Rhbt W] = ldtft(rhbt,512);

i = find(abs(Rhbt)==0);                 
Rhbt(i) = 1e-100*ones(size(i)); 

figure;
plot(W/pi,20*log10(abs(Rhbt/max(Rhbt))));
axis([-1 1 -80 0]);
grid;
xlabel('Omega/pi');    
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB');
title([tltstr,' m: ', num2str(window_length)]);

figure; 
stem(0:window_length-1,rhbt)
grid; 
xlabel('k'); 
ylabel('f(k)'); 
title(tltstr);
# END SOLUTION
```

### b)

Das Kaiser-Fenster stellt ein flexibles Fenster dar, das wie folgt definiert ist:

$$
f(k) :=
\begin{cases}
\dfrac{I_0\!\left( \beta\sqrt{1-\left(1-\tfrac{2}{m}k\right)^2} \right)}{I_0(\beta)}, & 0 \leq k \leq m \\
0, & \text{sonst.}
\end{cases}
$$

Dabei ist $I_0(x)$ die modifizierte Besselfunktion erster Art, 0. Ordnung. Die MATLAB-Funktion `kaiser(m,beta)` berechnet dieses Fenster. Im Gegensatz zu den Fenstern aus Aufgabe a) hat das Kaiser-Fenster zwei Parameter: die Länge $m+1$ und einen Parameter $\beta$, der die Steigung beeinflusst.

Durch Variation von $m$ und $\beta$ können Länge und Form des Fensters verändert werden, um einen Kompromiss zwischen der Amplitude des ersten Überschwingers und der Breite des Hauptmaximums zu erreichen. Berechne das Kaiser-Fenster für $m=21$ und $\beta=0,3,6$. Berechne die jeweiligen Spektren und stelle sie zusammen mit der Darstellung der Fenster im Zeitbereich dar.

```python
# BEGIN SOLUTION
w0 = kaiser(21,0); [W0 W] = ldtft(w0,512);
w3 = kaiser(21,3);     W3 = ldtft(w3,512);
w6 = kaiser(21,6);     W6 = ldtft(w6,512);
k=0:20;

figure; 
plot(W/pi,20*log10(abs(W0/max(W0)))); 
hold on;
plot(W/pi,20*log10(abs(W3/max(W3))),'--');
plot(W/pi,20*log10(abs(W6/max(W6))),'-.'); 
axis([-1 1 -80 0]);
hold off; 
xlabel('Omega/pi'); 
ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB'); 
title('Kaiser-Fenster: beta: - 0,  -- 3,  -. 6'); 
grid;

figure; 
plot(k,w0); 
hold on; 
plot(k,w3,'--');
plot(k,w6,'-.'); 
axis([0 20 0 1.2]); 
hold off;
xlabel('k');
ylabel('f(k)');
title('Kaiser-Fenster: beta: - 0,  -- 3,  -. 6      '); 
grid;
# END SOLUTION
```

Der Zeitverlauf jeder Fensterfunktion verdeutlicht die Mittensymmetrie, wodurch eine lineare Phase garantiert ist.  
Bei der Auswertung der Fourier-Transformation der Fensterfunktionen, um das Filterverhalten zu analysieren, zeigt sich:  
Mit steigendem Filtergrad $m$ wird das Spektrum proportional gestaucht. Daher reduziert sich das Passband (der Bereich bis zum ersten Nullpunkt des Spektrums) $\Omega < \Omega_s$ entsprechend bei einer Erhöhung von $m=21$ auf $m=42$.
 
Das Stopbandverhalten ($\Omega \rightarrow \pi$) zeigt, dass ein höherer Filtergrad vorteilhaft ist.  
Berücksichtigt man die Dämpfung des ersten Überschwingers, so zeigt sich eine dämpfungsunabhängige Größe von $m$, z. B. $a_N \approx 57 \,\text{dB}$ beim Blackman-Fenster.  

Ein Vergleich der Fenster zeigt einen Kompromiss zwischen dem erreichbaren Passband und der Dämpfung des ersten Überschwingers.  
Während das Rechteckfenster das kleinste Passband von $2\pi/(m+1)$ und die geringste Dämpfung von $a_N \approx 13.3 \,\text{dB}$ hat, besitzt das Blackman-Fenster das breiteste Passband von $6\pi/(m+1)$, aber gleichzeitig auch die höchste Dämpfung des ersten Überschwingers.
