# Aufgabe: Eigenschaften der DFT

Betrachte die Serie

$$
x(k) = 
\begin{cases}
k + j \cdot (7-k), & k = 0 \dots 15\\
0, & \text{sonst}
\end{cases}.
$$

**a)** Berechne die DFT $X(n)$ mit der Funktion `fft(x)`.  
      Stelle den Realteil $\Re X(n)$ und den Imaginärteil $\Im X(n)$ von $X(n)$ dar.

**b)** Stelle die DFT von $\Re x(k)$ und $j\,\Im x(k)$ dar  
      (ebenfalls Real- und Imaginärteil).  
      Welche Symmetrien zeigen die DFTs von $\Re x(k)$ und $j\,\Im x(k)$?

**c)** Erzeuge eine zyklische Verschiebung von $x(k)$ um
      $\lambda=4$ bzw. $\lambda=8$ Abtastwerte:  

$$
x_1(k) = x((k+\lambda) \bmod M)
$$

Welche Beziehung besteht zwischen $X_1(n) = \mathrm{DFT} x_1(k)$ und $X(n)$?

**d)** Berechne die diskrete Fourier-Transformation $X_2(n)$ von  

$$
x_2(k) = x(k) \cdot \exp\Big(-j \, 2\pi \, 3 k / 16\Big)
$$

Welche Beziehung besteht zwischen $X_2(n)$ und $X(n)$?

```python
# BEGIN SOLUTION
% #####  Teilaufgabe a:  DFT von x(k) berechnen  #####
kk = 0:15;
x  = kk + j*(7-kk);
xr = real(x);  
xi = imag(x);

X  = fft(x); 
Xr = real(X);  
Xi = imag(X);

% # grafische Ausgabe # %
% Realteil
figure;
stem(kk,Xr);
title('Re(X(n))');
xlabel('n'); 
grid;

% Imagin�rteil
figure;
stem(kk,Xi);
title('Im(X(n))');
xlabel('n');
grid;

input('....Fuer naechste Teilaufgabe: RETURN druecken');

% #####  Teilaufgabe b:  DFT von Re{x(k)} und Im{x(k)}  #####
% # grafische Ausgabe # %
figure;stem(kk,real(fft(xr)));  title('Re(FFT(Re(x(k))))');  xlabel('n');grid;
figure;stem(kk,imag(fft(xr)));  title('Im(FFT(Re(x(k))))');  xlabel('n');grid;
figure;stem(kk,real(fft(j*xi)));title('Re(FFT(j*Im(x(k))))');xlabel('n');grid;
figure;stem(kk,imag(fft(j*xi)));title('Im(FFT(j*Im(x(k))))');xlabel('n');grid;

input('....Fuer naechste Teilaufgabe: RETURN druecken');

% #####  Teilaufgabe c:  Zyklische Verschiebung  #####
xtmp = [x x];
x1_04 = xtmp(kk+5); 
X1_04 = fft(x1_04);
x1_08 = xtmp(kk+9); 
X1_08 = fft(x1_08);

figure;stem(kk,Xr);         title('Re(X(n))');           xlabel('n');grid;
figure;stem(kk,Xi);         title('Im(X(n))');           xlabel('n');grid;
figure;stem(kk,real(X1_04));title('Re(X1(n)), lambda=4');xlabel('n');grid;
figure;stem(kk,imag(X1_04));title('Im(X1(n)), lambda=4');xlabel('n');grid;
figure;stem(kk,real(X1_08));title('Re(X1(n)), lambda=8');xlabel('n'); grid;
figure;stem(kk,imag(X1_08));title('Im(X1(n)), lambda=8');xlabel('n'); grid;

input('....Fuer naechste Teilaufgabe: RETURN druecken');

% #####  Teilaufgabe d:  Multiplikation mit exp(j*...)  #####
x2 = x.*exp(-j*2*pi*3*kk/16);
X2 = fft(x2);

figure;stem(kk,Xr);      title('Re(X(n))'); xlabel('n');grid;
figure;stem(kk,Xi);      title('Im(X(n))'); xlabel('n');grid;
figure;stem(kk,real(X2));title('Re(X2(n))');xlabel('n');grid;
figure;stem(kk,imag(X2));title('Im(X2(n))');xlabel('n');grid;
# END SOLUTION
```

## Teil (b)

Die DFT eines reellen (imaginären) Signals ist konjugiert gerade (ungerade), das heißt der Realteil ist gerade (ungerade) bezüglich der Frequenz $0$ und der Imaginärteil ist ungerade (gerade). Um die Symmetrieeigenschaften zu verstehen, muss berücksichtigt werden, dass die Spektren in beiden Richtungen (positiv und negativ) kontinuierlich periodisch sind. Laut Tabelle 6.1 entspricht eine Verschiebung im Zeitbereich um $\lambda$ Abtastwerte einer Multiplikation von $X(n)$ mit $\exp(j\,2\pi n \lambda / N)$ im Frequenzbereich. Für $N=16$ und $\lambda =4$ bzw. $\lambda =8$ erhalten wir:

## Teil (d)

Laut Tabelle 6.1 gilt:

$$
\text{DFT}\{ \exp(-j\,2\pi i k/N) x(k) \} = X((n+i))_N
$$

Die Notation $((i))_N$ beschreibt eine Modulo-$N$ Indizierung. Die Multiplikation von $x(k)$ mit $\exp(-j\,2\pi\,3k/16)$ führt zu einer Frequenzverschiebung um $3 f_A/16$, d.h.

$$
X_2(n) = X((n+3))_{16}
