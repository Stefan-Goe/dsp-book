# Aufgabe: Interpolation durch Zero Padding

**Aufgabe:**

Berechne die DFT von  

$$
x(k) = 1 - \cos\left(\frac{2\pi k}{64}\right), \quad k = 0, \dots, 63
$$

Gibt es gemeinsame Eigenschaften mit der diskreten Fourier-Transformation des Signals

$$
x_0(k) =
\begin{cases}
x(k), & k = 0 \dots 63 \\
0,    & k = 64 \dots 127
\end{cases}
$$

---

### MATLAB-Hinweise

```matlab
N = 64;
kk = 0:N-1;
x = 1 - cos(2*pi*kk/N);   % definiert x(k)
x0 = [x, zeros(1,N)];     % hängt einen Nullvektor (1xN) an

````python
# BEGIN SOLUTION
N  = 64;
x  = 1-cos(2*pi/N*(0:N-1));    
X  = fft(x);

x0 = [x zeros(1,N)];           
X0 = fft(x0);

% # grafische Ausgabe # %
figure; 
stem(0:N-1,abs(X));
axis([0 N-1 0 max(X)]);
title('abs(X(n))');
xlabel('n');

figure; 
stem(0:2*N-1,abs(X0)); 
axis([0 2*N-1 0 max(X)]);
title('abs(X0(n))');
xlabel('n'); 
# END SOLUTION
````

## Lösung: Interpolation durch Zero Padding

Durch *Zero Padding* (Auffüllen eines Vektors mit Nullen) kann die spektrale Auflösung der DFT verbessert werden.  
Da die Folge $x(k)$ in dieser Aufgabe durch Zero Padding die doppelte Länge erhält, verdoppelt sich auch die spektrale Auflösung:  
Jeder zweite Wert von $X_0(n)$ stimmt mit dem entsprechenden Wert von $X(n)$ überein:

$$
X_0(2n) \equiv X(n), \quad n = 0, 1, 2, \dots, 63
$$

---

Aus der rechten Abbildung erkennen wir Folgendes: Wenn die Anzahl der hinzugefügten Nullen gegen unendlich geht, wird nicht das Spektrum des periodischen kontinuierlichen Signals  

$$
\tilde{x}(t) = 1 - \cos\left(\frac{2\pi t}{T}\right), \quad \forall t
$$

approximiert, sondern vielmehr das Spektrum des nichtperiodischen Signals

$$
x(t) =
\begin{cases}
1 - \cos\left(\frac{2\pi t}{T}\right), & 0 \leq t < T \\
0, & \text{sonst}
\end{cases}
$$
