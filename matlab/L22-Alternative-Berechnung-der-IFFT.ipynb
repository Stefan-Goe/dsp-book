# Aufgabe: Alternative Berechnung der IFFT

Führe die folgenden Schritte aus:

1. Berechne die FFT der zeitdiskreten Folge $x_0(k)$ wie in Aufgabe **dsv6a02e**.  
   (Zur Erinnerung: $x_0(k)$ ist die Folge der Länge 128, die für $k=0,\dots,63$ gleich $x(k)$ ist und für $k=64,\dots,127$ gleich null.)

2. Vertausche Real- und Imaginärteil des transformierten Signals (d.h. ersetze jeden komplexen Wert $X(n)=\Re\{X(n)\}+j\Im\{X(n)\}$ durch $\Re\{X(n)\}^\prime + j\Im\{X(n)\}^\prime$, wobei $\Re\{X(n)\}^\prime=\Im\{X(n)\}$ und $\Im\{X(n)\}^\prime=\Re\{X(n)\}$ gilt).

3. Transformiere dieses modifizierte Spektrum erneut mit der FFT.

4. Vertausche Real- und Imaginärteile der resultierenden Zeitbereichsfolge.

Warum ist die resultierende Folge – bis auf einen Skalierungsfaktor – gleich der ursprünglichen Folge $x_0(k)$?

```python
# BEGIN SOLUTION
N  = 64;
x  = 1 - cos(2*pi/N*(0:N-1));
x0 = [x zeros(1,N)];

X0 = fft(x0);                % FFT
X1 = imag(X0) + j*real(X0);  % Re() und Im() vertauschen
x1 = fft(X1);                % Nochmalige FFT
x2 = imag(x1) + j*real(x1);  % Re() und Im() vertauschen

% # grafische Ausgabe # %
figure; 
bar(0:2*N-.5,x0); 
title('x0(k)'); 
xlabel('k'); 
axis([-.5 2*N-1 0 2.1]);

figure; 
bar(0:2*N-.5,x2); 
title('x2(k)'); 
xlabel('k'); 
axis([-.5 2*N-1 0 269]);
# END SOLUTION
```

## Lösung: Alternative Berechnung der IFFT

Die *m-Datei* für die Aufgabe *Alternative Berechnung der IFFT* implementiert, entsprechend der Beschreibung in Absatz 6.5,  
eine IFFT durch zweimalige Anwendung der FFT. Wie in den Abbildungen zu sehen ist, entsprechen die Ergebnisse der ursprünglichen Datenfolge — bis auf den Faktor $N$.

Wenn $x_2(k)$ identisch zu $x_0(k)$ ist, muss das Signal $x_2(k)$ im letzten Schritt durch die FFT-Länge $N = 128$ normalisiert werden,  
aufgrund der Relation

$$
\text{IDFT}\{X(n)\} = \frac{1}{N}\, j \, \left( \text{DFT}\{ j \, X^*(n) \} \right)^*
$$
