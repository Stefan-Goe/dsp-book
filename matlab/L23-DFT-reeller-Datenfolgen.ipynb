# Aufgabe: DFT von reellen Folgen

Bei der Transformation einer reellen Folge mittels diskreter Fourier-Transformation (DFT) enthält das Ergebnis Redundanz: Es ist konjugiert symmetrisch (conjugate-even).

Diese Eigenschaft kann genutzt werden, um eine reelle Folge $x_1(k)$ der Länge $2N$ mit einer $N$-Punkt-FFT zu transformieren.  

**Aufgabe:**

- Schreibe eine *MATLAB-Funktion*, die diese Eigenschaft ausnutzt.
- Schreibe ein kleines Skript, um die Korrektheit dieses Ansatzes zu überprüfen.

```python
# BEGIN SOLUTION
N = 2^10;
x = randn(1, N);
X1 = fft(x);
X2 = lfft(x);
difference = mean(abs(X1 - X2).^2)
# END SOLUTION
```
