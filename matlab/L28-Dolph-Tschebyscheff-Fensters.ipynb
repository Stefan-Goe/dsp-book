### Lösung: Dolph-Chebyshev-Fenster

Die MATLAB-Funktion `lcheby(N, Ω_s, ΔΩ)` berechnet die **Spektralfunktion**

$$
F(e^{jn\Delta\Omega}),
$$

die **kausale Zeitreihe** $f(k)$ sowie die erreichte **Sperrbanddämpfung** $a_\mathrm{min}$ des Dolph-Chebyshev-Fensters mit den Parametern $N$ und $\Omega_s$.

Sie stellt die **normierte Fensterfunktion** $f(k)/f_\mathrm{max}$ und den **Betrag der normierten Spektralfunktion**

$$
\left| \frac{F(e^{jn\Delta\Omega})}{F(e^{j0})} \right|
$$

in dB dar.

Aus den folgenden Abbildungen erkennt man die **gleichmäßige Annäherung im Sperrband**:

* Die Wahl $N=32$ und $\Omega_s/(2\pi) = f_s T = 1/20 = 0.05$ führt zu einer minimalen Sperrbanddämpfung von $a_\mathrm{min} = 36,5~\mathrm{dB}$.

**Warum ist in MATLAB keine abschnittsweise Definition der Spektralfunktion notwendig?**

Da

$$
\cosh(x) = 0,5 \left( e^x + e^{-x} \right), \quad \cos(x) = 0,5 \left( e^{jx} + e^{-jx} \right),
$$

gilt:

$$
\cosh(x) = \cos(-jx).
$$

Mit

$$
y(\Omega) = \frac{\cos(\Omega/2)}{\cos(\Omega_s/2)}
$$

im **Durchlassband** des nicht-kausalen Dolph-Chebyshev-Fensters gilt:

\begin{align*}
F_0^{DT}(e^{j\Omega}) &= \cosh\big( N \cdot \operatorname{arcosh}(y(\Omega)) \big) \
&= \cos\Big( (-j) \cdot N \cdot \operatorname{arcosh}(y(\Omega)) \Big), \quad 0 \le \Omega \le \Omega_s
\end{align*}

Im **Sperrband**:

$$
F_0^{DT}(e^{j\Omega}) = \cos\left( N \cdot \arccos(y(\Omega)) \right), \quad \Omega_s \le \pi
$$

Es bleibt zu zeigen:

\begin{align*}
(-j) \cdot \operatorname{arcosh}(y(\Omega)) &\stackrel{!}{=} \arccos(y(\Omega)) \
\cos(-j \cdot \operatorname{arcosh}(y(\Omega))) &\stackrel{!}{=} y(\Omega)
\end{align*}

Aufgrund der obigen Beziehung – unter Verwendung der **komplexwertigen Definition hyperbolischer Funktionen** – verschmelzen beide Fälle.

Da MATLAB-Funktionen auf **komplexwertigen Definitionen** basieren, ist eine **abschnittsweise Definition** der Spektralfunktion des Dolph-Chebyshev-Fensters nicht notwendig.
