### Aufgabe — Effekt der Hamming-Fensterung auf die FFT

Stellen Sie die zweimal wiederholte Fortsetzung der Reihen dar  
$$
x_1(k) = \sin\left(\frac{k\pi}{8}\right)
\quad \text{und} \quad
x_2(k) = \sin\left(\frac{k\pi}{10}\right)
\quad \text{für} \quad k = 0 \dots 63,
$$
sowie den Betrag ihrer 64-Punkte-FFT  
$$
X_i(n) = \text{FFT}\{x_i(k)\}.
$$

Multiplizieren Sie die Datenreihen  
$$
x_i^{Hm}(k) := x_i(k) \cdot f^{Hm}(k)
$$
mit einem 64-Punkte-**Hamming-Fenster** $f^{Hm}(k)$,  
und stellen Sie erneut die zweimal wiederholte Fortsetzung von $x_i^{Hm}(k)$ dar,  
sowie den Betrag ihrer 64-Punkte-FFT  
$$
X_i^{Hm}(n).
$$

Welche Effekte hat die Hamming-Fensterung auf die Beträge  
der FFT-Koeffizienten?

---

**MATLAB-Hinweis:**  
Die Berechnung und Darstellung der Fensterungsergebnisse kann z.B. mit  
```matlab
stem(0:63, abs(fft(xi .* hamming(64).')))

````python
# BEGIN SOLUTION
N    = 64;
kk   = (0:N-1);            kk2  = (0:2*N-1);
x1   = sin(kk*pi/8);       X1   = fft(x1);
x1Hm = x1.*hamming(N).';   X1Hm = fft(x1Hm);
x2   = sin(kk*pi/10);      X2   = fft(x2);
x2Hm = x2.*hamming(N).';   X2Hm = fft(x2Hm);

% # grafische Ausgabe # %
figure; 
stem(kk2, [x1 x1]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k');
ylabel('x1((k))');
title('x1(k)=sin(k*pi/8), k=0..63, zweifach periodisch fortgesetzt'); 

figure; 
stem(kk, abs(X1));  
axis([0 N 0 1.1*N/2]);  
xlabel('n');
ylabel('|X1(n)|');
title('Betragsspektrum von x1(k)');

figure; 
stem(kk2, [x1Hm x1Hm]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k');
ylabel('x1_Hm((k))');
title('Hamming-gefenstertes x1(k), zweifach periodisch fortgesetzt');

figure; 
stem(kk, abs(X1Hm));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X1_Hm(n)|');
title('Betragsspektrum bei Hamming-Fensterung'); 

input('......fuer 2.Datenfolge: RETURN druecken');

figure; 
stem(kk2, [x2 x2]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k'); 
ylabel('x2((k))');
title('x2(k)=sin(k*pi/10), k=0..63, zweifach periodisch fortgesetzt'); 

figure; 
stem(kk, abs(X2));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X2(n)|');
title('Betragsspektrum von x2(k)'); 

figure; 
stem(kk2, [x2Hm x2Hm]);  
axis([0 2*N -1.1 1.1]);
xlabel('k');
ylabel('x2_Hm((k))');
title('Hamming-gefenstertes x2(k), zweifach periodisch fortgesetzt'); 

figure; 
stem(kk, abs(X2Hm));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X2_Hm(n)|');
title('Betragsspektrum bei Hamming-Fensterung');
# END SOLUTION
````

### Lösung — Reduzierung von Leakage durch Fensterung

**Reihe** $x_1(k)$ **ohne Fensterung:**  
Wenn die DFT-Länge $N=64$ gewählt wird, fallen genau $m=4$ Perioden des Sinussignals  
in das DFT-Fenster der Länge $N T_A$. Die dargestellte periodische Wiederholung der Reihe $x_1(k)$ nach $N$ Werten ergibt eine kontinuierliche Darstellung ohne Sprünge.  
Der Leakage-Effekt tritt nicht auf, d.h. die Spektralfunktion $X_1(n)$ reproduziert korrekt das Spektrum der kontinuierlichen Sinusschwingung.

---

**Reihe** $x_1(k)$ **mit Fensterung:**  
Bei gleicher DFT-Länge wird die Reihe mit einem Hamming-Fenster gewichtet, d.h. im Zeitbereich multipliziert mit

$$
f^{Hm}(k) =
\begin{cases}
0.54 - 0.46 \cdot \cos\left(\frac{2\pi k}{N-1}\right), & 0 \le k \le N-1 \\
0, & \text{sonst}.
\end{cases}
$$

Dies entspricht in der Spektraldomäne einer Faltung mit drei verschobenen Spektralfunktionen eines Rechteckfensters, wobei die mittlere mit dem Faktor $0.54$ bewertet wird.  
Als Folge haben die Hauptspektrallinien kleinere Werte.  
Aufgrund der reduzierten spektralen Auflösung treten zwei zusätzliche Linien auf.

**Reihe** $x_2(k)$ **ohne Fensterung:**  
Für $N=64$ fallen $m = N/20 = 3.2$ Perioden des Sinussignals in das DFT-Fenster.  
Die dargestellte periodische Wiederholung von $x_2(k)$ nach $N$ Werten zeigt eine unstetige Darstellung mit Sprüngen.  
Da die DFT eine periodische Fortsetzung der Reihe außerhalb des betrachteten DFT-Fensters annimmt, tritt aufgrund der Sprünge der **Leakage-Effekt** auf: Wie in der Abbildung zu sehen, "leaken" die Hauptspektrallinien bei $n=3$ und $n=61$ durch alle DFT-Bins.

**Reihe** $x_2(k)$ **mit Fensterung:**  
Wird ein Hamming-Fenster angewendet, werden die Sprünge geglättet.  
Dies führt zu einer Reduzierung des Leakage-Effekts auf Kosten der spektralen Auflösung.
