### Übung: Transformation reeller Bandpass-Signale

Basierend auf dem verallgemeinerten Abtasttheorem kann die Mittenfrequenz des Bandpass-Signals durch Downsampling beeinflusst werden.

Die Funktion  

$$
lband(N, \tfrac{f_m}{f_A}, \tfrac{b}{f_A})
$$  

erzeugt ein diskretes Bandpass-Signal mit der Mittenfrequenz $f_m$ und der Bandbreite $b$.

Vergleiche für $N = 1000$, $f_m / f_A = 0.2$ und $b / f_A = 0.02$  
die Betragsverläufe der Spektren des Originalsignals und der
um die Faktoren 9, 10, 11 und 12 heruntergetasteten Signale.

---

*Matlab-Hinweise:*  

```matlab
b1 = lband(1000, 0.2, 0.02);   % Bandpass-Signal
b2 = b1(1:10:1000);            % 10-fach heruntergetastetes Bandpass-Signal
plot(abs(fft(b2 .* hanning(length(b2)))))

````python
# BEGIN SOLUTION
NFFT = 2^10;
b1 = lband(1000,0.2,0.02);  
fT = 0:1/NFFT:1-1/NFFT;

% Originalsignal
b2 = b1; 
lb = length(b2); 
aB2_hn = abs(fft(b2.*hanning(lb),NFFT));

% # grafische Ausgabe # %
figure; 
plot(fT, aB2_hn);
axis([0 1 0 1.1*max(aB2_hn)]); 
xlabel('Omega/2pi = f T');
ylabel('|B(exp(j*Omega))|');

for k=[9 10 11 12]                     %Unterabtastung um Faktor k
    b2 = b1(1:k:1000); 
    lb = length(b2); 
    aB2_hn = abs(fft(b2.*hanning(lb),NFFT));
    
    % # grafische Ausgabe # %
    figure; 
    plot(fT,aB2_hn); 
    axis([0 1 0 1.1*max(aB2_hn)]); 
    xlabel('Omega/2pi = f T');
    ylabel('|B_M(exp(j*Omega))|'); 
    title(sprintf('Unterabtastung um Faktor M=%d',k));
end;
# END SOLUTION
````

### Lösung: Spektraltransformation reeller Bandpass-Signale

Die Funktion  

$$
b = lband\!\left(N, \tfrac{f_m}{f_A}, \tfrac{b}{f_A}\right)
$$  

liefert das zeitdiskrete reelle Bandpass-Signal

$$
b(k) = \frac{1}{11}\sum_{\nu=-5}^{5} \left(1+\frac{\nu}{10}\right)\cos\!\left(2\pi\left[f_m + \nu\frac{b}{10}\right]\frac{k}{f_A}\right), 
\quad k = 0,\ldots,N-1
$$

mit der Bandbreite $b$ und Mittenfrequenz $f_m$.  
Da $b(k)$ reell und gerade ist, ist das zugehörige ($2\pi$-periodische) Spektrum $B(e^{j\Omega})=\mathcal{F}\{b(k)\}$ konjugiert gerade.  
Damit ist auch das Betragsspektrum $|B(e^{j\Omega})|$ des Originalsignals gerade.

Für den gegebenen Bandpass-Prozess gilt $f_m=f_A/5$ und $b=f_A/50$,  
d.h. die Mittenfrequenz ist **10-mal so hoch wie die Bandbreite**:  
$f_m = 10\,b$.  

Unter Berücksichtigung der $2\pi$-Periodizität ergibt sich das Betragsspektrum $|B(e^{j\Omega})|$ mit den Mittenfrequenzen

- $f_m = f_A/5 \;\Rightarrow\; \Omega_m/2\pi = 1/5$  
- $f_m' = -f_A/5 = -f_A/5 + f_A = 4f_A/5 \;\Rightarrow\; \Omega_m'/2\pi = 4/5$  

---

Wenn das Bandpass-Signal $b(k)$ nun um den Faktor $M$ abwärtsgetastet wird  
(downsampling $\;\Rightarrow b_M(k)$), so erhält man das Spektrum

$$
|B_M(e^{j\Omega})| = \left|\mathcal{F}\{b_M(k)\}\right|
$$

indem  

1. die Frequenzachse des Originalspektrums $|B(e^{j\Omega})|$ mit $M$ gestreckt wird (Mittenfrequenzen und Bandbreiten werden mit $M$ multipliziert),  
2. das resultierende Spektrum $2\pi$-periodisch fortgesetzt wird,  
3. nur der Bereich $0 \leq \Omega \leq 2\pi$ dargestellt wird.  

Die Mittenfrequenzen $\Omega_m$ und $\Omega_m'$ sind also mit $M$ zu multiplizieren und dann durch eine Modulo-$2\pi$-Operation in den interessierenden Frequenzbereich zu verschieben.  

---

### Tabelle der Mittenfrequenzen

| $M$ | 9 | 10 | 11 | 12 |
|-----|---|----|----|----|
| $\Omega_{m,M}=M\,\Omega_m$ | $\tfrac{9}{5}\cdot 2\pi$ | $\tfrac{10}{5}\cdot 2\pi$ | $\tfrac{11}{5}\cdot 2\pi$ | $\tfrac{12}{5}\cdot 2\pi$ |
| $\Omega_{m,M} \bmod 2\pi$ | $\tfrac{4}{5}\cdot 2\pi$ | $0\cdot 2\pi$ | $\tfrac{1}{5}\cdot 2\pi$ | $\tfrac{2}{5}\cdot 2\pi$ |
| $\Omega'_{m,M}=M\,\Omega'_m$ | $\tfrac{36}{5}\cdot 2\pi$ | $\tfrac{40}{5}\cdot 2\pi$ | $\tfrac{44}{5}\cdot 2\pi$ | $\tfrac{48}{5}\cdot 2\pi$ |
| $\Omega'_{m,M} \bmod 2\pi$ | $\tfrac{1}{5}\cdot 2\pi$ | $0\cdot 2\pi$ | $\tfrac{4}{5}\cdot 2\pi$ | $\tfrac{3}{5}\cdot 2\pi$ |
| Bemerkung | Kehrlage, kein Aliasing | Original, Aliasing | Original, kein Aliasing | Kehrlage, Aliasing |
