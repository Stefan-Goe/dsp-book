## Function Generator (Chapter 2.1)

In this exercise a *function generator* shall be programmed which generates a signal of frequency $f_0$  at a sampling frequency of $f_\mathrm{A}$. This generator shall be able to generate a sinusoidal and a sawtooth signal. For a time continuous system it is common to calculate using the  angular frequency $\omega_0 = 2 \pi f_0$.

Thus a time continuous sine signal is given by
$$
x_\mathrm{sin,K}(t)=\sin \left( \omega_0 \cdot t \right)
$$

Within a time discrete system the sinusoidal signal is sampled with periods of the length $T = 1/f_\mathrm{A}$, such that the time $t=T \cdot k$ can be written by the indices $k =0, 1 , 2,...$. Therefore we obtain
$$
x_\mathrm{sin}(k)= \left.x_\mathrm{sin,K}(t)\right|_{t=kT} = \sin \left( \omega_0 \cdot T \cdot k \right)
$$

For time discrete systems it makes sense to introduce the normalized angular frequency $\Omega_0$, which is without any unit.
$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T \\
         & =  & 2 \pi f_0 \cdot T \\
         & =  & 2 \pi f_0 / f_\mathrm{A}
\end{array}
$$

This definition will be used throughout the further exercises, since it allows handling of digital systems without any limitation of the sampling frequency. Thus,
$$
x_\mathrm{sin}(k) = \sin \left( \Omega_0 \cdot k \right)
$$

### a)

Create a vector `x` with 60 samples of a sinusoidal signal with the frequency $f_0=400\,\text{Hz}$ at a sampling frequency of $8\,\text{kHz}$. In order to do so initialize a coefficient vector `k=[0, 1, ..., 59]`, which represents the time index of the sampled signal $x(k)$. Make use of MATLAB’s `sin()` function. Display the signal with the help of the function `stem()`. What is the period of the sampled signal?

```python
close all;      % Close all figures
clear all;      % Clear all variables

fa = 8000;      % Sampling frequency


% #####  Part a:  Sine Signal  #####
% # Calculation of the signal vector # %

# BEGIN SOLUTION
k = 0:59;               % Create a vector of length 60 with entries 0 to 59
f0 = 400;               % Signal frequency
Omega0 = 2*pi*f0/fa;    % Frequency normalized to sampling rate
xsin = sin(Omega0*k);   % Sine signal vector

% # Graphical output # %
figure;                 % Open a new figure
stem(k, xsin);          % Plot the discrete signal xsin; k is the vector
                        % with values on the x-axis
xlabel('k');            % Label the x-axis
ylabel('x_{sin}(k)');   % Label the y-axis
# END SOLUTION
```

### b)

Change the frequency of the sampled sinusoidal signal to $f_0=960\,\text{Hz}$ and display the resulting signal again by using `stem()`. What is the period of the sampled signal now?

```python
% #####  Part b:  Sine Signal, Modified Period  #####

# BEGIN SOLUTION
% # Calculation of the signal vector # %
f0 = 960;               % Signal frequency
Omega0 = 2*pi*f0/fa;    % Normalized frequency
xsin = sin(Omega0*k);   % Sine signal vector

% # Graphical output # %
figure;         
stem(k, xsin);   
xlabel('k');
ylabel('x_{sin}(k)');
# END SOLUTION
```

### c)

Now create a sawtooth signal with frequency $f_0=960\,\text{Hz}$ at a sampling frequency of $f_\mathrm{A}=8\,\text{kHz}$.

The sawtooth signal shall have an amplitude range of $0...1$.  For calculation use the MATLAB function `mod()`.  

The normalization of the signal frequency $f_0$ by the sampling frequency $f_\mathrm{A}$ can be done in an analogue way to the one above. Take into consideration that by doing so you set the period of the `mod()` yourself — in contrast to the $2\pi$-periodicity of the `sin()` function.

Display the sawtooth using `stem()` and compare the result with the sinusoidal signal. Is the signal frequency correct?

```python
% #####  Part c:  Sawtooth Signal  #####

# BEGIN SOLUTION
xsaw = mod(f0/fa*k, 1);    % Sawtooth signal vector

% # Graphical output # %
figure; 
stem(k, xsaw); 
xlabel('k'); 
ylabel('x_{saw}(k)');
# END SOLUTION
```

### d)

Create a vector of a sinusoidal as well as a sawtooth signal with  
$f_0=1\,\text{kHz}$ at a sampling frequency of $f_\mathrm{A}=8\,\text{kHz}$.

The signals shall be played through a soundcard and have a duration of $d=2\,\mathrm{s}$.  
How many samples are necessary for this?

Play the signals using the MATLAB command `sound(x, fa)`.

```python
% #####  Part d:  De-normalized Signals, Sound Generation  #####

# BEGIN SOLUTION
f0 = 1000;              % Signal frequency
d = 2;                  % Duration in seconds
w = d * fa;             % Number of samples
k = 0:w;                % Time index coefficient vector
xsin = sin(2*pi*f0/fa*k); % Sine signal vector
xsaw = mod(f0/fa*k, 1); % Sawtooth signal vector

% # Playback of the signals # % 
disp('Playing: 2s Sine Signal');     sound(xsin, fa);
pause(length(xsin) / fa)
disp('Playing: 2s Sawtooth Signal'); sound(xsaw, fa);
pause(length(xsaw) / fa)
# END SOLUTION
```

## Function Generator (Chapter 2.1)

**a)**  
The sinus signal has the following curve:

<img src="./pngs/ldsv2a20_a1.png" width="250">

The period of the continuous signal $x_\text{sin,K}(t)=\sin(2\pi f_0 \cdot t)$ is $T_0 = 1 / f_0$.  
Since the number of samples per period of the time continuous signal is an integer, the discrete signal has to have a period of
$$
\begin{array}{rcl}
k_p &=& T_0 / T \\
    &=& f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 400\,\text{Hz} \\
    &=& 20
\end{array}
$$

values.

**b)**  
After raising the signal frequency to $f_0 = 960\,\text{Hz}$  
the period of the time discrete signal and the period of the time continuous signal do not match.

<img src="./pngs/ldsv2a20_b1.png" width="250">

The period of the time discrete signal appears after

$$
\begin{array}{rcl}
k_p &=& L \cdot T_0 / T \\
    &=& L \cdot f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 960\,\text{Hz} \\
    &=& L \cdot 8.\overline{3} \\
    &\Downarrow& L = 3 \\
    &=& 25
\end{array}
$$

samples.  
For this, $L=3$ is the number of periods of the continuous signal, after which the periodicity of the sampled signal settles.

**c)**  
Similar to the sinusoidal signal, the normalization by the sampling frequency can be done in the same way:  

$$
x_\text{saw}(k) = \mathrm{mod}\!\left( \frac{f_0}{f_A} \cdot k , 1 \right)
$$

The plot shows the sawtooth signal, whose period  
(similar to the sinusoidal signal of b) contains $k_p = 25$ samples.

<img src="./pngs/ldsv2a20_c1.png" width="250">

**d)**  
For a duration of $d = 2\,\text{s}$,  

$$
w = \frac{d}{T} = d \cdot f_A = 16000

samples are necessary.
