## Exercise: The Equation of Euler

The equation of Euler states the connection between the exponential function (series)
and the trigonometric functions (series).

$$
e^{j\phi k} = \cos(\phi k) + j \sin(\phi k)
$$

resp.

$$
\cos(\phi k) = \frac{1}{2}\left(e^{j\phi k} + e^{-j\phi k}\right)
$$

$$
\sin(\phi k) = \frac{1}{2j}\left(e^{j\phi k} - e^{-j\phi k}\right)
$$

This connection shall be analyzed in this exercise.

- **a)** Plot a sinusoidal signal from 0 until $6\pi$ with 32 samples per period.

- **b)** Plot the associated complex exponential series on the complex plane (imaginary part = Y-axis, real part = X-axis), as well as its real, imaginary and absolute value with respect to the time index $k$.

- **c)** Use Euler’s equation and create a sinusoidal series using complex exponential sequences. Compare the results with the sinusoidal series of exercise a).

```python
# BEGIN SOLUTION
% #####  Part a:  Sinusoidal Signal  #####
k=0:95;
x=sin(2*pi/32*k);

% # graphical output # %
figure;
stem(k/32,x);           % Display of the samples
hold on; 
plot(k/32,x,'y');       % Display of the continuous sinusoidal signal
plot([0 3], [0 0],'k'); 
hold off;
title('Sinusoidal Function x(k)'); 
xlabel('k/32'); 
ylabel('x(k)');

% #####  Part b:  Real and Imaginary Parts  #####
y=exp(j*2*pi/32*k);

% # graphical output # %
% Complex Plane
figure; 
plot(real(y),imag(y),'bo-');
hold on; 
plot([-1 1], [0 0],'k'); 
plot([0 0 ],[-1 1],'k'); 
hold off; 
axis square
title('Complex Exponential Sequence y(k)'); 
xlabel('Real part'); 
ylabel('Imaginary part'); 

% Real part
figure; 
plot(k/32,real(y));
hold on; 
plot([0 3], [0 0],'k'); 
hold off;
xlabel('k/32'); 
ylabel('Re\{y(k)\}');

% Imaginary part
figure; 
plot(k/32,imag(y)); 
hold on; 
plot([0 3], [0 0],'k'); 
hold off;
xlabel('k/32'); 
ylabel('Im\{y(k)\}');

% Magnitude
figure; 
plot(k/32,abs(y));  
hold on; 
plot([0 3], [0 0],'k'); 
hold off;
axis([-0.05 3 0 1.1]); 
xlabel('k/32'); 
ylabel('|y(k)|');


% #####  Part c:  Euler's Relationship => Sinusoidal #####
z = -j*0.5 *(exp(j*2*pi/32*k) - exp(-j*2*pi/32*k));

% # graphical output # %
figure
stem(k/32,z,'y');
hold on; 
plot(k/32,x,'b-'); 
plot([0 3], [0 0],'k'); 
hold off;
title('Sinusoidal Function by Euler'); 
xlabel('k/32'); 
ylabel('Amplitude');
# END SOLUTION
```
