## Exercise: Special Series

This exercise demonstrates some fundamental series for digital
signal processing.

- **a)** Generate an impulse series  
  $$
  x(k) = 1.2\cdot \delta(k-4), \qquad -5 \leq k \leq 10
  $$
  with
  $$
  \delta(k):=\begin{cases}
  1 & k=0\\
  0 & k\neq 0
  \end{cases}
  $$
  and plot this series using the function `stem()`.

- **b)** Plot the following Gaussian- and Laplacian series using the function
  `stem()`.  
  Why is it unfavorable to use the `plot()` command to represent the sequences?

 | Nr. | Signaldefinition | Wertebereich | Parameter | Beschreibung |
|:---:|:-----------------|:--------------|:-----------|:--------------|
| 1 | $x_1(k) = e^{-\alpha k^2}$ | $-20 \leq k \leq 20$ | $\alpha = 0.01$ | Gaussian series |
| 2 | $x_2(k) = e^{-\beta\,\|k\|}$ | $-20 \leq k \leq 20$ | $\beta = 0.2$ | Laplace series |
| 3 | $x_3(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{4})$ | $-30 \leq k \leq 30$ | $\beta = 0.1$ | mod. Laplacian series |
| 4 | $x_4(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{80})$ | $-600 \leq k \leq 600$ | $\beta = 0.005$ | — |

```python
# BEGIN SOLUTION
% #####  Part a:  Impulse Sequence  #####
k_a = -5:10;    % initializing the coefficient vector k with values from 
                % -5 to 10;
x = zeros(1,16);% x is initialized as a ROW vector with 16 zeros;
x(5+1+4) = 1;   % The Dirac impulse '1' should appear at time index k=4;
                % The time index k=-5 corresponds to Matlab index '1',
                % hence for k=4, Matlab index 10 should be chosen;
                 
x = 1.2 * x;    % Multiplying the sequence by the factor 1.2 ;

% # graphical output # %
figure; 
stem(k_a,x);
set(gca,'XTick',[-4:2:10]); 
title('x(k)=1.2 \delta(k-4)'); 
xlabel('Sampling Values k'); 
ylabel('Amplitude');

% #####  Part b:  Gaussian and Laplace Sequences  #####
% Coefficient vectors
k_b1 = -20:20;
k_b2 = -20:20;
k_b3 = -30:30;
k_b4 = -600:600;

% Alpha factor for Gaussian sequence
alpha = 0.01;

% Beta factors for Laplace sequences
beta_2 = 0.2;
beta_3 = 0.1;
beta_4 = 0.005;

% Calculation of sequences
x1 = exp(-alpha *k_b1.^2);
x2 = exp(-beta_2*abs(k_b2));
x3 = exp(-beta_3*abs(k_b3)) .* cos(2*pi/4*k_b3);
x4 = exp(-beta_4*abs(k_b4)) .* cos(2*pi/80*k_b4);

% # graphical output # %
% Gaussian sequence
figure;
plot(k_b1,x1);
hold on; 
stem(k_b1,x1,'k'); 
hold off;
title('Gaussian Sequence x_1(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% Laplace sequence
figure;
plot(k_b2,x2);
hold on;
stem(k_b2,x2,'k'); 
hold off;
title('Laplace Sequence x_2(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% Modified Laplace sequence beta = 0.1
figure;
plot(k_b3,x3);
hold on;
stem(k_b3,x3,'k');  
hold off;
grid on; 
title('Modified Laplace Sequence x_3(k)'); 
xlabel('k'); 
ylabel('Amplitude');

% Modified Laplace sequence beta = 0.005
figure;
plot(k_b4,x4,'k');
grid on;
title('Modified Laplace Sequence  x_4(k)'); 
xlabel('k'); 
ylabel('Amplitude');
# END SOLUTION
```
