### Solution: Analysis of a 2nd order system

---

#### a)
The schematic corresponds to the following equation

$$
Y(z) = Y(z) z^{-1} - 0.5 \cdot Y(z) z^{-2}
      + X(z) + 1.5 \cdot X(z) z^{-1} + X(z) z^{-2}
$$

Transformation to the time domain yields the difference equation

$$
y(k) = y(k-1) - 0.5 \cdot y(k-2)
       + x(k) + 1.5 \cdot x(k-1) + x(k-2).
$$

Transforming the transfer function results in the familiar expression:

$$
Y(z)\left(1 - z^{-1} + 0.5 \cdot z^{-2}\right)
=
X(z)\left(1 + 1.5 \cdot z^{-1} + z^{-2}\right)
$$

$$
H(z) = \frac{Y(z)}{X(z)}
     = \frac{1 + 1.5 \cdot z^{-1} + z^{-2}}
            {1 - z^{-1} + 0.5 \cdot z^{-2}}.
$$

---

#### b)
The impulse response can be obtained by exciting the system with

$$
x(k)=
\begin{cases}
1 & \text{for } k=0 \\
0 & \text{otherwise}
\end{cases}
$$

Its first 10 values are:

$$
h(k)=
\left[
1,\ 2.5,\ 3,\ 1.75,\ 0.25,\ -0.625,\ -0.75,\ -0.4375,\ -0.0625,\ 0.1562,\ \dots
\right]
$$

---

#### c)
The coefficient vectors are

$$
\mathbf{b} = \{1,\ 1.5,\ 1\}
$$

and

$$
\mathbf{a} = \{1,\ -1,\ 0.5\}.
$$

The connection between frequency response and zero-pole plot can be demonstrated by well-known relationships.
Zeros on the unit circle yield a break-in of the frequency response, while the poles are responsible for the maximum.
