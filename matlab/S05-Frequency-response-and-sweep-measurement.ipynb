**d)** In practice, a frequency response can usually be obtained using the so-called sweep measurements.  
Using sweep measurements, a system is excited with oscillations of different frequencies.  
An LTI-system responds to an oscillation with the same frequency but different phase and amplitude.  
If the frequency is varied across the part of the spectrum, which we are interested in ($0 \dots \pi$),  
the amplitude- and phase-response can be determined.

In this exercise the amplitude response shall be measured using an excitation

$$
x(k) = \cos(\Omega_i k)
$$

Compare the results with the outcome of b).

**e)** Excite the system with a step series $\varepsilon(k)$, and demonstrate that the first difference of the step response equals the impulse response.

```python
# BEGIN SOLUTION
H = [1,-1] ; % Impulse Response = Polynomial H(z) = Numerator Polynomial (H(z)) for FIR 
N = 32;      % Number of values to output

% #####  Part b:  Amplitude and Phase Response  #####

[y,x] = freqz(H,1,256,'whole') ;

% # graphical output # %
% Amplitude Response
figure;
plot(x/(2*pi),abs(y));
title('Amplitude Response'); 
xlabel('\Omega/2\pi'); 
ylabel('Amplitude');

% Phase Response
figure;
plot(x/(2*pi),-angle(y)/pi*180) ; % Note: -angle() because of phase definition
set(gca,'YTick',[-90,-45,0,45,90]);
title('Phase Response'); 
xlabel('\Omega/2\pi'); 
ylabel('Phase in Degrees');

% #####  Part c:  Impulse Response, Pole-Zero Plot  #####

h  = impz(H,1,4); % 5 samples
lh = length(h);

% # graphical output # %
% Impulse Response
figure;
stem(0:lh-1,h); 
axis([-0.1 3.1 -1 1]);
hold on; 
plot([-0.1,3.1],[0,0],'k'); 
hold off;
title('Impulse Response'); 
xlabel('k'); 
ylabel('Amplitude') ;

% Pole-Zero Plot
figure;
zplane(H,1);
title('Pole-Zero Plot');

% #####  Part d:  Sweep Measurement - Cosine Excitation #####
lf = length(H) ;
k_sweep = (0:N-1+2*(lf-1));    % Sequence for sweeping of length N+2*(lf-1)
n_sweep = 0:N/2-1;             % normalized frequencies
erg = [];

for n = n_sweep
    sweep = cos(2*pi*n/N*k_sweep) ;                 % Actual sweep measurement
    foo = conv(H,sweep);                          % Sweeping
    erg = [erg;max(abs(foo(1+(lf-1):length(foo)-(lf-1))))]; % without transient
end

[c,d] = freqz(H,1,N);  % Comparison Sweep - Amplitude Response

% # graphical output # %
figure;
plot(2*n_sweep/N,erg,'r+'); 
hold on;
plot(d/pi,abs(c),'b-') 
hold off;
title('+ Sweep Measurement, - theoretical Amplitude Response'); 
xlabel('\Omega/\pi'); 
ylabel('Amplitude');

% #####  Part e:  Sweep Measurement - Exponential Excitation #####
erg = [];
for n = n_sweep
    sweep = exp(j*2*pi*n/N*k_sweep);                 % Actual sweep measurement
    foo = conv(H,sweep);                              % Sweeping
    erg = [erg; max(abs(foo(1+(lf-1):length(foo)-(lf-1))))]; % without transient
end

% # graphical output % #
figure;
plot(2*n_sweep/N,erg,'r+');
hold on;
plot(d/pi,abs(c),'b-') 
hold off;
title('+ Sweep Measurement, - theoretical Amplitude Response'); 
xlabel('\Omega/\pi'); 
ylabel('Amplitude');

% #####  Part f:  Step Response  #####

epsilon = [zeros(10,1); ones(10,1)];
step_response = conv(h,epsilon);
hdiff=diff(step_response);

% # graphical output # %
% Step Sequence
figure;
stem((0:length(epsilon)-1)-10,epsilon);
title('Step Sequence'); 
xlabel('k'); 
ylabel('\epsilon(k)');

% Step Response
figure;
stem((0:length(step_response)-1)-10,step_response);
hold on; 
plot([-10,15],[0,0],'k'); 
hold off;
title('Step Response'); 
xlabel('k'); 
ylabel('Amplitude');

% Impulse Response
figure;
stem((0:length(hdiff)-1)-10,hdiff);
hold on; 
plot([-10,15],[0,0],'k'); 
hold off;
title('Impulse Response'); 
xlabel('k'); 
ylabel('Amplitude');
# END SOLUTION
```

### Solution: frequency response and sweep measurement

**a)**

$$
H(z) = 1 - z^{-m}
$$

The **amplitude response** $\left|H(e^{j \Omega})\right|$ of
the system is calculated as:

$$
H(e^{j \Omega}) = 1 - e^{-j \Omega m}
$$

$$
\begin{array}{rcl}
\left|H(e^{j \Omega})\right|
  & = & \left| 1 - e^{-j \Omega m} \right| \\
  & = & \left|  e^{-j \frac{\Omega}{2} m} \left( e^{j \frac{\Omega}{2} m} - e^{-j  \frac{\Omega}{2} m} \right) \right| \\
  & = & \underbrace{\left|  e^{-j \frac{\Omega}{2} m} \right|}_{1} \left|\left( e^{j \frac{\Omega}{2} m} - e^{-j \frac{\Omega}{2} m} \right) \right| \\
  & = & \left| \left( e^{j \frac{\Omega}{2} m} - e^{-j \frac{\Omega}{2} m} \right) \right| \\
  & = & \left| 2 j \sin \left( \frac{\Omega}{2} m\right) \right| \\
\left|H(e^{j \Omega})\right|  & = & 2 \left| \sin \left( \frac{\Omega}{2} m\right) \right|\\
\end{array}
$$

The **phase response** $b(\Omega)$ is according to the
definition: ATTENTION: The phase response is defined here
with a negative sign, so that the group delay is obtained
through positive derivation of the phase response $\tau_g(\Omega)
= \frac{\mathrm{d}\, b(\Omega)}{\mathrm{d}\, \Omega}$ \\
Normally, the phase response is also defined positive, so that
for the determination of the group delay, a negative sign is
needed.
$$
H(e^{j \Omega}) = \left| H(e^{j \Omega}) \right| e^{- j b(\Omega) }
$$
is established through:


$$
\begin{aligned}
b(\Omega) &= - \arg\!\left(H(e^{j\Omega})\right) \\
          &= - \arctan\!\left(\frac{\Im\{H(e^{j\Omega})\}}{\Re\{H(e^{j\Omega})\}}\right)
\end{aligned}
$$


For this purpose, it is practical to convert the transfer function
analogue to the above calculation:
$$
\begin{array}{rcl}
H(e^{j \Omega}) & = & 1 - e^{-j \Omega m} \\
                & = & e^{-j \frac{\Omega}{2} m}  \left[ e^{j \frac{\Omega}{2} m} - e^{-j \frac{\Omega}{2} m}\right] \\
        & = & \left[ \cos \left( \frac{\Omega}{2} m \right) - j \sin \left( \frac{\Omega}{2} m \right) \right]
        \left[ 2 j \sin \left( \frac{\Omega}{2} m \right) \right] \\
        & = & 2 \sin^2 \left( \frac{\Omega}{2} m \right) + j 2 \sin \left( \frac{\Omega}{2} m \right)
        \cos \left( \frac{\Omega}{2} m \right) \\
\end{array}
$$

$$
\begin{aligned}
b(\Omega) &= - \arctan \frac{\Im\{H(e^{j\Omega})\}}{\Re\{H(e^{j\Omega})\}} \\
          &= - \arctan \frac{ 2 \sin \left( \tfrac{\Omega}{2} m \right)
        \cos \left( \tfrac{\Omega}{2} m \right) }{ 2 \sin^2 \left( \tfrac{\Omega}{2} m \right)  } \\
          &= - \arctan \frac{ \cos \left( \tfrac{\Omega}{2} m \right) }{ \sin \left( \tfrac{\Omega}{2} m \right)  } \\
          &= - \arctan \cot \left( \tfrac{\Omega}{2} m \right)  \\
          & \downarrow \quad \text{mit } \arctan x = \tfrac{\pi}{2} - \operatorname{arccot} x \\
      &= - \tfrac{\pi}{2} + \operatorname{arccot}\, \cot \left( \tfrac{\Omega}{2} m \right)  \\
b(\Omega) &= - \tfrac{\pi}{2} + \tfrac{\Omega}{2} m
\end{aligned}
$$
