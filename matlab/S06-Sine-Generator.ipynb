# Sine Generator

In this exercise a sine generator for a fixed frequency shall be synthesized, which
is realized without using any tables or polynomial approximations.
The basis of this generator is a 2nd order system which, once excited,
produces the values of a sine signal.

For the synthesis process the following correspondence of the Z-transform shall be used:

$$
h(k)=\sin!\big(\Omega_0 k\big)
$$

$$
H(z) = Z{h(k)} = \frac{z,\sin(\Omega_0)}{z^2 - 2z\cdot\cos(\Omega_0) + 1}
$$

This correspondence can be interpreted as the impulse response $h(k)$ of a system
$H(z)=Y(z)/X(z)$.

### d)

Now the effects of quantizing the coefficients of the system shall be investigated.

Therefore replace the expression $2\cdot\cos(\Omega_0)$ in the coefficient vector `a`
by $2.1\cdot\cos(\Omega_0)$.

Display the zero–pole plot and the impulse response of the system.

```python
# BEGIN SOLUTION
% Normalized dimensionless angular frequency
Omega0 = 2*pi*1/32;

% #####  Part b:  Sinus Generator  #####
b = [0 sin(Omega0) 0];
a = [1 -2*cos(Omega0) 1];

% Sampled sine signal
k = 0:99; 
x = sin(Omega0*k);

% # Graphical output # %
% Poles and zeros in the z-plane
figure; 
zplane(b, a);
xlabel('Real Part'); 
ylabel('Imaginary Part');
title('z-Plane');

% Impulse response
figure; 
impz(b, a, 100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulse Response');
hold on; 
plot(k, x, 'r-'); 
hold off;

% #####  Part c:  Forward Path  #####
b = [1 0 0];
a = [1 -2*cos(Omega0) 1];

% # Graphical output # %
figure; 
impz(b, a, 100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulse Response');

% #####  Part d:  Instability  #####
b = [0 sin(Omega0) 0];
a = [1 -2.1*cos(Omega0) 1];

% # Graphical output # %
% Poles and zeros in the z-plane
figure; 
zplane(b, a);
xlabel('Real Part'); 
ylabel('Imaginary Part');
title('z-Plane');

% Impulse response
figure; 
impz(b, a, 100); 
xlabel('k'); 
ylabel('h(k)');
title('Impulse Response');
# END SOLUTION
```

## Solution: Sine Generator

#### a)

$$
H(z) = \frac{Y(z)}{X(z)} = \frac{z^{-1} \sin(\Omega_0)}{1 - 2 z^{-1} \cos(\Omega_0) + z^{-2}}
