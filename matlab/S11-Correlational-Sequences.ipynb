# Exercise 2.14: Correlation Series

Beside the moments of exercise~\ref{dsv2a11e} there are some more characteristical parameters for stochastic processes, i.e. *the autocorrelation, cross-correlation and the auto-covariance* resp. *cross-covariance*.

The common property of these functions is: they are values for the dependency of the observed processes and their random variables. In this exercise the autocorrelation series is observed exemplarily. The following stochastic processes are expected to be *ergodic*, see section (2.6). A main problem is the estimation of the theoretical autocorrelation series (2.6.6):

$$
r_{XX}(\kappa) = E\{X^\ast(k)\, X(k+\kappa)\}
$$

based on a finite number $N$ of samples.

The MATLAB function `xcorr()` estimates an
auto-/cross-correlation series of a stochastic
process.  
Calling `r = xcorr(x)` creates a not
normalized autocorrelation series from vector `x` using
formula  

$$
\hat{r}_{XX}(\kappa) = \sum_{k=0}^{N-|\kappa|-1} x(k)\,x^{\ast}(k+\kappa)
$$

In contrast to eq. (2.6.6) the autocorrelation series is defined conjugate even using MATLAB. Therefore the results of `xcorr()` are always conjugated.

Calling `r = xcorr(x, 'option')` allows to normalize the correlation series. As an `'option'` you have the following alternatives:

- biased:
$$
\hat{r}_{XX}(\kappa) = \frac{1}{N} \sum_{k=0}^{N-|\kappa|-1} x(k) \, x^*(k+\kappa)
$$

- unbiased
$$
\hat{r}'_{XX}(\kappa) = \frac{1}{N - |\kappa|} \sum_{k=0}^{N - |\kappa| - 1} x(k) \, x^*(k+\kappa)
$$

- coeff  
normalizes the correlation series $\hat{r}''_{XX}(\kappa)$ to have a value of one at $\kappa=0$

**a)**  
  Create a white Gaussian noise sequence of length $N = 8000$.  
  Estimate the autocorrelation series using function `xcorr()`  
  and plot the results.  
  Use all three different `'option'` settings.

```python
# BEGIN SOLUTION
n = randn(800,1);
kappa = -799:799;
s_biased   = xcorr(n,'biased').';
s_unbiased = xcorr(n,'unbiased').';
s_coeff    = xcorr(n,'coeff').';

figure; stem(kappa,s_biased);   grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autocorrelation Sequence: Biased');
figure; stem(kappa,s_unbiased); grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autocorrelation Sequence: Unbiased');
figure; stem(kappa,s_coeff);    grid; xlabel('kappa'); ylabel('Amplitude'); 
title('Autocorrelation Sequence: Normalized at Time-Lag Zero');
# END SOLUTION
```

**b)**  
  The Fourier transformation of the autocorrelation series is the
  (periodic) power density spectrum.  
  Use function `ldtft()` and plot the estimation for
  the power density spectrum.

```python
# BEGIN SOLUTION
n = randn(200,1);
s = xcorr(n,'biased');
[S W] = ldtft(s,512);

figure; semilogy(W/pi, abs(S) ); % normalized frequency
grid; xlabel('Omega/pi'); ylabel('$|S_{xx}(e^{j\Omega})|$', 'Interpreter', 'latex'); 
title('Power Density Spectrum');
# END SOLUTION
```
