# Design of Non-Recursive Filters Using Windowing Techniques

The complete filter design by windowing an impulse response of a time-shifted, ideal low-pass filter is implemented in MATLAB using the function `fir1()`.

Calling

```matlab
b = fir1(m, Wn)
```

generates $m+1$ filter coefficients of a low-pass filter with order `m` and cut-off frequency `Wn`.
The function call

```matlab
b = fir1(m, Wn, 'high')
```

yields the corresponding high-pass filter with cut-off frequency `Wn`.

```matlab
b = fir1(m, [Wn1 Wn2])
```

yields a pass-band filter with cut-off frequencies `Wn1` and `Wn2`.

The Hamming window is used as **default**.
In addition, the last parameter may be a vector with samples of an arbitrary window function:

```matlab
b = fir1(m, Wn, window)
```

This vector must cover $m+1$ samples.

Group delay and linearity of a system are closely connected.
Calculate the **group delay** of the low-pass filter from exercise a) (Hamming window) using MATLAB’s function `grpdelay()`.

````python
# BEGIN SOLUTION
% #####  Task a: Design Lowpass Filter  #######################################
% Filter coefficients            % Frequency response
bR  = fir1(50,0.4,boxcar(51));   [HR, W]  = freqz(bR, 1, 512);
bH  = fir1(50,0.4,hanning(51));  [HH, W]  = freqz(bH, 1, 512);
bHm = fir1(50,0.4,hamming(51));  [HHM, W] = freqz(bHm, 1, 512);
bB  = fir1(50,0.4,blackman(51)); [HB, W]  = freqz(bB, 512);
bK  = fir1(50,0.4,kaiser(51,8)); [HK, W]  = freqz(bK, 512);

% # Graphical Output # %
% Magnitude frequency response
figure;
plot(W/pi, 20*log10(abs(HR/max(HR))));        
hold on;
plot(W/pi, 20*log10(abs(HH/max(HH))), '--');
plot(W/pi, 20*log10(abs(HHM/max(HHM))), '-.'); 
hold off;
xlabel('Omega/pi'); 
ylabel('|H(exp(j*Omega))/H(1)| in dB'); 
grid
title(['Lowpass: - Rectangular -- von Hann  -. Hamming']); 
axis([0 1 -150 0]); 

figure;
plot(W/pi, 20*log10(abs(HB/max(HB))), '-');    
hold on;
plot(W/pi, 20*log10(abs(HK/max(HK))), '--');   
hold off;
xlabel('Omega/pi'); ylabel('|H(exp(j*Omega))/H(1)| in dB'); grid
title(['Lowpass: - Blackman -- Kaiser with beta=8']);   axis([0 1 -150 0]);

% #####  Task b: Design Highpass Filter  ######################################
% Filter coefficients    % Frequency response 
b = fir1(33, 0.4, 'high'); [H, W] = freqz(b, 1, 512);

% # Graphical Output # %
% Highpass
figure;
plot(W/pi, 20*log10(abs(H/max(H)))); 
grid;
xlabel('Omega/pi'); 
ylabel('|H(exp(j*Omega))/H(-1)| in dB'); 
title('Highpass');

% #####  Task c: Determine Group Delay  #######################################
figure;
b = fir1(50, 0.4);  % Filter coefficients
grpdelay(b, 1, 256); % Calculate group delay and plot
title('Group Delay in Samples'); 
ylabel(''); 
xlabel('Omega/pi');
axis([0 1 24 26]);
# END SOLUTION
````

When designing a high-pass with MATLAB’s function `fir1()`, MATLAB generates the following warning:

```
For highpass and bandstop filters, order must be even.
Order is being increased by 1.
```

The `fir1()` function designs filters of type I or II depending on whether the order is even or odd.
When the order is odd (type II), the frequency response must have a zero at the Nyquist frequency $\Omega/\pi=1$, which contradicts the demanded high-pass behavior.
Therefore, MATLAB increases the filter order automatically (type I).
