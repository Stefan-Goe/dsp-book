# Remez Design (Chebyshev-Approximation)

By designing IIR filters using Chebyshev approximation, it was shown that this approximation delivers a great benefit because it utilizes the whole tolerance scheme. Such a design method (applying a Chebyshev approximation) for FIR systems is implemented in MATLAB’s `remez()` function (Parks–McClellan algorithm). The approximation takes place in the pass and stop bands.

The call of `b=remez(m,F,G)` designs an \(m\)-th order filter, whose specification is given by vectors `F` and `G`. `F` includes a pair of frequencies (between 0 and 1) to set the pass or stop bands and `G` is the frequency response (amplitude) corresponding to `F`.

**Example:**
```matlab
F = [0 0.3 0.5 1]
G = [1 1 0 0]

## Tasks
a)
Calculate a filter with the upper frequency response. Experiment with different orders and count the number of extreme values within the approximation interval.

b)
The call of ```b=remez(m,F,G,W)``` weights a desired frequency response where frequency bands with a lower weight are less accurately approximated than frequency bands with a higher weight. The number of weights within vector W is half the number of entries in F resp. G (for each frequency band one weight factor).

The frequency band between 0.0 and 0.3 (pass-band) doesn’t have to approximate the desired frequency response very closely. A higher weight should be used for the frequency band between 0.5 and 1. Calculate the filter again and compare the frequency response with the filter designed in task a).

c)
Design the low-pass - high-pass combination as specified below, where the passbands should have a (by factor 4) higher weight than the stopbands:

| Band        | Frequency Range |
|-------------|-----------------|
| Passband 1  | 0 – 0.2         |
| Stopband 1  | 0.26 – 0.44     |
| Passband 2  | 0.5 – 0.7       |
| Stopband 2  | 0.76 – 1        |


Filter order $m$: 128

**Note:** The original `remez()` function has been replaced by `firpm()` in MATLAB 7 and later. While the basic principle remains the same, check the documentation for possible differences in parameter syntax.

````python
# BEGIN SOLUTION
% #####  Task a: Design Filter  ###############################################
f   = [0 0.3 0.5 1.0];     % Frequencies of the stopband and passband
g   = [1 1 0 0];           % Impulse response
v_m = [9 10];              % Vector with filter orders

for m = v_m
    b = firpm(m, f, g);         % Design filter using Remez
    [H, W] = freqz(b, 1, 512);  % Compute the frequency response

    % # Graphical Output # %
    string = sprintf('Desired/Actual Response - Filter Order: %d', m);
    figure;
    plot(f, g, W/pi, abs(H));
    line([0 0.3], [2-max(abs(H)) 2-max(abs(H))], 'linestyle', '--'); % Create line
    line([0 0.5], [max(abs(H)) max(abs(H))], 'linestyle', '--');     % in plot
    line([0.5 1], [max(abs(H(400:end))) max(abs(H(400:end)))], 'linestyle', '--');
    line([0.3 0.3], [2-max(abs(H)) 0], 'linestyle', '--');
    line([0.5 0.5], [max(abs(H)) max(abs(H(400:end)))], 'linestyle', '--');
    grid;
    xlabel('Omega/pi');
    ylabel('|H(exp(j*Omega))|');
    title(string);
end;

% #####  Task b: Weight Frequency Response  ###################################
f   = [0 0.3 0.5 1.0];     % Frequencies of the stopband and passband
g   = [1 1 0 0];           % Impulse response
w   = [0.5 2];             % Weights of the passband and stopband
v_m = [9 10];              % Vector with filter orders

for m = v_m
    b = firpm(m, f, g, w);
    [H, W] = freqz(b, 1, 512);
    
    % # Graphical Output # %
    string = sprintf('Desired/Actual Response - Filter Order: %d', m);
    figure; plot(f, g, W/pi, abs(H));
    line([0 0.3], [2-max(abs(H)) 2-max(abs(H))], 'linestyle', '--');
    line([0 0.5], [max(abs(H)) max(abs(H))], 'linestyle', '--');
    line([0.5 1], [max(abs(H(400:end))) max(abs(H(400:end)))], 'linestyle', '--');
    line([0.3 0.3], [2-max(abs(H)) 0], 'linestyle', '--');
    line([0.5 0.5], [max(abs(H)) max(abs(H(400:end)))], 'linestyle', '--');
    grid;
    xlabel('Omega/pi');
    ylabel('|H(exp(j*Omega))|');
    title(string);
end;

% #####  Task c: Lowpass-Bandpass Combination  ###############################
f = [0 0.2 0.26 0.44 0.5 0.7 0.76 1]; % Vector of corner frequencies
g = [1  1   0    0    1   1   0   0]; % Vector of frequency responses
w = [1       4         1       4];    % Vector of weights
m = 128;                              % Filter order
b = firpm(m, f, g, w); 
[H, W] = freqz(b, 1, 512);

% # Graphical Output # %
figure;
plot(W/pi, 20*log10(abs(H)));
grid; 
xlabel('Omega/pi');
ylabel('|H(exp(j*Omega))| in dB');
title('LP-BP Combination');
# END SOLUTION
````
