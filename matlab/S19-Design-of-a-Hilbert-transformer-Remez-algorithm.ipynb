### Exercise: Design of a Hilbert Transformer

In this exercise a Hilbert transformer with an ideal frequency response 

$$
H(e^{j\Omega}) = j\,\mathrm{sgn}(\Omega)
$$

shall be approximated using the remez algorithm (`remez()` in MATLAB).

---

#### a) 

Design two Hilbert transformers of order 30 and 31 using the additional argument `'h'`:

```matlab
b = remez(m, F, G, 'h');
````

Plot the amplitude response, impulse response, and the pole-zero planes.
What are the types of the created filters?

---

#### b)

Create a sinusoidal signal with a duration of two seconds, a signal frequency of $50\text{Hz}$, and a sampling frequency of $1000\text{Hz}$.

Filter this signal with the Hilbert transformer of order $m=30$ and plot the input and output signals.

*Hint:* The group delay has to be taken into account, because the output signal is delayed.

---

#### c)

For the Hilbert transform (important in communications), MATLAB has its own function:

```matlab
y = hilbert(x);
```

The return value `y` is the analytic signal of `x`, the real part of `y` is equal to `x`, and the imaginary part of `y` is the Hilbert transform of `x`.

Calculate the Hilbert transform of the signal of exercise **b** and compare the results.

`````python
# BEGIN SOLUTION
% #####  Part a:  Design Hilbert Transformer  #####
b1 = remez(30,[0.05 0.95],[1 1],'h');
b2 = remez(31,[0.05 1],[1 1],'h');

[H1 W1] = freqz(b1,1,512);  
[H2 W2] = freqz(b2,1,512);  

% # Plot frequency response # %
figure; 
plot(W1/pi,abs(H1)); 
grid; 
xlabel('Omega/pi');
ylabel('|H_H(exp(j*Omega))|'); 
title('Filter Order: 30 Type II');

figure; 
plot(W2/pi,abs(H2)); 
grid; 
xlabel('Omega/pi');
ylabel('|H_H(exp(j*Omega))|'); 
title('Filter Order: 31 Type IV');

% # Plot impulse responses # %
figure; 
stem([0:30], b1); 
axis([0 30 -0.8 0.8]);     
grid; 
xlabel('k');
ylabel('Impulse Response');       
title('Filter Order: 30 Type II');

figure; 
stem([0:31], b2); 
axis([0 31 -0.8 0.8]);     
grid; 
xlabel('k');
ylabel('Impulse Response');       
title('Filter Order: 31 Type IV');

% #####  Part b:  Filter the input signal  #####
b  = remez(30,[0.05 0.95],[1 1],'h'); 
Fs = 1000;                            
t  = (0:1/Fs:2)';                     
x  = sin(2*pi*50*t);                  
xh = filter(b,1,x);                  

% # Plot pole-zero diagram # %
figure;
zplane(b); 
xlabel('Real Part'); 
ylabel('Imaginary Part');
title('Pole-Zero Diagram');

xd = [zeros(15,1); x(1:length(x)-15)];  % Delay input signal by 15 samples

% # Compare delayed input and filtered output # %
figure; 
plot(t(1:50),xd(1:50),t(1:50),xh(1:50),'--'); 
grid; 
xlabel('Time (s)'); 
ylabel('Amplitude'); 
title(' - Input Signal   -- Output Signal');

% #####  Part c:  Compute Hilbert Transform using built-in function  #####
Fs = 1000;
t  = (0:1/Fs:2)';
x  = sin(2*pi*50*t);
y = hilbert(x);   % Analytical signal (Hilbert transform)

% # Plot real and imaginary parts # %
figure; 
plot(t(1:50), real(y(1:50))); 
grid; 
hold on;
plot(t(1:50), imag(y(1:50)),'--'); 
xlabel('Time (s)'); 
ylabel('Amplitude');
title('-- Hilbert Transform   - Original');
% #####  EOF  #####

# END SOLUTION
`````

# Solution: Design of a Hilbert Transformer

### a) Filter Design and Impulse/Amplitude Response

If we consider the optional parameter[^1], we will get filter type II for a filter order $m=30$ and filter type IV for filter order $m=31$. The following figures show the amplitude responses and the impulse responses for the designs.

Here, it also shows that the quality of approximation is higher for odd filter orders because of the missing zero at $\Omega = \pi$.

In order to use a Hilbert transformer to produce analytic signals, the output signal (imaginary part) and the input signal (real part) must be delayed equally and combined. This means that an even group delay should exist. By a non-recursive, linear system, the contribution of every zero to group delay is $-0.5$. Therefore, in the case of a Hilbert transformer, an even filter order must exist. This is a contradiction to achieving high approximation quality.

To clarify the position of zeros, the following figure shows the zero distribution of the filter of order $m=30$.

<img src="./../pngs/ldsv5a05_a5.png" width="250">

There are $15$ zeros, which lie either on the unit circle or as a pair reflected on the unit circle, because the system is linear and non-recursive.

[^1]: If this parameter is not fixed, then we get the filter from type I/III. The design is therefore non-impulse invariant, see exercise 5.4.a.
