# Exercise: Properties of the DFT

Start with the series

$$
x(k) = 
\begin{cases}
k + j \cdot (7-k), & k = 0 \dots 15\\
0, & \text{otherwise}
\end{cases}.
$$

**a)** Calculate the DFT $X(n)$ using the function `fft(x)`.  
      Plot the real part $\Re X(n)$ and the imaginary part $\Im X(n)$ of $X(n)$.

**b)** Plot the DFT of $\Re x(k)$ and $j\,\Im x(k)$  
      (again, real and imaginary parts).  
      What kind of symmetries do the DFTs of $\Re x(k)$ and $j\,\Im x(k)$ show?

**c)** Produce a cyclic shift of $x(k)$ by
      $\lambda=4$ and $\lambda=8$ cycles:  

$$
x_1(k) = x((k+\lambda) \bmod M)
$$

What is the relation between $X_1(n) = \mathrm{DFT} x_1(k)$ and $X(n)$?

**d)** Calculate the discrete Fourier transform $X_2(n)$ of  

$$
x_2(k) = x(k) \cdot \exp\Big(-j \, 2\pi \, 3 k / 16\Big)
$$

What is the relation between $X_2(n)$ and $X(n)$?

```python
# BEGIN SOLUTION
% #####  Task a:  Compute DFT of x(k)  #####
kk = 0:15;
x  = kk + j*(7-kk);
xr = real(x);  
xi = imag(x);

X  = fft(x); 
Xr = real(X);  
Xi = imag(X);

% # graphical output # %
% Real part
figure;
stem(kk,Xr);
title('Re(X(n))');
xlabel('n'); 
grid;

% Imaginary part
figure;
stem(kk,Xi);
title('Im(X(n))');
xlabel('n');
grid;

input('....For next task: press RETURN');

% #####  Task b:  DFT of Re{x(k)} and Im{x(k)}  #####
% # graphical output # %
figure;stem(kk,real(fft(xr)));  title('Re(FFT(Re(x(k))))');  xlabel('n');grid;
figure;stem(kk,imag(fft(xr)));  title('Im(FFT(Re(x(k))))');  xlabel('n');grid;
figure;stem(kk,real(fft(j*xi)));title('Re(FFT(j*Im(x(k))))');xlabel('n');grid;
figure;stem(kk,imag(fft(j*xi)));title('Im(FFT(j*Im(x(k))))');xlabel('n');grid;

input('....For next task: press RETURN');

% #####  Task c:  Cyclic Shift  #####
xtmp = [x x];
x1_04 = xtmp(kk+5); 
X1_04 = fft(x1_04);
x1_08 = xtmp(kk+9); 
X1_08 = fft(x1_08);

figure;stem(kk,Xr);         title('Re(X(n))');           xlabel('n');grid;
figure;stem(kk,Xi);         title('Im(X(n))');           xlabel('n');grid;
figure;stem(kk,real(X1_04));title('Re(X1(n)), lambda=4');xlabel('n');grid;
figure;stem(kk,imag(X1_04));title('Im(X1(n)), lambda=4');xlabel('n');grid;
figure;stem(kk,real(X1_08));title('Re(X1(n)), lambda=8');xlabel('n'); grid;
figure;stem(kk,imag(X1_08));title('Im(X1(n)), lambda=8');xlabel('n'); grid;

input('....For next task: press RETURN');

% #####  Task d:  Multiplication with exp(j*...)  #####
x2 = x.*exp(-j*2*pi*3*kk/16);
X2 = fft(x2);

figure;stem(kk,Xr);      title('Re(X(n))'); xlabel('n');grid;
figure;stem(kk,Xi);      title('Im(X(n))'); xlabel('n');grid;
figure;stem(kk,real(X2));title('Re(X2(n))');xlabel('n');grid;
figure;stem(kk,imag(X2));title('Im(X2(n))');xlabel('n');grid;
# END SOLUTION
```

# Solution: Properties of the DFT

## Part (a)

For a better illustration of the properties of the discrete Fourier transform (DFT), a periodic signal with 16 periods and not a finite signal with length 16 (like given in the exercise question) is considered. The DFT of $x(k)$ (real and imaginary part) is plotted beside the figures of exercise part (b).

## Part (b)

The DFT of a real (imaginary) signal is conjugate even (odd), that means the real part is even (odd) with respect to the frequency $0$ and the imaginary part is odd (even). To understand the symmetry properties, it has to be taken into account that the spectra are continuously periodic in both the positive and negative directions. From table 6.1, it can be seen how to calculate $\Re X(n)$ (or $\Im X(n)$) from $X(n)$.

## Part (c)

According to table 6.1, a displacement in the time domain by $\lambda$ samples corresponds to a multiplication of $X(n)$ with $\exp(j\,2\pi n\lambda/N)$ in the frequency domain. For $N=16$ and $\lambda =4$ resp. $\lambda =8$ we obtain:

$$
X_1(n) = j^n X(n) \quad \text{resp.} \quad X_1(n) = (-1)^n X(n)
$$

## Part (d)

According to table 6.1, it is obtained that:

$$
\text{DFT}\{ \exp(-j\,2\pi i k/N) x(k) \} = X((n+i))_N
$$

The notation $((i))_N$ describes a modulo-$N$ indexing. The multiplication of $x(k)$ with $\exp(-j\,2\pi\,3k/16)$ leads to a frequency offset of $3 f_A/16$, i.e.

$$
X_2(n) = X((n+3))_{16}
$$
