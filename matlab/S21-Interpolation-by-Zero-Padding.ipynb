# Exercise: Interpolation by Zero Padding

**Task:**

Calculate the DFT of  

$$
x(k) = 1 - \cos\left(\frac{2\pi k}{64}\right), \quad k = 0, \dots, 63
$$

Do common characteristics exist with the discrete Fourier transform of the signal

$$
x_0(k) =
\begin{cases}
x(k), & k = 0 \dots 63 \\
0,    & k = 64 \dots 127
\end{cases}
$$

---

### MATLAB Hints

```matlab
N = 64;
kk = 0:N-1;
x = 1 - cos(2*pi*kk/N);   % defines x(k)
x0 = [x, zeros(1,N)];     % appends a zero vector (1xN)

````python
# BEGIN SOLUTION
N  = 64;
x  = 1 - cos(2*pi/N*(0:N-1));    
X  = fft(x);

x0 = [x zeros(1, N)];           
X0 = fft(x0);

% # graphical output # %
figure; 
stem(0:N-1, abs(X));
axis([0 N-1 0 max(X)]);
title('abs(X(n))');
xlabel('n');

figure; 
stem(0:2*N-1, abs(X0)); 
axis([0 2*N-1 0 max(X)]);
title('abs(X0(n))');
xlabel('n'); 
# END SOLUTION
````

## Solution: Interpolation by Zero Padding

Through *zero padding* (filling up a vector with zeros) the spectral resolution of the DFT can be improved.  
Because the series $x(k)$ in this exercise has double length through zero padding, the spectral resolution is also doubled:  
every second value of $X_0(n)$ matches with the respective value of $X(n)$:

$$
X_0(2n) \equiv X(n), \quad n = 0, 1, 2, \dots, 63
$$

---

From the right-hand figure, we see the following: if the number of added zeros is infinite, the spectra of the periodic continuous signal  

$$
\tilde{x}(t) = 1 - \cos\left(\frac{2\pi t}{T}\right), \quad \forall t
$$

will not be approximated, but rather the spectra of the non-periodic signal

$$
x(t) =
\begin{cases}
1 - \cos\left(\frac{2\pi t}{T}\right), & 0 \leq t < T \\
0, & \text{else}
\end{cases}
$$
