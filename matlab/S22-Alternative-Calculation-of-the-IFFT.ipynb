# Exercise: Alternative Calculation of the IFFT

Perform the following steps:

1. Calculate the FFT of the time-discrete sequence $x_0(k)$ as used in exercise **dsv6a02e**.  
   (Recall: $x_0(k)$ is the length-128 sequence which equals $x(k)$ for $k=0,\dots,63$ and is zero for $k=64,\dots,127$.)

2. Exchange the real and imaginary parts of the transformed signal (i.e. replace each complex value $X(n)=\Re\{X(n)\}+j\Im\{X(n)\}$ by $\Re\{X(n)\}^\prime + j\Im\{X(n)\}^\prime$ where $\Re\{X(n)\}^\prime=\Im\{X(n)\}$ and $\Im\{X(n)\}^\prime=\Re\{X(n)\}$).

3. Transform this modified spectrum using the FFT again.

4. Exchange the real and imaginary parts of the resulting time-domain sequence.

Why is the resulting sequence — up to a scaling factor — equal to the original sequence $x_0(k)$?

```python
# BEGIN SOLUTION
N  = 64;
x  = 1 - cos(2*pi/N*(0:N-1));
x0 = [x zeros(1, N)];

X0 = fft(x0);                % FFT
X1 = imag(X0) + j*real(X0);  % Swap Re() and Im()
x1 = fft(X1);                % Another FFT
x2 = imag(x1) + j*real(x1);  % Swap Re() and Im()

% # graphical output # %
figure; 
bar(0:2*N-0.5, x0); 
title('x0(k)'); 
xlabel('k'); 
axis([-.5 2*N-1 0 2.1]);

figure; 
bar(0:2*N-0.5, x2); 
title('x2(k)'); 
xlabel('k'); 
axis([-.5 2*N-1 0 269]);
# END SOLUTION
```

## Solution: Alternative Calculation of the IFFT

The *m-file* for the exercise *Alternative Calculation of the IFFT* implements, according to the description in paragraph 6.5,  
an IFFT using the FFT twice. As shown in the figures, the results equal the original data sequence — up to the factor $N$.

If $x_2(k)$ is identical to $x_0(k)$, the signal $x_2(k)$ must be normalized by the FFT length $N = 128$ in the final stage,  
because of the relation

$$
\text{IDFT}\{X(n)\} = \frac{1}{N}\, j \, \left( \text{DFT}\{ j \, X^*(n) \} \right)^*
$$
