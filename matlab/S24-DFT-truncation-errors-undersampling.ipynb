### Exercise: DFT: truncation error / sub-sampling

Calculate the discrete fourier transform  

$$
X_i(n) = \mathrm{DFT}\{x_i(k)\}
$$  

of the following series:


\begin{aligned}
x_1(k) &= \sin(k \pi / 8), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63 \\
x_2(k) &= \sin(k \pi / 2), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63 \\
x_3(k) &= \sin(3 k \pi / 2), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63
\end{aligned}


Use MATLAB's `fft(x)`-function.

Plot the time domain signal $x_i(k)$, as well as the absolute values of the spectra $|X_i(n)|$ for both ranges of time indices.

In which cases is the spectrum of the associated continuous sine-signal $x_i(t)$ reproduced exactly?

```python
# BEGIN SOLUTION
kk08  = 0:7;
kk64  = 0:63;
kk100 = 0:8/100:8; 

x1_08 = sin(kk08 * pi / 8);    x1_64 = sin(kk64 * pi / 8);
x2_08 = sin(kk08 * pi / 2);    x2_64 = sin(kk64 * pi / 2);
x3_08 = sin(kk08 * pi * 3 / 2);  x3_64 = sin(kk64 * pi * 3 / 2);

% # Graphical output # %
figure; 
plot(kk100, sin(kk100 * pi / 8), ':'); 
hold on;
stem(kk08, x1_08); 
axis([0 8 -1.2 1.2]); 
title('x1(k) = sin(k * pi / 8), k = 0...7');

figure;
stem(kk08, abs(fft(x1_08))); 
axis([0 7 -1 6]); 
title('abs(X1(n)), n = 0...7');

figure;
stem(kk64, x1_64);
axis([0 64 -1.2 1.2]);
title('x1(k) = sin(k * pi / 8), k = 0...63');

figure;
stem(kk64, abs(fft(x1_64)));
axis([0 64 -1 40]);
title('abs(X1(n)), n = 0...63');

input('....To continue: press RETURN');

figure; 
plot(kk100, sin(kk100 * pi / 2), ':'); 
hold on;
stem(kk08, x2_08); 
axis([0 8 -1.2 1.2]); 
title('x2(k) = sin(k * pi / 2), k = 0...7');

figure;
stem(kk08, abs(fft(x2_08))); 
axis([0 7 -1 6]); 
title('abs(X2(n)), n = 0...7');

figure;
stem(kk64, x2_64);
axis([0 64 -1.2 1.2]);
title('x2(k) = sin(k * pi / 2), k = 0...63');

figure;
stem(kk64, abs(fft(x2_64)));
axis([0 64 -1 40]);
title('abs(X2(n)), n = 0...63');

input('....To continue: press RETURN');

figure;
plot(kk100, sin(kk100 * pi * 3 / 2), ':'); 
hold on;
stem(kk08, x3_08); 
axis([0 8 -1.2 1.2]); 
title('x3(k) = sin(k * pi * 3 / 2), k = 0...7');

figure;
stem(kk08, abs(fft(x3_08))); 
axis([0 8 -1 6]); 
title('abs(X3(n)), n = 0...7'); 
axis([0 7 -1 6]); 

figure;
stem(kk64, x3_64);
axis([0 64 -1.2 1.2]);
title('x3(k) = sin(k * pi * 3 / 2), k = 0...63');

figure;
stem(kk64, abs(fft(x3_64)));
axis([0 64 -1 40]);
title('abs(X3(n)), n = 0...63');
# END SOLUTION
```

### Solution: DFT: truncation error / sub-sampling

$\tilde{x}_\nu(k)$ is the periodic continuation of the series
$x_\nu(k)$. The spectrum of the corresponding continuous
sinusoidal signal $x_\nu(t)$ with period $T$ is exactly described
by $\mathrm{DFT}\{\tilde{x}_\nu(k)\}$, if

1. $\tilde{x}_\nu(k)$ is identical with the sampled continuous sinusoidal signal $x_\nu(t)|_{t=kT_A}$  
2. AND the sampling theorem is fulfilled, i.e. the sampling rate $1/T_A$ is higher than double the frequency of the continuous sinusoidal signal $x_\nu(t)$.  
   In other words: $T_A < T/2$

---

**Signal $x_1(k)$:** In both cases, the sampling theorem is
fulfilled (the continuous sinusoidal signal is plotted dotted). For $k=0..7$, condition (1) is however violated,
so that $X_1(n)$ shows spectral components at many frequencies 
(instead of only two). For $k=0..63$, both conditions are
fulfilled. The spectrum of the continuous sinusoidal signal is obtained.

**Signal $x_3(k)$:** The sampling theorem is violated here.  
Even though the frequency of the continuous sinusoidal signal
is higher than $x_2(t)$, the sampling rate is not increased
accordingly. Because $x_3(k)$ corresponds with $x_2(k)$ until
the sign, the FFT yields exactly the same results as the signal
$x_2(k)$. The spectrum $X_3(n)$ does not describe the spectrum of
the continuous sinusoidal signal $x_3(t)$.
