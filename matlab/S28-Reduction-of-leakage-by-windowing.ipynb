### Exercise — Effect of Hamming Windowing on the FFT

Plot the two-times repeated continuation of the series  
$$
x_1(k) = \sin\left(\frac{k\pi}{8}\right)
\quad \text{and} \quad
x_2(k) = \sin\left(\frac{k\pi}{10}\right)
\quad \text{for} \quad k = 0 \dots 63,
$$
as well as the absolute value of their  
64-point FFT  
$$
X_i(n) = \text{FFT}\{x_i(k)\}.
$$

Multiply the data series  
$$
x_i^{Hm}(k) := x_i(k) \cdot f^{Hm}(k)
$$
with a 64-point **Hamming window** $f^{Hm}(k)$,  
and again plot the two-times repeated continuation of $x_i^{Hm}(k)$,  
as well as the absolute value of their 64-point FFT  
$$
X_i^{Hm}(n).
$$

What are the effects of Hamming windowing on the absolute values  
of the coefficients of the FFT?

---

**MATLAB hint:**  
Calculation and plotting of the windowing results can be done with  
```matlab
stem(0:63, abs(fft(xi .* hamming(64).')))

````python
# BEGIN SOLUTION
N    = 64;
kk   = (0:N-1);            kk2  = (0:2*N-1);
x1   = sin(kk*pi/8);       X1   = fft(x1);
x1Hm = x1.*hamming(N).';   X1Hm = fft(x1Hm);
x2   = sin(kk*pi/10);      X2   = fft(x2);
x2Hm = x2.*hamming(N).';   X2Hm = fft(x2Hm);

% # graphical output # %
figure; 
stem(kk2, [x1 x1]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k');
ylabel('x1((k))');
title('x1(k)=sin(k*pi/8), k=0..63, doubly periodized'); 

figure; 
stem(kk, abs(X1));  
axis([0 N 0 1.1*N/2]);  
xlabel('n');
ylabel('|X1(n)|');
title('Magnitude spectrum of x1(k)');

figure; 
stem(kk2, [x1Hm x1Hm]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k');
ylabel('x1_Hm((k))');
title('Hamming-windowed x1(k), doubly periodized');

figure; 
stem(kk, abs(X1Hm));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X1_Hm(n)|');
title('Magnitude spectrum with Hamming windowing'); 

input('......for 2nd dataset: press RETURN');

figure; 
stem(kk2, [x2 x2]);  
axis([0 2*N -1.1 1.1]);  
xlabel('k'); 
ylabel('x2((k))');
title('x2(k)=sin(k*pi/10), k=0..63, doubly periodized'); 

figure; 
stem(kk, abs(X2));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X2(n)|');
title('Magnitude spectrum of x2(k)'); 

figure; 
stem(kk2, [x2Hm x2Hm]);  
axis([0 2*N -1.1 1.1]);
xlabel('k');
ylabel('x2_Hm((k))');
title('Hamming-windowed x2(k), doubly periodized'); 

figure; 
stem(kk, abs(X2Hm));  
axis([0 N 0 1.1*N/2]);
xlabel('n');
ylabel('|X2_Hm(n)|');
title('Magnitude spectrum with Hamming windowing');
# END SOLUTION
````

### Solution — Leakage Reduction by Windowing

**Series** $x_1(k)$ **without windowing:**  
If the DFT length $N=64$ is chosen, exactly $m=4$ periods of the sinusoidal signal  
fall in the DFT window of length $N T_A$. The portrayed periodic repetition of the series $x_1(k)$ after $N$ values gives a continuous plot without jumps.  
The leakage effect does not exist, i.e., the spectral function $X_1(n)$ reproduces the correct spectra of the continuous sinusoidal oscillation.

<img src="./pngs/ldsv7a07_x1ohne1.png" width="250">
<img src="./pngs/ldsv7a07_x1ohne2.png" width="250">

---

**Series** $x_1(k)$ **with windowing:**  
With the same DFT length, the series is weighted next with a Hamming window, i.e., in the time domain, multiplied with

$$
f^{Hm}(k) =
\begin{cases}
0.54 - 0.46 \cdot \cos\left(\frac{2\pi k}{N-1}\right), & 0 \le k \le N-1 \\
0, & \text{else}.
\end{cases}
$$

This corresponds to a convolution with three displaced spectral functions of a rectangular window in the spectral domain, where the middle is evaluated with a factor $0.54$.  
As a result, the main spectral lines have smaller values.  
Due to the reduced spectral resolution, two additional lines emerge.

<img src="./pngs/ldsv7a07_x1ham1.png" width="250">
<img src="./pngs/ldsv7a07_x1ham2.png" width="250">

---

**Series** $x_2(k)$ **without windowing:**  
For $N=64$, $m = N/20 = 3.2$ periods of the sinusoidal signal fall into the DFT window.  
The portrayed periodic repetition of $x_2(k)$ after $N$ values shows a discontinuous trend with jumps.  
Because the DFT assumes a periodic continuation of the series outside the considered DFT window, the **leakage effect** exists due to the jumps: as seen in the figure below, the main spectral lines at $n=3$ and $n=61$ "leak" through all DFT bins.

<img src="./pngs/ldsv7a07_x2ohne1.png" width="250">
<img src="./pngs/ldsv7a07_x2ohne2.png" width="250">

---

**Series** $x_2(k)$ **with windowing:**  
If we apply a Hamming window, the jumps are smoothed.  
This leads to a reduction of the leakage effect at the cost of spectral resolution.
