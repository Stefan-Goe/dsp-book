### Exercise: Transformation of real valued bandpass signals

Based on the generalized sampling theorem, the mean frequency of
the bandpass signal can be influenced by down-sampling.

The function  

$$
lband(N, \tfrac{f_m}{f_A}, \tfrac{b}{f_A})
$$  

creates a discrete bandpass process with mean frequency $f_m$ and bandwidth $b$.

Compare, for $N = 1000$, $f_m / f_A = 0.2$ and $b / f_A = 0.02$,  
the absolute values in the spectra of the original signal and the
down-sampled signals by factors 9, 10, 11 and 12.

---

*Matlab hints:*  

```matlab
b1 = lband(1000, 0.2, 0.02);   % bandpass signal
b2 = b1(1:10:1000);            % 10-times down-sampled bandpass signal
plot(abs(fft(b2 .* hanning(length(b2)))))

````python
# BEGIN SOLUTION
NFFT = 2^10;
b1 = lband(1000, 0.2, 0.02);  
fT = 0:1/NFFT:1-1/NFFT;

% Original signal
b2 = b1; 
lb = length(b2); 
aB2_hn = abs(fft(b2 .* hanning(lb), NFFT));

% # Graphical output # %
figure; 
plot(fT, aB2_hn);
axis([0 1 0 1.1 * max(aB2_hn)]); 
xlabel('Omega/2pi = f T');
ylabel('|B(exp(j*Omega))|');

for k = [9 10 11 12]  % Undersampling by a factor of k
    b2 = b1(1:k:1000); 
    lb = length(b2); 
    aB2_hn = abs(fft(b2 .* hanning(lb), NFFT));
    
    % # Graphical output # %
    figure; 
    plot(fT, aB2_hn); 
    axis([0 1 0 1.1 * max(aB2_hn)]); 
    xlabel('Omega/2pi = f T');
    ylabel('|B_M(exp(j*Omega))|'); 
    title(sprintf('Undersampling by a factor of M=%d', k));
end;
# END SOLUTION
````

### Solution: Transformation of real valued bandpass signals

The function 
$$
b = lband\!\left(N, \tfrac{f_m}{f_A}, \tfrac{b}{f_A}\right)
$$   
yields the time-discrete real bandpass signal

$$
b(k) = \frac{1}{11}\sum_{\nu=-5}^{5} \left(1+\frac{\nu}{10}\right)
\cos\!\left(2\pi \left[f_m + \nu\frac{b}{10}\right]\frac{k}{f_A}\right),
\quad k = 0,\dots,N-1
$$

with bandwidth $b$ and center frequency $f_m$.  
Because $b(k)$ is real and even, the corresponding ($2\pi$-periodic) spectrum 
$B(e^{j\Omega})=\mathcal{F}\{b(k)\}$ is conjugate-even.  
Therefore, the absolute spectrum $|B(e^{j\Omega})|$ of the original signal is also even.

---

For the given bandpass process, $f_m=f_A/5$ and $b=f_A/50$.  
That means the center frequency is $10$ times higher than the bandwidth: $f_m = 10\,b$.  

Considering the $2\pi$-periodicity of the spectrum, we obtain the absolute spectrum 
$|B(e^{j\Omega})|$, which is shown in the first plot with center frequencies

- $f_m = f_A/5 \;\Rightarrow\; \Omega_m/2\pi = 1/5$  
- $f_m' = -f_A/5 = -f_A/5 + f_A = 4f_A/5 \;\Rightarrow\; \Omega_m'/2\pi = 4/5$  

---

### Downsampling

If the bandpass signal $b(k)$ is downsampled by factor $M$, i.e. $b_M(k)$,  
the resulting absolute spectrum is

$$
|B_M(e^{j\Omega})| = \left|\mathcal{F}\{b_M(k)\}\right|,
$$

which is obtained by

1. stretching the frequency axis of the original spectrum $|B(e^{j\Omega})|$ by factor $M$  
   (i.e. center frequencies and bandwidth are multiplied by $M$),  
2. continuing the resulting spectrum $2\pi$-periodically,  
3. representing only the section $0 \leq \Omega \leq 2\pi$.  

Thus, the center frequencies $\Omega_m$ and $\Omega'_m$ are multiplied by $M$  
and then mapped into the frequency range of interest by a modulo-$2\pi$ operation.  

---

### Table of center frequencies for downsampling factors

| $M$ | 9 | 10 | 11 | 12 |
|-----|---|----|----|----|
| $\Omega_{m,M} = M\Omega_m$ | $9/5 \cdot 2\pi$ | $10/5 \cdot 2\pi$ | $11/5 \cdot 2\pi$ | $12/5 \cdot 2\pi$ |
| $\Omega_{m,M}\ \mathrm{mod}\ 2\pi$ | $4/5 \cdot 2\pi$ | $0$ | $1/5 \cdot 2\pi$ | $2/5 \cdot 2\pi$ |
| $\Omega'_{m,M} = M\Omega'_m$ | $36/5 \cdot 2\pi$ | $40/5 \cdot 2\pi$ | $44/5 \cdot 2\pi$ | $48/5 \cdot 2\pi$ |
| $\Omega'_{m,M}\ \mathrm{mod}\ 2\pi$ | $1/5 \cdot 2\pi$ | $0$ | $4/5 \cdot 2\pi$ | $3/5 \cdot 2\pi$ |
| Remark | reversed, no aliasing | original, aliasing | original, no aliasing | reversed, aliasing |
