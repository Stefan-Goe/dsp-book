### Exercise: Traditional spectrum estimation using the correlogram method

The Blackman–Tuckey algorithm yields an estimation $\hat{S}_{XX}^{BT}(\mathrm{e}^{j\Omega})$ of the power density spectrum $S_{XX}(\mathrm{e}^{j\Omega})$ based on the autocorrelation $r_{XX}(\kappa)$.

This includes:

* a biased estimation of the ACF series
  $\hat{r}_{XX}(\kappa)$ for $|\kappa| \leq M-1$
  using Rader’s algorithm,

* a multiplication of the ACF series with a window of length $2M-1$:

$$
\hat{r}_{XX}^{BT}(\kappa)

\hat{r}_{XX}(\kappa)\cdot f_M(\kappa)
$$

* a DFT to calculate the power density spectrum:

$$
\hat{S}_{X_iX_i}^{BT}(\mathrm{e}^{j\Omega})

\mathrm{DFT}{\hat{r}_{XX}^{BT}(\kappa)}
$$

Estimate the power density spectra of both processes of exercise dsv8a02e using the Blackman–Tuckey algorithm with a rectangular or triangular window function for $M = 8$, $32$, and $128$.

Why is the spectrum estimation of the MA model remarkably good for $M = 8$?

MATLAB hints:

$rxx = lrader(M, x)$  estimation using Rader’s algorithm $triang(2*M-1)$   triangle (Bartlett) window of length $2*M-1$

```python
# BEGIN SOLUTION
% Modeling noise processes
N = 2^12;           % number of samples to set the system in steady-state
N_pre = 500;        % number of samples to set the system in steady-state

n = randn(1,N+N_pre);  

h1 = [1 -4 6 -4 +1]; x1_all = filter(h1,1,n); x1 = x1_all(N_pre+1:length(x1_all));
h2 = [1 0.8];        x2_all = filter(1,h2,n); x2 = x2_all(N_pre+1:length(x2_all));

% Compute "true" power spectral density for comparison
NFFT = 2^12;  
Sx1x1 = abs(fft(h1,NFFT)).^2;                axis_Sx1=[0 1 0 1.35*max(Sx1x1)];
Sx2x2 = ones(1,NFFT)./abs(fft(h2,NFFT)).^2;  axis_Sx2=[0 1 0 1.35*max(Sx2x2)];
fT = 0:1/NFFT:1-1/NFFT;

% Estimate power spectral density for M=8, 32, 128
win = 'triang';

Sxx_x1_8   = lblack(x1,8,win,NFFT);
Sxx_x1_32  = lblack(x1,32,win,NFFT); 
Sxx_x1_128 = lblack(x1,128,win,NFFT);

Sxx_x2_8   = lblack(x2,8,win,NFFT);
Sxx_x2_32  = lblack(x2,32,win,NFFT); 
Sxx_x2_128 = lblack(x2,128,win,NFFT); 

% # Graphical Output # %
figure;
plot(fT,Sx1x1,'--'); 
hold on;  
plot(fT,Sxx_x1_8); 
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X1X1}(exp(j\cdot \Omega)), M=8 (Triangular)');
axis(axis_Sx1);  

figure;   
plot(fT,Sx1x1,'--'); 
hold on;  
plot(fT,Sxx_x1_32); 
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X1X1}(exp(j\cdot \Omega)), M=32 (Triangular)'); 
axis(axis_Sx1); 
 
figure;  
plot(fT,Sx1x1,'--');
hold on; 
plot(fT,Sxx_x1_128);
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X1X1}(exp(j\cdot \Omega)), M=128 (Triangular)'); 
axis(axis_Sx1); 

input('....Press RETURN to continue:');

figure;   
plot(fT,Sx2x2,'--'); 
hold on;  
plot(fT,Sxx_x2_8); 
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X2X2}(exp(j\cdot \Omega)), M=8 (Triangular)');
axis(axis_Sx2);  

figure;
plot(fT,Sx2x2,'--'); 
hold on; 
plot(fT,Sxx_x2_32); 
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X2X2}(exp(j\cdot \Omega)), M=32 (Triangular)'); 
axis(axis_Sx2); 

figure;  
plot(fT,Sx2x2,'--');
hold on; 
plot(fT,Sxx_x2_128); 
xlabel('\Omega/2\pi = f \cdot T'); 
ylabel('S_{XX}(exp(j\cdot \Omega))');
title('Blackman-Tukey Estimate of S_{X2X2}(exp(j\cdot \Omega)), M=128 (Triangular)'); 
axis(axis_Sx2); 
# END SOLUTION
```

### Solution: Traditional spectrum estimation using the correlogram method

For different values of $M$, the following figures show a Blackman–Tukey estimation $\hat{S}{XX}^{BT}$ of the power density spectra $S{XX}$ using $N = 2^{12}$ samples of the noise processes $X_1(k)$ and $X_2(k)$. A triangular window of length $2M-1$ is used. The true power spectral density $S_{\pi X_i \pi X_i}(\mathrm{e}^{j\Omega})$ is plotted as a dashed line.

From the first image of the first figure, we can see that the true power density spectrum is well approximated for $M = 8$.
This is because the true autocorrelation function $r_{XX}$ vanishes outside the interval $|\kappa| \leq m$ for an MA model of order $m = 4$.
In our example, the MA model $H_1(z)$ has order $m = 4$, so the true ACF $r_{XX}$ is restricted to the interval $|\kappa| \leq 4$.

In contrast, the estimated ACF $\hat{r}{XX}$ does not vanish outside this interval, but only assumes small values depending on the data length $N$.
The variance of the estimation can be improved by damping or completely suppressing these disturbing ACF values $\hat{r}{XX}$ with $|\kappa| > 4$ by windowing.

This represents the optimal case for $M = 8$, since the resulting window of length 15 sets all ACF values $\hat{r}_{XX}$ with $|\kappa| > 7$ to zero.
However, multiplication with a short window in the time domain corresponds to a convolution with a wide spectral window in the frequency domain, which increases the bias for small values of $M$ compared to larger ones.

<img src="./pngs/ldsv8a03_x11.png" width="250">
<img src="./pngs/ldsv8a03_x12.png" width="250">
<img src="./pngs/ldsv8a03_x13.png" width="250">

We observe that the bias decreases with increasing values of $M$. However, the variance increases proportionally to $\sum_\kappa f_M^2(\kappa)$. For the triangular window, the approximation

$$
\sum_\kappa f_M^2(\kappa) \approx \frac{2}{3} M
$$

holds, which implies that the variance is proportional to $M$.

In contrast to the MA model $H_1(z)$, the AR model $H_2(z)$ has an infinitely long impulse response and therefore an infinitely extended ACF $r_{XX}$. Windowing thus distorts the estimated power density spectrum more strongly in this case.

**Remark:** If a rectangular window is used, strong oscillations occur at low and high frequencies for small values of $M$.
