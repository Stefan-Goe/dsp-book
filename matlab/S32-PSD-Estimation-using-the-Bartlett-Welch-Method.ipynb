## Exercise: Traditional Spectrum Estimation — Bartlett & Welch (8.4)

To decrease the variance of the periodogram  
$\mathrm{Per}_N^X(\Omega)$, both the Bartlett and the Welch algorithms
compute the mean of multiple ($K$) periodograms of subsequences
with length

$$
L = \frac{N}{K}.
$$

An extension of Bartlett’s algorithm is provided by Welch’s method,
which additionally applies windowing to each subsequence.
Bartlett’s algorithm is included as a special case of Welch’s
algorithm by using a rectangular window.

Compare the realizations of both processes of length

$$
N = 2^{10}
$$

from Exercise `dsv8a02e`.

Analyze the influence of  
- different numbers of periodograms used to compute the mean  
  ($K = 1, 4, 16, 64$), and  
- different window functions (rectangular, Hamming)  

on the spectrum estimation.

---

### MATLAB Hints

```matlab
Sxx = lwelch(x, K, windowtype)
```

````python
# BEGIN SOLUTION
N     = 2^10; 
N_pre = 500; 

n  = randn(1,N+N_pre);             

h1 = [1 -4 6 -4 +1];  x1_all = filter(h1,1,n); x1=x1_all(N_pre+1:length(x1_all));
h2 = [1 0.8];         x2_all = filter(1,h2,n); x2=x2_all(N_pre+1:length(x2_all));

% "True" power spectral density for comparison
NFFT = 2^10;                    
Sx1x1 = abs(fft(h1,NFFT)).^2;                axis_Sx1=[0 1 0  1.5*max(Sx1x1)];
Sx2x2 = ones(1,NFFT)./abs(fft(h2,NFFT)).^2;  axis_Sx2=[0 1 0  1.5*max(Sx2x2)];
fT = 0:1/NFFT:1-1/NFFT;
v_K = [1 4 16 64];

% BARTLETT for x1
win='boxcar';                                       
for k=v_K
    S_xx=lwelch(x1,k,win,NFFT);
    
    % # Graphical Output # %
    figure; 
    plot(fT,S_xx); 
    hold on;
    plot(fT,Sx1x1,'--');
    axis(axis_Sx1);
    xlabel('\Omega/2\pi = f \cdot T');
    ylabel('S_{XX}(exp(j\cdot \Omega))');
    title(sprintf('Bartlett Estimate of S_{X1X1}(exp(j*Om.); K=%d and L=%d/K',k, N));
    text(0.73,900,'== Periodogram');
end;

input('......Press RETURN to continue:');

% WELCH method with Hamming for x1
win='hamming';                       
for k=v_K
    S_xx=lwelch(x1,k,win,NFFT);

    % # Graphical Output # %
    figure;  
    plot(fT,S_xx);  
    hold on; 
    plot(fT,Sx1x1,'--'); 
    axis(axis_Sx1);
    xlabel('\Omega/2\pi = f\cdot T');
    ylabel('S_{XX}(exp(j\cdot \Omega))');
    title(sprintf('Welch Estimate of S_{X1X1}(exp(j*Om.)); K=%d and L=%d/K (Hamming)',k, N));
end;

input('......Press RETURN to continue:');

% BARTLETT for x2
win='boxcar';                                       
for k=v_K
    S_xx=lwelch(x2,k,win,NFFT);
    
    % # Graphical Output # %
    figure; 
    plot(fT,S_xx);  
    hold on;
    plot(fT,Sx2x2,'--');
    axis(axis_Sx2); 
    xlabel('\Omega/2\pi = f\cdot T');
    ylabel('S_{XX}(exp(j \cdot \Omega))');
    title(sprintf('Bartlett Estimate of S_{X2X2}(exp(j*Om.)); K=%d and L=%d/K',k, N));
    text(0.73,900,'== Periodogram');
end;

input('......Press RETURN to continue:');

% WELCH method with Hamming for x2
win='hamming';                       
for k=v_K
    S_xx=lwelch(x2,k,win,NFFT);
    
    % # Graphical Output # %
    figure; 
    plot(fT,S_xx); 
    hold on;
    plot(fT,Sx2x2,'--');
    axis(axis_Sx2); 
    xlabel('\Omega/2\pi = f\cdot T');
    ylabel('S_{XX}(exp(j\cdot \Omega))');
    title(sprintf('Welch Estimate of S_{X2X2}(exp(j*Om.)); K=%d and L=%d/K (Hamming)',k, N));
end; 
% #####  EOF  #####

# END SOLUTION
````
