# Quantisierung von Filterkoeffizienten

In Abschnitt 4.4.1 wurde während der Analyse der **Parameterquantisierung** gezeigt, dass die Pole empfindlicher werden, je mehr Pole vorhanden sind und je näher sie zueinander liegen. Bei den Nullstellen ist ein ähnliches Verhalten zu beobachten. Daher ist die **dritte kanonische Form** (Übung 4.4.1) von großer Bedeutung für die Realisierung höherer Ordnung von Systemen.

Diese Aufgabe zeigt ein Beispiel, wie die **Quantisierung von Filterkoeffizienten** die Systemfunktion $H(z)$ beeinflusst.
Die Funktion `lfxquant()` implementiert einen linearen Quantisierer:

$$
-1 \leq [v]_Q \leq 1 - 2^{-l+1},
$$

wobei $l$ die Anzahl der Bits ist und $v$ die zu quantisierende Zahl darstellt.
Um die Koeffizienten eines digitalen Filters zu quantisieren, müssen die Koeffizienten **normalisiert** werden.
Die Funktion `[aq, n] = lcoefrnd(a, l)` berechnet den Vektor der quantisierten Koeffizienten `aq` sowie den Normalisierungsfaktor `n`.
