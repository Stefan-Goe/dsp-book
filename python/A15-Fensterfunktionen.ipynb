
# Fensterfunktionen

Im Gegensatz zu einem rekursiven Filter, dessen Entwurf auf der Transformation von kontinuierlichen IIR-Systemen in diskrete IIR-Systeme basiert, beruht der Entwurf nicht-rekursiver (FIR-)Systeme auf der Approximation der gewünschten Frequenzantwort des diskreten Systems. Auf der Basis einer konstanten Gruppendelay werden meist **linearphasige** Filter angewendet, deren Entwurf im Folgenden behandelt wird.

FIR-Systeme mit linearer Phase basieren auf der Fensterung idealer Tiefpassfilter. Um die Linearphasenbedingung einzuhalten, müssen die Fensterfunktionen symmetrisch sein.

Die am häufigsten verwendeten Fensterfunktionen sind die sogenannten Cosinus-Fenster, die folgende Form haben:

$$
f(k) =
\begin{cases}
a - b \cdot \cos\!\left( \frac{2\pi k}{m} \right) + c \cdot \cos\!\left( \frac{4\pi k}{m} \right), & 0 \leq k \leq m \\
0, & \text{sonst.}
\end{cases}
$$

| Fenster       | Python / SciPy Funktion      | a    | b    | c    |
|---------------|-----------------------------|------|------|------|
| Rechteck      | `scipy.signal.boxcar`       | 1.0  | 0.0  | 0.0  |
| von Hann      | `scipy.signal.hann`         | 0.5  | 0.5  | 0.0  |
| Hamming       | `scipy.signal.hamming`      | 0.54 | 0.46 | 0.0  |
| Blackman      | `scipy.signal.blackman`     | 0.42 | 0.5  | 0.08 |

Ein weiteres Fenster ist das Dreiecks- oder Bartlett-Fenster (`scipy.signal.bartlett`).
