$$
x(k) = 1 - \cos\left(\frac{2\pi k}{64}\right), \quad k = 0, \dots, 63
$$

$$
x_0(k) =
\begin{cases}
x(k), & k = 0 \dots 63 \\
0,    & k = 64 \dots 127
\end{cases}
$$
