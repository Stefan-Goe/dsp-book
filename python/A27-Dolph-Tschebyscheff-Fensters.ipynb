### Aufgabe: Dolph-Chebyshev-Fenster

Der Einfluss des **Leck-Effekts** auf die Ergebnisse einer DFT kann durch die Verwendung von Fensterfunktionen minimiert werden (siehe Übung~\ref{dsv7a07e}). Traditionelle Fensterfunktionen sind im **Zeitbereich** als endliche Folgen definiert.

Eine Ausnahme bildet das **Dolph-Chebyshev-Fenster**, bei dem die Fensterfunktion im **Frequenzbereich** definiert ist.

Zeigen Sie anhand einiger Beispiele, dass die **Rücktransformation** dieser Spektralfunktionen zu einer **endlichen Folge** der Länge $N$ im Zeitbereich führt.

Entwickeln Sie eine **Python-Funktion** `lcheby(N, Ω_s, ΔΩ)` mit folgenden Parametern:

* Fensterlänge `N`
* Sperrband-Grenzfrequenz `Ω_s`
* Abtastabstand `ΔΩ`

Die Funktion soll die **Spektralfunktion**

$$
F^{DT}(e^{jn\Delta\Omega})
$$

gemäß (7.4.26) und (7.4.25) berechnen und zudem die **kausale Zeitbereichsfolge**

$$
f(k) = \mathcal{F}^{-1} { F^{DT}(e^{jn\Delta\Omega}) }
$$

des Dolph-Chebyshev-Fensters erzeugen.
