Linear filters are subdivided by two components: Nonrecursive filters and Recursive filters. In this exercise, we will focus on designing of nonrecursive filters using some windowing schemes. 

Nonrecursive filters have finite impulse responses. The complete filter design by windowing an impulse response of a time shifted, ideal low/high pass filter is implemented in Python function firwin(). 

 The function b = firwin(numtaps,cutoff, pass_zero=false) yields the corresponding high pass with a cut-off frequency Wn. 
 b = firwin(numtaps,cutoff) generates m + 1 filter coefficients of a low-pass with order m and cut-off frequency Wn.
