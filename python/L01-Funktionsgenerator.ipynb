## Funktionsgenerator (Kapitel 2.1)

In dieser Übung soll ein Funktionsgenerator programmiert werden, der ein Signal mit der Frequenz $f_0$  
bei einer Abtastfrequenz von $f_\mathrm{A}$ erzeugt.  
Dieser Generator soll in der Lage sein, sowohl ein Sinus- als auch ein Sägezahnsignal zu generieren.

Für ein zeitkontinuierliches System wird üblicherweise mit der Kreisfrequenz  
$\omega_0 = 2 \pi f_0$ gerechnet.  
Damit ergibt sich ein zeitkontinuierliches Sinussignal zu

$$
x_\mathrm{sin,K}(t) = \sin(\omega_0 \cdot t)
$$

In einem zeitdiskreten System wird das Sinussignal mit Abtastperioden der Länge  
$T = 1 / f_\mathrm{A}$ abgetastet, sodass die Zeit  
$t = T \cdot k$ durch die Indizes $k = 0, 1, 2, ...$ beschrieben werden kann.  
Daraus ergibt sich

$$
x_\mathrm{sin}(k) = x_\mathrm{sin,K}(t)|_{t = kT} = \sin(\omega_0 \cdot T \cdot k)
$$

Für zeitdiskrete Systeme ist es sinnvoll, die **normierte Kreisfrequenz** $\Omega_0$ einzuführen,  
die keine Einheit besitzt:

$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T \\
         & =  & 2 \pi f_0 \cdot T \\
         & =  & 2 \pi f_0 / f_\mathrm{A}
\end{array}
$$

Diese Definition wird in den folgenden Aufgaben verwendet, da sie die Behandlung digitaler Systeme  
ohne Einschränkung der Abtastfrequenz ermöglicht.  
Damit gilt:

$$
x_\mathrm{sin}(k) = \sin(\Omega_0 \cdot k)
$$

### a)

Erzeuge einen Vektor `x` mit 60 Abtastwerten eines Sinussignals mit der Frequenz  
$f_0 = 400\,\text{Hz}$ bei einer Abtastfrequenz von $8\,\text{kHz}$.

Initialisiere dazu einen Koeffizientenvektor  
`k = np.arange(60)`,  
der den Zeitindex des abgetasteten Signals $x(k)$ darstellt.  
Verwende **Pythons** Funktion `numpy.sin()`.

Stelle das Signal mit der Funktion `matplotlib.pyplot.stem()` dar.  
Wie groß ist die Periode des abgetasteten Signals?

### b)

Ändere die Frequenz des abgetasteten Sinussignals auf  
$f_0 = 960\,\text{Hz}$
und stelle das resultierende Signal erneut mit `matplotlib.pyplot.stem()` dar.

Wie groß ist nun die Periode des abgetasteten Signals?

### c)

Erzeuge nun ein Sägezahnsignal mit der Frequenz  
$f_0 = 960\,\text{Hz}` bei einer Abtastfrequenz von $f_\mathrm{A} = 8\,\text{kHz}$.

Das Sägezahnsignal soll einen Amplitudenbereich von $0...1$ besitzen.  
Zur Berechnung verwende **Pythons** Funktion `numpy.mod()`.  
Die Normierung der Signalfrequenz $f_0$ durch die Abtastfrequenz $f_\mathrm{A}$  
kann analog zu oben erfolgen.  
Beachte dabei, dass du die Periode der `mod()`-Funktion selbst festlegst –  
im Gegensatz zur $2\pi$-Periodizität der `sin()`-Funktion.

Stelle das Sägezahnsignal mit `matplotlib.pyplot.stem()` dar und vergleiche es mit dem Sinussignal.  
Ist die Signalfrequenz korrekt?

### d)

Erzeuge einen Vektor sowohl für ein Sinus- als auch für ein Sägezahnsignal mit  
$f_0 = 1\,\text{kHz}` bei einer Abtastfrequenz von $f_\mathrm{A} = 8\,\text{kHz}$.

Die Signale sollen über eine Soundkarte wiedergegeben werden und eine Dauer von  
$d = 2\,\text{s}` haben.  
Wie viele Abtastwerte sind dafür erforderlich?

Spiele die Signale mit dem **Python**-Befehl  
`sounddevice.play(x, fa)` ab.

```python
import numpy as np
import matplotlib.pyplot as plt
import sounddevice as sd
# BEGIN SOLUTION
fa = 8000  # Setting the sampling rate

# ##### Subtask a: Sine Wave #####
# Calculation of the signal vector
k = np.arange(0, 60)  # Creating a vector of length 60 with entries from 0 to 59
f0 = 400  # Signal frequency
Omega0 = 2 * np.pi * f0 / fa  # Normalized frequency
xsin = np.sin(Omega0 * k)  # Sine wave vector

# Plotting
plt.figure()
plt.stem(k, xsin)
plt.xlabel('k')
plt.ylabel('x_{sin}(k)')
plt.title('Sine Wave with f0 = 400 Hz')
plt.show()

# ##### Subtask b: Sine Wave, Altered Period #####
# Calculation of the signal vector
f0 = 960  # Signal frequency
Omega0 = 2 * np.pi * f0 / fa  # Normalized frequency
xsin = np.sin(Omega0 * k)  # Sine wave vector

# Plotting
plt.figure()
plt.stem(k, xsin)
plt.xlabel('k')
plt.ylabel('x_{sin}(k)')
plt.title('Sine Wave with f0 = 960 Hz')
plt.show()

# ##### Subtask c: Sawtooth Wave #####
xsaw = np.mod(f0 / fa * k, 1)  # Sawtooth wave vector

# Plotting
plt.figure()
plt.stem(k, xsaw)
plt.xlabel('k')
plt.ylabel('x_{saw}(k)')
plt.title('Sawtooth Wave')
plt.show()

# ##### Subtask d: Denormalized Signals, Sound Generation #####
f0 = 1000  # Signal frequency
d = 2  # Playback duration in seconds
w = d * fa  # Number of samples
k = np.arange(0, w + 1)  # Coefficient vector of the time index
xsin = np.sin(2 * np.pi * f0 / fa * k)  # Sine wave vector
xsaw = np.mod(f0 / fa * k, 1)  # Sawtooth wave vector

# Playback of the signals
print('Listening to: 2s Sine Wave')
sd.play(xsin, fa)
sd.wait()

print('Listening to: 2s Sawtooth Wave')
sd.play(xsaw, fa)
sd.wait()
# END SOLUTION
```

Die Periode des zeitkontinuierlichen Signals  
$x_\text{sin,K}(t) = \sin(2\pi f_0 \cdot t)$  
beträgt $T_0 = 1 / f_0$.  
Da die Anzahl der Abtastwerte pro Periode des zeitkontinuierlichen Signals eine ganze Zahl ist,  
muss das zeitdiskrete Signal eine Periodenlänge von

$$
\begin{array}{rcl}
k_p &=& T_0 / T \\
    &=& f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 400\,\text{Hz} \\
    &=& 20
\end{array}
$$

Werten haben.

Nach Erhöhung der Signalfrequenz auf $f_0 = 960\,\text{Hz}$ stimmen die Perioden des zeitdiskreten und des zeitkontinuierlichen Signals nicht mehr überein.

<figure style="text-align:center; margin: 0 auto; width: fit-content;">

Das Signal wiederholt sich nach 25 Abtastwerten. Schlichte technische Diagrammdarstellung.">
    <figcaption><em>Abbildung: Abgetastetes Sinussignal bei 960 Hz und 8 kHz Abtastrate mit einer Periode von 25 Abtastwerten.</em></figcaption>
</figure>

Die Periode des zeitdiskreten Signals tritt nach

$$
\begin{array}{rcl}
k_p &=& L \cdot T_0 / T \\
    &=& L \cdot f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 960\,\text{Hz} \\
    &=& L \cdot 8.\overline{3} \\
    &\Downarrow& L = 3 \\
    &=& 25
\end{array}
$$

Abtastwerten auf.  
Dabei ist $L = 3$ die Anzahl der Perioden des kontinuierlichen Signals, nach denen sich die Periodizität des abgetasteten Signals einstellt.
