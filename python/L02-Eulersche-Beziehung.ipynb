## Aufgabe: Die Eulersche Gleichung

Die Eulersche Gleichung beschreibt die Verbindung zwischen der Exponentialfunktion (Reihe)
und den trigonometrischen Funktionen (Reihen).

$$
e^{j\phi k} = \cos(\phi k) + j \sin(\phi k)
$$

bzw.

$$
\cos(\phi k) = \frac{1}{2}\left(e^{j\phi k} + e^{-j\phi k}\right)
$$

$$
\sin(\phi k) = \frac{1}{2j}\left(e^{j\phi k} - e^{-j\phi k}\right)
$$

Diese Beziehung soll in dieser Aufgabe untersucht werden.

- **a)** Stelle ein sinusförmiges Signal von 0 bis $6\pi$ mit 32 Abtastwerten pro Periode dar.

- **b)** Stelle die zugehörige komplexe Exponentialreihe in der komplexen Ebene dar (Imaginärteil = Y-Achse, Realteil = X-Achse) sowie ihren Realteil, Imaginärteil und Betrag in Abhängigkeit vom Zeitindex $k$.

- **c)** Verwende die Eulersche Gleichung und erzeuge eine sinusförmige Folge mithilfe komplexer Exponentialfolgen. Vergleiche das Ergebnis mit der sinusförmigen Folge aus Teilaufgabe a).

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
# ##### Subtask a: Sine Signal #####
k = np.arange(0, 96)
x = np.sin(2 * np.pi / 32 * k)

# Plotting
plt.figure()
plt.stem(k / 32, x, basefmt=" ")
plt.plot(k / 32, x, 'y')
plt.plot([0, 3], [0, 0], 'k')
plt.title('Sine Function x(k)')
plt.xlabel('k/32')
plt.ylabel('x(k)')
plt.show()

# ##### Subtask b: Real and Imaginary Part #####
y = np.exp(1j * 2 * np.pi / 32 * k)

# Plotting
# Complex Plane
plt.figure()
plt.plot(np.real(y), np.imag(y), 'bo-')
plt.plot([-1, 1], [0, 0], 'k')
plt.plot([0, 0], [-1, 1], 'k')
plt.axis('square')
plt.title('Complex Exponential Sequence y(k)')
plt.xlabel('Real part')
plt.ylabel('Imaginary part')
plt.show()

# Real part
plt.figure()
plt.plot(k / 32, np.real(y))
plt.plot([0, 3], [0, 0], 'k')
plt.xlabel('k/32')
plt.ylabel('Re{y(k)}')
plt.show()

# Imaginary part
plt.figure()
plt.plot(k / 32, np.imag(y))
plt.plot([0, 3], [0, 0], 'k')
plt.xlabel('k/32')
plt.ylabel('Im{y(k)}')
plt.show()

# Magnitude
plt.figure()
plt.plot(k / 32, np.abs(y))
plt.plot([0, 3], [0, 0], 'k')
plt.axis([-0.05, 3, 0, 1.1])
plt.xlabel('k/32')
plt.ylabel('|y(k)|')
plt.show()

# ##### Subtask c: Euler's Relationship => Sine #####
z = -1j * 0.5 * (np.exp(1j * 2 * np.pi / 32 * k) - np.exp(-1j * 2 * np.pi / 32 * k))

# Plotting
plt.figure()
plt.stem(k / 32, np.real(z), basefmt=" ")
plt.plot(k / 32, x, 'b-')
plt.plot([0, 3], [0, 0], 'k')
plt.title('Sine Function using Euler')
plt.xlabel('k/32')
plt.ylabel('Amplitude')
plt.show()
# END SOLUTION
```
