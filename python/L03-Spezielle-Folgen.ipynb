## Übung: Spezielle Reihen

Diese Übung demonstriert einige grundlegende Reihen für die digitale Signalverarbeitung.

* **a)** Erzeuge eine Impulsreihe
  $$
  x(k) = 1.2\cdot \delta(k-4), \qquad -5 \leq k \leq 10
  $$
  mit
  $$
  \delta(k):=\begin{cases}
  1 & k=0\
  0 & k\neq 0
  \end{cases}
  $$
  und stelle diese Reihe mit der Funktion `stem()` dar.

* **b)** Stelle die folgenden gaußschen und laplaceschen Reihen mit der Funktion `stem()` dar.
  Warum ist es ungünstig, den Befehl `plot()` zur Darstellung der Folgen zu verwenden?

 | Nr. | Signaldefinition | Wertebereich | Parameter | Beschreibung |
|:---:|:-----------------|:--------------|:-----------|:--------------|
| 1 | $x_1(k) = e^{-\alpha k^2}$ | $-20 \leq k \leq 20$ | $\alpha = 0.01$ | Gaussian series |
| 2 | $x_2(k) = e^{-\beta\,\|k\|}$ | $-20 \leq k \leq 20$ | $\beta = 0.2$ | Laplace series |
| 3 | $x_3(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{4})$ | $-30 \leq k \leq 30$ | $\beta = 0.1$ | mod. Laplacian series |
| 4 | $x_4(k) = e^{-\beta\,\|k\|} \cdot \cos(2\pi\, \frac{k}{80})$ | $-600 \leq k \leq 600$ | $\beta = 0.005$ | — |


```python
import numpy as np
import matplotlib.pyplot as plt 
# BEGIN SOLUTION
# ##### Subtask a: Impulse Sequence #####
k_a = np.arange(-5, 11)      # Initialize the coefficient vector k with values from -5 to 10
x = np.zeros(len(k_a))       # Initialize x as a ROW vector with 16 zeros
x[4 + 5] = 1                 # The Dirac impulse '1' should appear at temporal index k=4
                             # The temporal index k=-5 corresponds to Python index 0,
                             # so for k=4, you need to use index 4 - (-5) = 9
x = 1.2 * x                  # Multiply the sequence by the factor 1.2


# Plotting
plt.figure()
plt.stem(k_a, x)
plt.xticks(np.arange(-4, 11, 2))
plt.title('x(k)=1.2 δ(k-4)')
plt.xlabel('Sample points k')
plt.ylabel('Amplitude')
plt.show()

# ##### Subtask b: Gaussian and Laplace Sequences #####
# Coefficient vectors
k_b1 = np.arange(-20, 21)
k_b2 = np.arange(-20, 21)
k_b3 = np.arange(-30, 31)
k_b4 = np.arange(-600, 601)

# Alpha factor for Gaussian sequence
alpha = 0.01

# Beta factors for Laplace sequences
beta_2 = 0.2
beta_3 = 0.1
beta_4 = 0.005

# Compute the sequences
x1 = np.exp(-alpha * k_b1**2)
x2 = np.exp(-beta_2 * np.abs(k_b2))
x3 = np.exp(-beta_3 * np.abs(k_b3)) * np.cos(2 * np.pi / 4 * k_b3)
x4 = np.exp(-beta_4 * np.abs(k_b4)) * np.cos(2 * np.pi / 80 * k_b4)

# Plotting
# Gaussian sequence
plt.figure()
plt.plot(k_b1, x1)
plt.stem(k_b1, x1, 'k', basefmt=" ")
plt.title('Gaussian Sequence x_1(k)')
plt.xlabel('k')
plt.ylabel('Amplitude')
plt.show()

# Laplace sequence
plt.figure()
plt.plot(k_b2, x2)
plt.stem(k_b2, x2, 'k', basefmt=" ")
plt.title('Laplace Sequence x_2(k)')
plt.xlabel('k')
plt.ylabel('Amplitude')
plt.show()

# Modified Laplace sequence beta = 0.1
plt.figure()
plt.plot(k_b3, x3)
plt.stem(k_b3, x3, 'k', basefmt=" ")
plt.grid()
plt.title('Modulated Laplace Sequence x_3(k)')
plt.xlabel('k')
plt.ylabel('Amplitude')
plt.show()

# Modified Laplace sequence beta = 0.005
plt.figure()
plt.plot(k_b4, x4, 'k')
plt.grid()
plt.title('Modulated Laplace Sequence x_4(k)')
plt.xlabel('k')
plt.ylabel('Amplitude')
plt.show()
# END SOLUTION
```
