### c) Kommutativität der Faltung
Der Faltungsoperator ist **kommutativ**, d.h. man kann **Signal** und **System** vertauschen.

```python
import numpy as np
# BEGIN SOLUTION
% #####  Teilaufgabe a:  Eigene convmtx Funktion schreiben #####
def lconvmtx(h, n):
    """
    Generate a convolution matrix for a given impulse response vector.

    Parameters:
    - h: numpy array or list
        The impulse response vector.
    - n: int
        The number of columns in the convolution matrix.

    Returns:
    - H: numpy array
        The convolution matrix.

    Example:
    >>> h = [1, 2, 3]
    >>> n = 4
    >>> lconvmtx(h, n)
    array([[1., 0., 0., 0.],
           [2., 1., 0., 0.],
           [3., 2., 1., 0.]])
    """
    h = np.asarray(h).flatten()  # Ensure h is a column vector
    H = np.zeros((len(h) + n - 1, n))  # Create a matrix of zeros
    for l in range(n):
        H[l:l+len(h), l] = h
    
    return H

% #####  Teilaufgabe b:  Faltungsmatrix testen #####
h = [1 2 3 4 5].';           % Impulsantwort des Systems
x = sin(2*pi*1/10*[0:19].'); % Testsignal

yhx = conv(h,x);             % direkt ausgefuehrte Faltung

Hh = lconvmtx(h,length(x));  % Faltung mit Hilfe der Faltungsmatrix
yHhx = Hh*x;
differenz = yhx-yHhx         % Vergleich durch Diferenzbildung

% #####  Teilaufgabe c:  Vertauschung System <> Signal #####
disp('1.Variante: Faltungsmatrix mit h gebildet');

Hh = lconvmtx(h,length(x));
Hhsize = size(Hh)           % 24 x 20 Matrix
Hhsizeprod = prod(Hhsize)   % Matrix hat 480 Eintraege
convHhx = Hh*x;

disp('2.Variante: Faltungsmatrix mit x gebildet');
Hx = lconvmtx(x,length(h));
Hxsize = size(Hx)           % 24 x 5 Matrix
Hxsizeprod = prod(size(Hx)) % Matrix hat 120 Eintraege

yHxx = Hx*h;
# END SOLUTION
```

### c) Effizienz verschiedener Faltungsmatrizen

Es sei  

- $n_x$ = Länge des Vektors $\mathbf{x}$  
- $n_h$ = Länge des Vektors $\mathbf{h}$  

---

Wird die Faltungsmatrix aus $\mathbf{h}$ aufgebaut, so hat sie die Dimensionen  

$$
(n_h + n_x - 1) \times n_x
$$  

und damit  

$$
n_1 = n_x n_h + n_x^2 - n_x
$$  

Einträge.

---

Wird die Faltungsmatrix aus $\mathbf{x}$ aufgebaut, so hat sie die Dimensionen  

$$
(n_x + n_h - 1) \times n_h
$$  

und damit  

$$
n_2 = n_x n_h + n_h^2 - n_h
$$  

Einträge.

---

Für den Fall $n_x > n_h$ und $n_h > 0$ gilt:  

$$
\begin{aligned}
n_1 &> n_2 \\
n_x n_h + n_x^2 - n_x &> n_x n_h + n_h^2 - n_h \\
n_x^2 - n_x &> n_h^2 - n_h
\end{aligned}
$$

Um die Anzahl der Einträge in der Faltungsmatrix zu minimieren, sollte die Matrix also aus der **längeren Sequenz** aufgebaut werden.
