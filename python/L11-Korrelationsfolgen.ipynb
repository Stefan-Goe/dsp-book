# Aufgabe 2.14: Korrelationsreihe

Neben den Momenten aus Aufgabe~\ref{dsv2a11e} gibt es noch weitere charakteristische Parameter für stochastische Prozesse, nämlich *die Autokorrelation, Kreuzkorrelation* sowie *Auto-Kovarianz* bzw. *Kreuzkovarianz*.

Gemeinsames Merkmal dieser Funktionen ist, dass sie Werte für die Abhängigkeit der beobachteten Prozesse und ihrer Zufallsvariablen liefern. In dieser Aufgabe wird beispielhaft die Autokorrelationsreihe betrachtet. Die zugrundeliegenden stochastischen Prozesse werden als *ergodisch* angenommen, siehe Abschnitt (2.6). Ein Hauptproblem ist die Schätzung der theoretischen Autokorrelationsreihe (2.6.6):

$$
r_{XX}(\kappa) = E\{X^\ast(k)\, X(k+\kappa)\}
$$

basierend auf einer endlichen Anzahl von $N$ Stichproben.

In Python kann die Autokorrelationsreihe mit Funktionen wie `numpy.correlate()` oder `scipy.signal.correlate()` geschätzt werden.  
Ein Aufruf von `r = numpy.correlate(x, x, mode='full')` erzeugt aus dem Vektor `x` eine nicht normalisierte Autokorrelationsreihe nach der Formel

$$
\hat{r}_{XX}(\kappa) = \sum_{k=0}^{N-|\kappa|-1} x(k)\,x^{\ast}(k+\kappa)
$$

Im Gegensatz zu Gleichung (2.6.6) ist die Autokorrelationsreihe in Python konjugiert definiert. Deshalb sind die Ergebnisse stets konjugiert.

Eine normalisierte Autokorrelationsreihe kann durch Division mit verschiedenen Normalisierungsfaktoren realisiert werden. Die folgenden Normalisierungsoptionen entsprechen:

- biased:
$$
\hat{r}_{XX}(\kappa) = \frac{1}{N} \sum_{k=0}^{N-|\kappa|-1} x(k) \, x^*(k+\kappa)
$$

- unbiased:
$$
\hat{r}'_{XX}(\kappa) = \frac{1}{N - |\kappa|} \sum_{k=0}^{N - |\kappa| - 1} x(k) \, x^*(k+\kappa)
$$

- coeff:  
normiert die Korrelationsreihe $\hat{r}''_{XX}(\kappa)$ so, dass sie beim Wert $\kappa=0$ den Wert Eins annimmt.

**a)**  
Erzeuge eine weiße Gaußsche Rauschfolge der Länge \(N = 8000\).  
Schätze die Autokorrelationsfolge mit der Funktion `numpy.correlate()`  
und plotte die Ergebnisse.  
Verwende alle drei Normalisierungsoptionen: `biased`, `unbiased` und `coeff`.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import correlate
from scipy.fftpack import fft
from helper_functions import ldtft
# BEGIN SOLUTION
n = np.random.randn(800)
kappa = np.arange(-799, 800)

s_biased = correlate(n, n, 'full') / len(n)
s_unbiased = correlate(n, n, 'full') / (len(n) - np.abs(np.arange(-len(n) + 1, len(n))))
s_coeff = correlate(n, n, 'full') / np.max(correlate(n, n, 'full'))

# Plotting
# Autocorrelation Sequence: Biased
plt.figure()
plt.stem(kappa, s_biased)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Biased')
plt.show()

# Autocorrelation Sequence: Unbiased
plt.figure()
plt.stem(kappa, s_unbiased)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Unbiased')
plt.show()

# Autocorrelation Sequence: Normalized at time-lag Zero
plt.figure()
plt.stem(kappa, s_coeff)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Normalized at time-lag Zero')
plt.show()
# END SOLUTION
```

**b)**  
Die Fourier-Transformation der Autokorrelationsreihe ist das  
(periodische) Leistungsdichtespektrum.  
Verwende die Python-Funktion `ldtft()` aus der Datei `helper_functions.py`  
und plotte die Schätzung des Leistungsdichtespektrums.

```python
# BEGIN SOLUTION
n = np.random.randn(200)
s = correlate(n, n, 'full') / len(n)
S, W = ldtft(s, 512)

# Plotting
plt.figure()
plt.semilogy(W / np.pi, np.abs(S))  # normalized frequency
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel(r'$|S_{xx}(e^{j\Omega})|$')
plt.title('Power Spectral Density')
plt.show()
# END SOLUTION
```
