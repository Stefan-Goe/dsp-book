
# Fensterfunktionen

Im Gegensatz zu einem rekursiven Filter, dessen Entwurf auf der Transformation von kontinuierlichen IIR-Systemen in diskrete IIR-Systeme basiert, beruht der Entwurf nicht-rekursiver (FIR-)Systeme auf der Approximation der gewünschten Frequenzantwort des diskreten Systems. Auf der Basis einer konstanten Gruppendelay werden meist **linearphasige** Filter angewendet, deren Entwurf im Folgenden behandelt wird.

FIR-Systeme mit linearer Phase basieren auf der Fensterung idealer Tiefpassfilter. Um die Linearphasenbedingung einzuhalten, müssen die Fensterfunktionen symmetrisch sein.

Die am häufigsten verwendeten Fensterfunktionen sind die sogenannten Cosinus-Fenster, die folgende Form haben:

$$
f(k) =
\begin{cases}
a - b \cdot \cos\!\left( \frac{2\pi k}{m} \right) + c \cdot \cos\!\left( \frac{4\pi k}{m} \right), & 0 \leq k \leq m \\
0, & \text{sonst.}
\end{cases}
$$

| Fenster       | Python / SciPy Funktion      | a    | b    | c    |
|---------------|-----------------------------|------|------|------|
| Rechteck      | `scipy.signal.boxcar`       | 1.0  | 0.0  | 0.0  |
| von Hann      | `scipy.signal.hann`         | 0.5  | 0.5  | 0.0  |
| Hamming       | `scipy.signal.hamming`      | 0.54 | 0.46 | 0.0  |
| Blackman      | `scipy.signal.blackman`     | 0.42 | 0.5  | 0.08 |

Ein weiteres Fenster ist das Dreiecks- oder Bartlett-Fenster (`scipy.signal.bartlett`).

## Aufgaben

### a)

Berechne die oben genannten Fenstertypen für $m=21$ und $m=42$. Stelle die Signale im Zeitbereich sowie deren zeitdiskrete Fourier-Transformation dar. Vergleiche die verschiedenen Fenstertypen hinsichtlich ihrer Durchlassbanddämpfung, der Dämpfung des ersten Überschwingers und der Sperrbanddämpfung.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import windows, freqz
from helper_functions import ldtft

# BEGIN SOLUTION
window_length = 21

# Rectangle window
tltstr = 'Rectangle Window'
rw = windows.boxcar(window_length)
Rw, W = ldtft(rw, 512)

Rw[np.abs(Rw) == 0] = 1e-100

plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(Rw / np.max(Rw))))
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title(f'{tltstr}, m: {window_length}')

plt.figure()
plt.stem(np.arange(window_length), rw)
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title(tltstr)

# von Hann
tltstr = 'von Hann Window'
rh = windows.hann(window_length)
Rh, W = ldtft(rh, 512)

Rh[np.abs(Rh) == 0] = 1e-100

plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(Rh / np.max(Rh))))
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title(f'{tltstr}, m: {window_length}')

plt.figure()
plt.stem(np.arange(window_length), rh)
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title(tltstr)

# Hamming
tltstr = 'Hamming Window'
rhm = windows.hamming(window_length)
Rhm, W = ldtft(rhm, 512)

Rhm[np.abs(Rhm) == 0] = 1e-100

plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(Rhm / np.max(Rhm))))
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title(f'{tltstr}, m: {window_length}')

plt.figure()
plt.stem(np.arange(window_length), rhm)
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title(tltstr)

# Blackman
tltstr = 'Blackman Window'
rhb = windows.blackman(window_length)
Rhb, W = ldtft(rhb, 512)

Rhb[np.abs(Rhb) == 0] = 1e-100

plt.figure()
plt.plot(W / np.pi,20 * np.log10(np.abs(Rhb / np.max(Rhb))))
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title(f'{tltstr}, m: {window_length}')

plt.figure()
plt.stem(np.arange(window_length), rhb)
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title(tltstr)

# Bartlett
tltstr = 'Bartlett Window'
rhbt = windows.bartlett(window_length)
Rhbt, W = ldtft(rhbt, 512)

Rhbt[np.abs(Rhbt) == 0] = 1e-100

plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(Rhbt / np.max(Rhbt))))
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title(f'{tltstr}, m: {window_length}')

plt.figure()
plt.stem(np.arange(window_length), rhbt)
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title(tltstr)
# END SOLUTION
```

### b)

Das Kaiser-Fenster ist ein flexibles Fenster, das wie folgt definiert ist:

$$
f(k) :=
\begin{cases}
\dfrac{I_0\!\left( \beta\sqrt{1-\left(1-\tfrac{2}{m}k\right)^2} \right)}{I_0(\beta)}, & 0 \leq k \leq m \\
0, & \text{sonst.}
\end{cases}
$$

Dabei ist $I_0(x)$ die modifizierte Bessel-Funktion erster Art, 0. Ordnung. Die Python / SciPy Funktion `scipy.signal.kaiser(m, beta)` berechnet dieses Fenster. Im Gegensatz zu den Fenstern aus Aufgabe a) hat das Kaiser-Fenster zwei Parameter: die Länge $m+1$ und einen Parameter $\beta$, der die Steigung beeinflusst.

Durch Variation von $m$ und $\beta$ können Länge und Form des Fensters angepasst werden, um einen Kompromiss zwischen der Amplitude des ersten Überschwingers und der Breite des Hauptmaximus zu erreichen. Berechne das Kaiser-Fenster für $m=21$ und $\beta=0,3,6$. Berechne die jeweiligen Spektren und stelle sie zusammen mit der Zeitbereichsdarstellung der Fenster dar.

```python
# BEGIN SOLUTION
w0 = windows.kaiser(21, 0)
W0, W = ldtft(w0, 512)
w3 = windows.kaiser(21, 3)
W3, W = ldtft(w3, 512)
w6 = windows.kaiser(21, 6)
W6, W = ldtft(w6, 512)
k = np.arange(21)

plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(W0 / np.max(W0))))
plt.plot(W / np.pi, 20 * np.log10(np.abs(W3 / np.max(W3))), '--')
plt.plot(W / np.pi, 20 * np.log10(np.abs(W6 / np.max(W6))), '-.')
plt.axis([-1, 1, -80, 0])
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|F(exp(j*Omega))/F(exp(j*0))| in dB')
plt.title('Kaiser Window: beta: - 0,  -- 3,  -. 6')

plt.figure()
plt.plot(k, w0)
plt.plot(k, w3, '--')
plt.plot(k, w6, '-.')
plt.axis([0, 20, 0, 1.2])
plt.grid()
plt.xlabel('k')
plt.ylabel('f(k)')
plt.title('Kaiser Window: beta: - 0,  -- 3,  -. 6')

plt.show()
# END SOLUTION
```

Der Zeitverlauf jeder Fensterfunktion verdeutlicht die Mittensymmetrie, wodurch eine lineare Phase garantiert ist.  
Bei der Auswertung der Fourier-Transformation der Fensterfunktionen, um das Filterverhalten zu analysieren, zeigt sich:  
Mit steigendem Filtergrad $m$ wird das Spektrum proportional gestaucht. Daher reduziert sich das Passband (d.h. der Bereich bis zum ersten Nullpunkt des Spektrums) $\Omega < \Omega_s$ entsprechend bei einer Erhöhung von $m=21$ auf $m=42$.  
  
Das Stopbandverhalten ($\Omega \rightarrow \pi$) zeigt, dass ein höherer Filtergrad vorteilhaft ist.  
Berücksichtigt man die Dämpfung des ersten Überschwingers, so zeigt sich eine dämpfungsunabhängige Größe von $m$, z. B. $a_N \approx 57 \,\text{dB}$ beim Blackman-Fenster.  

Ein Vergleich der Fenster zeigt einen Kompromiss zwischen dem erreichbaren Passband und der Dämpfung des ersten Überschwingers.  
Während das Rechteckfenster das kleinste Passband von $2\pi/(m+1)$ und die geringste Dämpfung von $a_N \approx 13.3 \,\text{dB}$ hat, besitzt das Blackman-Fenster das breiteste Passband von $6\pi/(m+1)$, aber gleichzeitig auch die höchste Dämpfung des ersten Überschwingers.
