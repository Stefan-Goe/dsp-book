# Aufgabe: Eigenschaften der DFT

Betrachte die Serie

$$
x(k) = 
\begin{cases}
k + j \cdot (7-k), & k = 0 \dots 15\\
0, & \text{sonst}
\end{cases}.
$$

**a)** Berechne die DFT $X(n)$ mit der Python-Funktion `numpy.fft.fft(x)`.  
      Stelle den Realteil $\Re X(n)$ und den Imaginärteil $\Im X(n)$ von $X(n)$ dar.

**b)** Stelle die DFT von $\Re x(k)$ und $j\,\Im x(k)$ dar  
      (ebenfalls Real- und Imaginärteil).  
      Welche Symmetrien zeigen die DFTs von $\Re x(k)$ und $j\,\Im x(k)$?

**c)** Erzeuge eine zyklische Verschiebung von $x(k)$ um
      $\lambda=4$ bzw. $\lambda=8$ Abtastwerte:  

$$
x_1(k) = x((k+\lambda) \bmod M)
$$

Welche Beziehung besteht zwischen $X_1(n)$, der DFT von $x_1(k)$, und $X(n)$?

**d)** Berechne die diskrete Fourier-Transformation $X_2(n)$ von  

$$
x_2(k) = x(k) \cdot \exp\Big(-j \, 2\pi \, 3 k / 16\Big)
$$

Welche Beziehung besteht zwischen $X_2(n)$ und $X(n)$?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
# #####  Part a: Compute DFT of x(k)  #####
kk = np.arange(16)
x = kk + 1j * (7 - kk)
xr = np.real(x)
xi = np.imag(x)

X = np.fft.fft(x)
Xr = np.real(X)
Xi = np.imag(X)

# # Graphical Output # #
# Real part
plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

# Imaginary part
plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part b: DFT of Re{x(k)} and Im{x(k)}  #####
# # Graphical Output # #
plt.figure()
plt.stem(kk, np.real(np.fft.fft(xr)))
plt.title('Re(FFT(Re(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(np.fft.fft(xr)))
plt.title('Im(FFT(Re(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(np.fft.fft(1j * xi)))
plt.title('Re(FFT(j*Im(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(np.fft.fft(1j * xi)))
plt.title('Im(FFT(j*Im(x(k))))')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part c: Cyclic shift  #####
xtmp = np.concatenate([x, x])
x1_04 = xtmp[kk + 5]
X1_04 = np.fft.fft(x1_04)
x1_08 = xtmp[kk + 9]
X1_08 = np.fft.fft(x1_08)

plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X1_04))
plt.title('Re(X1(n)), lambda=4')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X1_04))
plt.title('Im(X1(n)), lambda=4')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X1_08))
plt.title('Re(X1(n)), lambda=8')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X1_08))
plt.title('Im(X1(n)), lambda=8')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part d: Multiplication with exp(j*...)  #####
x2 = x * np.exp(-1j * 2 * np.pi * 3 * kk / 16)
X2 = np.fft.fft(x2)

plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X2))
plt.title('Re(X2(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X2))
plt.title('Im(X2(n))')
plt.xlabel('n')
plt.grid(True)
plt.show()
# END SOLUTION
```

# Lösung: Eigenschaften der DFT

## Teil (a)

Zur besseren Veranschaulichung der Eigenschaften der diskreten Fourier-Transformation (DFT) wird ein periodisches Signal mit 16 Perioden betrachtet und nicht ein endliches Signal der Länge 16 (wie in der Aufgabenstellung gegeben). Die DFT von $x(k)$ (Real- und Imaginärteil) wird neben den Figuren von Teilaufgabe (b) dargestellt.

## Teil (b)

Die DFT eines reellen (imaginären) Signals ist konjugiert gerade (ungerade), das heißt der Realteil ist gerade (ungerade) bezüglich der Frequenz $0$ und der Imaginärteil ist ungerade (gerade). Um die Symmetrieeigenschaften zu verstehen, muss berücksichtigt werden, dass die Spektren in beiden Richtungen (positiv und negativ) kontinuierlich periodisch sind. ## Teil (c)

Laut Tabelle 6.1 entspricht eine Verschiebung im Zeitbereich um $\lambda$ Abtastwerte einer Multiplikation von $X(n)$ mit $\exp(j\,2\pi n \lambda / N)$ im Frequenzbereich. Für $N=16$ und $\lambda =4$ bzw. $\lambda =8$ erhalten wir:

$$
X_1(n) = j^n X(n) \quad \text{bzw.} \quad X_1(n) = (-1)^n X(n)

## Teil (d)

Laut Tabelle 6.1 gilt:

$$
\text{DFT}\{ \exp(-j\,2\pi i k/N) x(k) \} = X((n+i))_N
$$

Die Notation $((i))_N$ beschreibt eine Modulo-$N$ Indizierung. Die Multiplikation von $x(k)$ mit $\exp(-j\,2\pi\,3k/16)$ führt zu einer Frequenzverschiebung um $3 f_A/16$, d.h.

$$
X_2(n) = X((n+3))_{16}
$$
