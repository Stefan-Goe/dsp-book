### Übung: DFT – Abbruchfehler / Unterabtastung

Berechnen Sie die diskrete Fourier-Transformation  

$$
X_i(n) = \mathrm{DFT}\{x_i(k)\}
$$  

der folgenden Datenfolgen:

\begin{aligned}
x_1(k) &= \sin(k \pi / 8), &\text{für } k = 0\ldots 7 \text{ sowie } k = 0\ldots 63 \\
x_2(k) &= \sin(k \pi / 2), &\text{für } k = 0\ldots 7 \text{ sowie } k = 0\ldots 63 \\
x_3(k) &= \sin(3 k \pi / 2), &\text{für } k = 0\ldots 7 \text{ sowie } k = 0\ldots 63
\end{aligned}

Verwenden Sie die Python-Funktion `numpy.fft.fft(x)`.

Stellen Sie sowohl das Zeitsignal $x_i(k)$ als auch die Betragswerte der Spektren $|X_i(n)|$ für beide Zeitbereichsgrößen grafisch dar.

In welchen Fällen wird das Spektrum des zugehörigen kontinuierlichen Sinussignals $x_i(t)$ exakt wiedergegeben?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
kk08 = np.arange(8)
kk64 = np.arange(64)
kk100 = np.arange(0, 8, 8/100)

x1_08 = np.sin(kk08 * np.pi / 8)
x1_64 = np.sin(kk64 * np.pi / 8)
x2_08 = np.sin(kk08 * np.pi / 2)
x2_64 = np.sin(kk64 * np.pi / 2)
x3_08 = np.sin(kk08 * np.pi * 3 / 2)
x3_64 = np.sin(kk64 * np.pi * 3 / 2)

# Plot x1(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi / 8), ':')
plt.stem(kk08, x1_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x1(k)=sin(k*pi/8), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x1_08)))
plt.axis([0, 7, -1, 6])
plt.title('abs(X1(n)), n=0...7')

# Plot x1(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x1_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x1(k)=sin(k*pi/8), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x1_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X1(n)), n=0...63')

input('....Press RETURN to continue: ')

# Plot x2(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi / 2), ':')
plt.stem(kk08, x2_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x2(k)=sin(k*pi/2), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x2_08)))
plt.axis([0, 7, -1, 6])
plt.title('abs(X2(n)), n=0...7')

# Plot x2(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x2_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x2(k)=sin(k*pi/2), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x2_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X2(n)), n=0...63')

input('....Press RETURN to continue: ')

# Plot x3(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi * 3 / 2), ':')
plt.stem(kk08, x3_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x3(k)=sin(k*pi*3/2), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x3_08)))
plt.axis([0, 8, -1, 6])
plt.title('abs(X3(n)), n=0...7')

# Plot x3(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x3_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x3(k)=sin(k*pi*3/2), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x3_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X3(n)), n=0...63')

plt.show()

# END SOLUTION
```

### Lösung: DFT – Abbruchfehler / Unterabtastung

Sei $\tilde{x}_\nu(k)$ die periodische Fortsetzung der Datenfolge $x_\nu(k)$.  
Das Spektrum des entsprechenden kontinuierlichen Sinussignals $x_\nu(t)$ mit der
Periodendauer $T$ wird durch $\mathrm{DFT}\{\tilde{x}_\nu(k)\}$ exakt wiedergegeben, wenn

1. $\tilde{x}_\nu(k)$ identisch mit dem abgetasteten kontinuierlichen Sinussignal $x_\nu(t)|_{t=kT_A}$ ist  
2. UND das Abtasttheorem erfüllt ist, d.h. die Abtastrate $1/T_A$ ist höher als die doppelte Frequenz des kontinuierlichen Sinussignals $x_\nu(t)$.  
   Mit anderen Worten: $T_A < T/2$

---

**Signal $x_1(k)$:** In beiden Fällen ist das Abtasttheorem erfüllt  
(das kontinuierliche Sinussignal ist gepunktet eingezeichnet).  
Für $k=0..7$ ist jedoch Bedingung (1) verletzt, sodass $X_1(n)$ hier  
Spektralanteile bei vielen Frequenzen aufweist (anstatt nur bei zwei).  
Für $k=0..63$ sind beide Bedingungen erfüllt. Es ergibt sich das Spektrum  
des kontinuierlichen Sinussignals.

**Signal $x_3(k)$:** Hier ist das Abtasttheorem verletzt.  
Obwohl die Frequenz des kontinuierlichen Sinussignals hier höher als bei $x_2(t)$ ist,  
wurde die Abtastrate nicht entsprechend erhöht.  
Da $x_3(k)$ bis auf das Vorzeichen mit $x_2(k)$ übereinstimmt, ergibt die FFT exakt  
dieselben Resultate wie bei Signal $x_2(k)$.  
Die Spektren $X_3(n)$ geben das Spektrum des kontinuierlichen Sinussignals $x_3(t)$ also nicht wieder.
