Stellen Sie die Koeffizienten $f(k)$ sowie den Betrag der
normalisierten Spektren $F(e^{j\Omega}) / F(e^{j0})$ (in dB)
für ein Rechteck-, Hann-, Hamming-, Blackman- und ein
Dolph-Chebyshev-Fenster (mit $48.6\ \mathrm{dB}$ bzw. $76.1\ \mathrm{dB}$ Stopbanddämpfung)
dar, jeweils mit $N=63$ Koeffizienten.

Vergleichen Sie die erreichbaren Stopbanddämpfungen und die

Die Fensterfunktionen der Länge $N$ können aus `scipy.signal` importiert werden:

```python

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import boxcar, hamming, blackman, chebwin
# BEGIN SOLUTION
N = 63
NFFT = 2**9
f_re = boxcar(N)
aF_re = np.abs(np.fft.fft(f_re, NFFT) / np.sum(f_re))
f_hn = np.hanning(N)
aF_hn = np.abs(np.fft.fft(f_hn, NFFT) / np.sum(f_hn))
f_hm = hamming(N)
aF_hm = np.abs(np.fft.fft(f_hm, NFFT) / np.sum(f_hm))
f_bl = blackman(N)
aF_bl = np.abs(np.fft.fft(f_bl, NFFT) / np.sum(f_bl))
a_d1 = 48.6
f_d1 = chebwin(N, at=a_d1)
aF_d1 = np.abs(np.fft.fft(f_d1, NFFT) / np.sum(f_d1))
a_d2 = 76.1
f_d2 = chebwin(N, at=a_d2)
aF_d2 = np.abs(np.fft.fft(f_d2, NFFT) / np.sum(f_d2))
kk = np.arange(N)
fT = np.arange(0, 1, 1/NFFT)

# Calculation of a_min (fsT:=Omega_s/(2*pi); Stopband: n_s+1<=n<=NFFT+1-n_s)
fsT = 1/N
n_s = int(round(NFFT * fsT))
a_re = 20 * np.log10(np.max(aF_re[n_s+1:NFFT+1-n_s]))
fsT = 2/N
n_s = int(round(NFFT * fsT))
a_hn = 20 * np.log10(np.max(aF_hn[n_s+1:NFFT+1-n_s]))
fsT = 2/N
n_s = int(round(NFFT * fsT))
a_hm = 20 * np.log10(np.max(aF_hm[n_s+1:NFFT+1-n_s]))
fsT = 3/N
n_s = int(round(NFFT * fsT))
a_bl = 20 * np.log10(np.max(aF_bl[n_s+1:NFFT+1-n_s]))

# Graphical Output
# RECTANGLE Window
plt.figure()
plt.bar(kk, f_re)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Rectangle Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()

plt.figure()
plt.plot(fT, 20 * np.log10(aF_re + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_re+13, f'a_min = {a_re:.1f} dB')
plt.plot([0, 1.9], [a_re, a_re], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# HANNING Window
plt.figure()
plt.bar(kk, f_hn)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Hanning Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_hn + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_hn+15, f'a_min = {a_hn:.1f} dB')
plt.plot([0, 1.9], [a_hn, a_hn], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# HAMMING Window
plt.figure()
plt.bar(kk, f_hm)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Hamming Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_hm + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_hm+15, f'a_min = {a_hm:.1f} dB')
plt.plot([0, 1.9], [a_hm, a_hm], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# BLACKMAN Window
plt.figure()
plt.bar(kk, f_bl)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Blackman Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_bl+15, f'a_min = {a_bl:.1f} dB')
plt.plot([0, 1.9], [a_bl, a_bl], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# DOLPH-CHEBYSHEV-1
plt.figure()
plt.bar(kk, f_d1)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title(f'Dolph-Cheb., {a_d1} dB')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_d1 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, -a_d1+15, f'a_min = {a_d1:.1f} dB')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# DOLPH-CHEBYSHEV-2
plt.figure()
plt.bar(kk, f_d2)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title(f'Dolph-Cheb., {a_d2} dB')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()

# Normalized Spectral Function
plt.figure()
plt.plot(fT, 20 * np.log10(aF_d2 + np.finfo(float).eps))
plt.axis([0, 1, -150, 10])
plt.text(.3, -a_d2+15, f'a_min = {a_d2:.1f} dB')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()
# END SOLUTION
````

# Vergleich von Fensterfunktionen

Die folgenden Abbildungen zeigen die Koeffizienten $f(k)$ auf der linken Seite und den Betrag des normalisierten Spektrums

$$
20 \log_{10} \left| \frac{F(e^{j\Omega})}{F(e^{j0})} \right|
$$

in dB auf der rechten Seite für verschiedene Fenster mit $N=63$ Koeffizienten.

Das **Rechteckfenster** erreicht eine minimale Stopbanddämpfung von nur $a_\text{min} = 13.3\,\text{dB}$, besitzt aber ein sehr schmales Passband $|\Omega| \le \Omega_s = 2\pi/N \approx 0.1$, was zu einer sehr guten spektralen Auflösung führt.

Beim **Hann-Fenster** werden die Oszillationen im Stopband deutlich gedämpft ($a_\text{min} = 31.5\,\text{dB}$), wodurch Leckeffekte gemindert werden. Dieser Vorteil geht jedoch auf Kosten der spektralen Auflösung, da das Passband doppelt so breit ist wie beim Rechteckfenster: $\Omega_s = 4\pi/N \approx 0.2$.

Das **Hamming-Fenster** minimiert die erste Seitenkeule im Stopband und hält eine fast konstante Dämpfung über das gesamte Stopband aufrecht ($a_\text{min} = 42.5\,\text{dB}$). Das Passband ist wie beim Hann-Fenster: $\Omega_s = 4\pi/N \approx 0.2$. Beachten Sie, dass die ersten und letzten Koeffizienten ($f(0)$ und $f(N-1)$) nicht null sind.

Das letzte traditionelle Fenster, das **Blackman-Fenster**, erreicht eine noch höhere Stopbanddämpfung ($58.2\,\text{dB}$), jedoch mit geringerer spektraler Auflösung: $\Omega_s = 6\pi/N \approx 0.3$.

Das **Dolph-Chebyshev-Fenster** besitzt eine gleichmäßige Annäherung im Stopband, d.h. die minimale Dämpfung ist für alle Frequenzen $\Omega_s \le \Omega \le \pi$ konstant. Anders als bei traditionellen Fenstern definiert die Fensterlänge $N$ **nicht** die Stopband-Grenzfrequenz $\Omega_s$. Wenn entweder der Wert von $a_\text{min}$ oder $\Omega_s$ gewählt wird, ist der andere Wert festgelegt, und das Fenster kann berechnet werden.

Nach Gleichung (7.4.28) ergeben sich die Werte $\Omega_s \approx 4\pi/N \approx 0.2$ und $6\pi/N \approx 0.3$ für $N=63$ mit $a_\text{min} = 48.6$ dB bzw. $76.1$ dB. Bei denselben Werten von $N$ und $\Omega_s$ erreicht das Fenster eine höhere minimale Stopbanddämpfung, etwa $6\,\text{dB}$ höher als das entsprechende Hamming-Fenster und $18\,\text{dB}$ höher als das Blackman-Fenster.

Für eine feste Fensterlänge $N$ zeigt die folgende Tabelle die Beziehung zwischen der erreichbaren minimalen Stopbanddämpfung $a_\text{min}$ und der Stopband-Grenzfrequenz $\Omega_s$:

| Fensterfunktion | Rechteck | Hann | Hamming | Blackman | Dolph-Chebyshev | Dolph-Chebyshev |
|-----------------|----------|------|---------|----------|-----------------|-----------------|
| Minimale Stopbanddämpfung $a_\text{min}$ | $13.3\,\text{dB}$ | $31.5\,\text{dB}$ | $42.5\,\text{dB}$ | $58.2\,\text{dB}$ | $48.6\,\text{dB}$ | $76.1\,\text{dB}$ |
| Pass Band |$\Omega$| $\le$ $\Omega_s$ | $2\pi/N$ | $4\pi/N$ | $4\pi/N$ | $6\pi/N$ | $\approx 4\pi/N$ | $\approx 6\pi/N$ |
| Hinweise | Schmalstes Passband, geringe Stopbanddämpfung | Oszillationen gedämpft, moderate Auflösung | Minimiert erste Seitenkeule, moderate Auflösung | Hohe Dämpfung, geringere Auflösung | Gleichmäßiges Stopband, Stopband durch $a_\text{min}$ oder $\Omega_s$ bestimmt | Gleichmäßiges Stopband, Stopband durch $a_\text{min}$ oder $\Omega_s$ bestimmt |

**Hinweis:** Bei traditionellen Fenstern sind die Werte $a_\text{min}$ **unabhängig von der Fensterlänge $N$**. Eine Vergrößerung von $N$ komprimiert lediglich die Seitenkeulen, deren Maxima bleiben unverändert. Dies gilt nicht für das Dolph-Chebyshev-Fenster.
