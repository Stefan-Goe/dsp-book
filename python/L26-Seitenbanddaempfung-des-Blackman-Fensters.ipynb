### Aufgabe — Seitenbanddämpfung des Blackman-Fensters

Stellen Sie den Betrag des normierten Spektrums dar  
$$
\frac{F(\mathrm{e}^{j\Omega})}{F(\mathrm{e}^{j0})}
$$
(in dB) eines Blackman-Fensters mit  
$N = 32$, $N = 64$ bzw.~$N = 128$ Koeffizienten.  

Ermitteln Sie (für diese Werte von $N$) das Verhältnis  
des maximalen Werts zum zweitgrößten Wert.

---

Welcher Effekt lässt sich hier — wie auch bei allen anderen  
traditionellen Fensterfunktionen — beobachten?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
NFFT = 2**10
f_bl032 = np.blackman(32)
aF_bl032 = np.abs(np.fft.fft(f_bl032, NFFT) / np.sum(f_bl032))
f_bl064 = np.blackman(64)
aF_bl064 = np.abs(np.fft.fft(f_bl064, NFFT) / np.sum(f_bl064))
f_bl128 = np.blackman(128)
aF_bl128 = np.abs(np.fft.fft(f_bl128, NFFT) / np.sum(f_bl128))
fT = np.arange(0, 1, 1/NFFT)

# Calculation of minimum stopband attenuation a_min:
# Os := Omega_s/(2*pi);  n_s: stopband range indices: n_s+1<=n<=NFFT+1-n_s
Os = 3/32
n_s = round(NFFT * Os)
a032 = 20 * np.log10(np.max(aF_bl032[n_s+1:NFFT+1-n_s]))
Os = 3/64
n_s = round(NFFT * Os)
a064 = 20 * np.log10(np.max(aF_bl064[n_s+1:NFFT+1-n_s]))
Os = 3/128
n_s = round(NFFT * Os)
a128 = 20 * np.log10(np.max(aF_bl128[n_s+1:NFFT+1-n_s]))

# Graphical Output
plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl032 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a032, f'a_min = {a032:.1f} dB')
plt.plot([0, 0.39], [a032, a032], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=32')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')

plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl064 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a064, f'a_min = {a064:.1f} dB')
plt.plot([0, 0.39], [a064, a064], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=64')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')

plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl128 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a128, f'a_min = {a128:.1f} dB')
plt.plot([0, 0.39], [a128, a128], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=128')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()
# END SOLUTION
```

### Blackman-Fenster

Die folgende Gleichung definiert das Blackman-Fenster:

$$
f^{BL}(k) =
\begin{cases}
0.42 - 0.5 \cdot \cos\left(\frac{2\pi k}{N - 1}\right) + 0.08 \cdot \cos\left(\frac{4\pi k}{N - 1}\right), & 0 \le k \le N - 1 \\
0, & \text{sonst}
\end{cases}
$$

Mit $a_{min} \approx 58\,\text{dB}$ zeigt es im Vergleich zu allen anderen traditionellen Fenstern  
die höchste Dämpfung im Sperrband  
(Rechteckfenster: $13.3\,\text{dB}$, Hann: $31.5\,\text{dB}$, Hamming: $42.5\,\text{dB}$).  

Auf der anderen Seite muss die starke Dämpfung mit einer breiten Durchlassbandbreite bezahlt werden:  
$\Omega_s = \frac{6\pi}{N}$ (Rechteckfenster: $\frac{2\pi}{N}$, Hann und Hamming: $\frac{4\pi}{N}$).

Wir bestätigen:
1. Die Sperrband-Grenzfrequenz ist umgekehrt proportional zur Fensterlänge:  
   $\Omega_s \sim \frac{1}{N}$.
2. Die minimale Sperrbanddämpfung ändert sich mit der Fensterlänge nur unwesentlich:  
   $a_{min} \approx \text{const}_N$.

**Bemerkung:**  
Für alle traditionellen Fenster (Rechteck-, Hann-, Hamming- und Blackman-Fenster)  
ist die *Seitenbanddämpfung unabhängig von der Fensterlänge* $N$.  
Eine Vergrößerung der Fensterlänge führt lediglich zu einer Kompression der Seitenbänder  
und kann zur Verbesserung der Frequenzauflösung genutzt werden.
