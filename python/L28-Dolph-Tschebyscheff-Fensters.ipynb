**Warum ist in Python keine abschnittsweise Definition der Spektralfunktion notwendig?**

Da

$$
\cosh(x) = 0.5 \left( e^x + e^{-x} \right), \quad \cos(x) = 0.5 \left( e^{jx} + e^{-jx} \right),
$$

gilt

$$
\cosh(x) = \cos(-jx).
$$

Mit

$$
y(\Omega) = \frac{\cos(\Omega/2)}{\cos(\Omega_s/2)}
$$

im **Passband** des nicht-kausalen Dolph-Chebyshev-Fensters gilt:

\begin{align*}
F_0^{DT}(e^{j\Omega}) &= \cosh\big( N \cdot \operatorname{arcosh}(y(\Omega)) \big)
&= \cos\Big( (-j) \cdot N \cdot \operatorname{arcosh}(y(\Omega)) \Big), \quad 0 \le \Omega \le \Omega_s
\end{align*}

Für das **Sperrband** gilt:

$$
F_0^{DT}(e^{j\Omega}) = \cos\left( N \cdot \arccos(y(\Omega)) \right), \quad \Omega_s \le \pi
$$

Es bleibt zu zeigen:

\begin{align*}
(-j) \cdot \operatorname{arcosh}(y(\Omega)) &\stackrel{!}{=} \arccos(y(\Omega)) \
\cos(-j \cdot \operatorname{arcosh}(y(\Omega))) &\stackrel{!}{=} y(\Omega)
\end{align*}

Aufgrund der obigen Beziehung – unter Verwendung der **komplexwertigen Definition hyperbolischer Funktionen** – fallen beide Fälle zusammen.

Da Python-Funktionen auf **komplexwertigen Definitionen** basieren, ist eine **abschnittsweise Definition** der Spektralfunktion des Dolph-Chebyshev-Fensters nicht notwendig.
