### Aufgabe: Yule–Walker

Ziel dieser Aufgabe ist die Analyse der Spektralschätzung mithilfe des Yule–Walker- bzw. des Autokorrelationsverfahrens. Für dieses Vorgehen ist es entscheidend, dass der zu analysierende Prozess $X(k)$ außerhalb des gemessenen Intervalls $0 \leq k \leq N-1$ auf null gesetzt wird. Damit wird das *Einschwingverhalten* des Prädiktors bei der Berechnung des Prädiktionsfehlers (Gl. (9.3.7)) berücksichtigt. Trotz des Einflusses dieses Einschwingverhaltens wird dieser Ansatz als stationär bezeichnet. Der Grund hierfür liegt in der Berechnung der Koeffizienten der Korrelationsmatrix, die die Form einer hermiteschen Toeplitz-Matrix besitzt. Eine solche Matrix tritt auch als Autokorrelationsmatrix stationärer Prozesse auf.

Zusätzlich wird die verzerrte Autokorrelationsfolge $\hat{r}_{XX}(\kappa)$ verwendet. Daher ist auch die Schätzung der AR-Koeffizienten verzerrt.

Für die Analyse wird ein (Allpol-)System $H(z)=1/A(z)$ betrachtet mit

$$
A(z)=1 + 0.3611z^{-1} - 0.1373z^{-2} + 0.2685z^{-3} - 0.0533z^{-4} - 0.3353z^{-5} - 0.307z^{-6}
$$

der Ordnung $m=6$, das durch einen stationären, mittelwertfreien, weißen Prozess $N(k)$ mit der Leistung $\sigma_N^2 = 1$ angeregt wird.

Stellen Sie das exakte Spektrum, das gemittelte geschätzte Spektrum sowie die oberen $2\sigma$-Grenzen (95 %-Grenzen) gemeinsam in einer Abbildung dar.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from helper_functions import lyw, zplane
# BEGIN SOLUTION
MA = np.array([1.0])
AR = np.array([1.0000, 0.3611, -0.1373, 0.2685, -0.0533, -0.3353, -0.307])
NFFT = 2**10
n = len(AR) - 1

# Task a: Pole-Zero Plot & Impulse Response  
# Execute your zplane function
zplane(MA, AR)

# Impulse response
impulse = np.zeros(80)
impulse[0] = 1.0
h_arma = signal.lfilter(MA, AR, impulse)

plt.figure()
plt.stem(np.arange(1, len(h_arma)+1), h_arma)
plt.title(f'Impulse Response of the AR({n}) Model')
plt.grid(True)
plt.show()
print('Samples required for transient settling: N_vor=80')

# --- Task b: Exact ACF, YW-Coefficients, and PSD ---
# Exact ACF via h_arma
rxx = signal.correlate(h_arma, h_arma, mode='full')
center = len(rxx) // 2
rxx_ex = rxx[center : center + n + 1]

ARcoeff_b, _, _ = lyw(h_arma, n, 0)
Omega_norm = np.linspace(0, 1, NFFT, endpoint=False)

# Frequency response calculation (PSD)
w, h_freq = signal.freqz(MA, AR, worN=NFFT, whole=True)
Sxx_ex = np.abs(h_freq)**2

plt.figure()
plt.stem(np.arange(n+1), rxx_ex, linefmt='r-', markerfmt='ro')
plt.grid(True)
plt.xlabel(r'$\kappa$')
plt.ylabel(r'$r_{xx}(\kappa)$')
plt.title(f'Exact ACF values for AR({n}): $r_{{xx}}(0)...r_{{xx}}({n})$')

plt.figure()
plt.stem(np.arange(n+1), ARcoeff_b, linefmt='r-', markerfmt='ro')
plt.grid(True)
plt.ylim([1.1*np.min(ARcoeff_b), 1.1*np.max(ARcoeff_b)])
plt.xlabel('i')
plt.ylabel('$a_i$')
plt.title(f'Exact AR({n}) Coefficients: $a_0...a_{{{n}}}$')

plt.figure()
plt.plot(Omega_norm, 10 * np.log10(Sxx_ex / np.max(Sxx_ex)), 'r--')
plt.grid(True)
plt.xlabel(r'$\Omega/2\pi$')
plt.ylabel(r'$S_{xx}(e^{j\Omega})$ in dB')
plt.title('Exact PSD')

# --- Task c: Influence of Data Length N (Monte Carlo) ---
N_list = [10, 256]
N_vor = 80
mc_run = 1000

print('Calculating Monte Carlo simulation...')

for N_val in N_list:
    Sxx_ar_stack = np.zeros((mc_run, NFFT))
    Coeff_stack = np.zeros((mc_run, n + 1))
    
    for r in range(mc_run):
        noise = np.random.randn(N_val + N_vor)
        x_all = signal.lfilter(MA, AR, noise)
        x = x_all[N_vor:] # Remove transient effect
        
        coeffs, _, _ = lyw(x, n, 1)
        Coeff_stack[r, :] = coeffs
        
        # Estimate PSD from coefficients
        _, h_est = signal.freqz([1.0], coeffs, worN=NFFT, whole=True)
        Sxx_ar_stack[r, :] = np.abs(h_est)**2
        
    Coeff_mean = np.mean(Coeff_stack, axis=0)
    Coeff_std = np.std(Coeff_stack, axis=0)
    C2sig_o = Coeff_mean + 2*Coeff_std
    C2sig_u = Coeff_mean - 2*Coeff_std
    Coeff_mse = np.mean(np.abs(AR - Coeff_mean)**2)
    
    Sxx_mean = np.mean(Sxx_ar_stack, axis=0)
    Sxx_std = np.std(Sxx_ar_stack, axis=0)
    Sxx_2sigma = Sxx_mean + 2*Sxx_std

    # Plot AR Coefficients Statistics
    plt.figure()
    plt.plot(np.arange(n+1), AR, 'ro', label='Exact', mfc='none')
    plt.plot(np.arange(n+1), Coeff_mean, 'bx', label=f'Mean / MSE={Coeff_mse:.2e}')
    plt.plot(np.arange(n+1), C2sig_o, 'go', label=f'2$\sigma$ / $\sigma_{{avg}}$={np.mean(Coeff_std):.2f}')
    plt.plot(np.arange(n+1), C2sig_u, 'go')
    
    # Draw vertical lines for the confidence interval
    for i in range(n+1):
        plt.plot([i, i], [C2sig_u[i], C2sig_o[i]], 'g')
        
    plt.title(f'AR({n}) Coefficients; N={N_val}')
    plt.xlabel('i')
    plt.ylabel('Coefficients')
    plt.legend()
    plt.grid(True)
    plt.ylim([-1.5, 1.5])

    # Plot PSD Statistics
    plt.figure()
    plt.plot(Omega_norm, 10*np.log10(Sxx_ex/np.max(Sxx_ex)), 'r--', label='Exact')
    plt.plot(Omega_norm, 10*np.log10(Sxx_mean/np.max(Sxx_ex)), 'b-', label='Mean')
    plt.plot(Omega_norm, 10*np.log10(Sxx_2sigma/np.max(Sxx_ex)), 'g-.', label='2$\sigma$ Bound')
    plt.ylabel(r'$S_{xx}(e^{j\Omega})$ in dB')
    plt.xlabel(r'$\Omega/2\pi$')
    plt.title(f'PSD of YW-AR({n}); N={N_val}')
    plt.legend()
    plt.grid(True)

plt.show()
# END SOLUTION
```

* **b)**
  Für stationäres, mittelwertfreies, weißes Rauschen gilt die Beziehung
  $r_{NN}(\kappa)=\sigma_N^2\delta(\kappa) ;\Leftrightarrow;
  S_{NN}(\mathrm{e}^{j\Omega})=\sigma_N^2$.
  Am Systemausgang liegen somit die wahre Autokorrelationsfunktion (ACF) sowie die exakte
  spektrale Leistungsdichte bei $\sigma_N^2=1$ vor:

  $$
  r_{XX}(\kappa)
  = r_{HH}^{(E)}(\kappa) \ast r_{NN}(\kappa)
  = \sigma_N^2 \cdot r_{hh}^{(E)}(\kappa)
  = r_{HH}^{(E)}(\kappa)
  := \sum_{k=-\infty}^{\infty} h^*(k),h(k+\kappa)
  $$

  $$
  S_{XX}(\mathrm{e}^{j\Omega})
  = |H(\mathrm{e}^{j\Omega})|^2 \cdot S_{NN}(\mathrm{e}^{j\Omega})
  = \sigma_N^2 \cdot |H(\mathrm{e}^{j\Omega})|^2
  = |H(\mathrm{e}^{j\Omega})|^2
  $$

  Zur Berechnung der AR-Koeffizienten mithilfe der Yule–Walker-Gleichung (9.3.17) wird die
  Autokorrelationsfolge an den Stellen
  $r_{XX}(0), \ldots, r_{XX}(n)$ betrachtet. Der Parameter $n$ beschreibt dabei die Ordnung
  des AR($n$)-Prozesses und muss nicht zwingend mit der Ordnung $m$ des zugrunde liegenden
  Systems $H(z)$ übereinstimmen (Unter- bzw. Überschätzung der Polanzahl).

  Zusätzlich zur exakten ACF-Folge sind in der folgenden Abbildung die exakten
  AR(6)-Koeffizienten $a_0,\ldots,a_6$ dargestellt. Die Berechnung erfolgt mithilfe der
  Levinson–Durbin-Rekursion zur Lösung der Yule–Walker-Gleichung (9.3.17).

  Für die Leistungsdichtespektren (PDS)
  $S_{XX}(\mathrm{e}^{j\Omega}) \leftrightarrow r_{XX}(\kappa)$ ist der Verlauf in der
  unteren Abbildung als gestrichelte Linie dargestellt. Die Berechnung der
  Leistungsdichtespektren erfolgt nach

  $$
  S_{XX}(\mathrm{e}^{j\Omega})
  = \sigma_N^2 \cdot |H(\mathrm{e}^{j\Omega})|^2
  = \left| \frac{B(\mathrm{e}^{j\Omega})}{A(\mathrm{e}^{j\Omega})} \right|^2
  = \frac{\left| \mathcal{F}{b(k)} \right|^2}{\left| \mathcal{F}{a(k)} \right|^2}.
  $$

  Für eine grafische Darstellung des wahren Leistungsdichtespektrums
  $S_{XX}(\mathrm{e}^{j\Omega})$ wird die zeitdiskrete Fourier-Transformation
  $\mathcal{F}{\cdot}$ durch eine $2^{10}$-Punkt-FFT approximiert.
  Da das Übertragungssystem $H(z)$ reellwertig ist, ergibt sich ein symmetrisches
  Leistungsdichtespektrum, normiert auf die Frequenz $\Omega/2\pi$.
