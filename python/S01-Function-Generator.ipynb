## Function Generator (Chapter 2.1)

In this exercise a function generator shall be programmed which generates a signal of frequency $f_0$  at a sampling frequency of $f_\mathrm{A}$.  This generator shall be able to generate a sinusoidal and a sawtooth signal.

For a time continuous system it is common to calculate using the angular frequency $\omega_0 = 2 \pi f_0$. Thus a time continuous sine signal is given by

$$
x_\mathrm{sin,K}(t)=\sin \left( \omega_0 \cdot t \right)
$$

Within a time discrete system the sinusoidal signal is sampled with periods of the length $T = 1/f_\mathrm{A}$, such that the time $t=T \cdot k$ can be written by the indices $k = 0, 1, 2, ...$. Therefore we obtain

$$
x_\mathrm{sin}(k)= x_\mathrm{sin,K}(t)|_{t=kT} = \sin \left( \omega_0 \cdot T \cdot k \right)
$$

For time discrete systems it makes sense to introduce the normalized angular frequency $\Omega_0$, which is without any unit:

$$
\begin{array}{rcl}
\Omega_0 & := & \omega_0 \cdot T \\
         & =  & 2 \pi f_0 \cdot T \\
         & =  & 2 \pi f_0 / f_\mathrm{A}
\end{array}
$$

This definition will be used throughout the further exercises, since it allows handling of digital systems without any limitation of the sampling frequency. Thus,

$$
x_\mathrm{sin}(k) = \sin \left( \Omega_0 \cdot k \right)
$$

### a)

Create a vector `x` with 60 samples of a sinusoidal signal with the frequency $f_0 = 400\,\text{Hz}$ at a sampling frequency of $8\,\text{kHz}$.

In order to do so initialize a coefficient vector  
`k = np.arange(60)`,  
which represents the time index of the sampled signal $x(k)$.  
Make use of **Python’s** `numpy.sin()` function.

Display the signal with the help of the function `matplotlib.pyplot.stem()`.  
What is the period of the sampled signal?

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

# BEGIN SOLUTION
fa = 8000  # Setting the sampling rate

# ##### Subtask a: Sine Wave #####
# Calculation of the signal vector
k = np.arange(0, 60)  # Creating a vector of length 60 with entries from 0 to 59
f0 = 400  # Signal frequency
Omega0 = 2 * np.pi * f0 / fa  # Normalized frequency
xsin = np.sin(Omega0 * k)  # Sine wave vector

# Plotting
plt.figure()
plt.stem(k, xsin)
plt.xlabel('k')
plt.ylabel('x_{sin}(k)')
plt.title('Sine Wave with f0 = 400 Hz')
plt.show()
# END SOLUTION
```

```python
# ##### Subtask b: Sine Wave, Altered Period #####

# Calculation of the signal vector
# BEGIN SOLUTION
f0 = 960  # Signal frequency
Omega0 = 2 * np.pi * f0 / fa  # Normalized frequency
xsin = np.sin(Omega0 * k)  # Sine wave vector

# Plotting
plt.figure()
plt.stem(k, xsin)
plt.xlabel('k')
plt.ylabel('x_{sin}(k)')
plt.title('Sine Wave with f0 = 960 Hz')
plt.show()
# END SOLUTION
```

```python
# ##### Subtask c: Sawtooth Wave #####

# BEGIN SOLUTION
xsaw = np.mod(f0 / fa * k, 1)  # Sawtooth wave vector

# Plotting
plt.figure()
plt.stem(k, xsaw)
plt.xlabel('k')
plt.ylabel('x_{saw}(k)')
plt.title('Sawtooth Wave')
plt.show()
# END SOLUTION
```

### d)

Create a vector of a sinusoidal as well as a sawtooth signal with  
$f_0 = 1\,\text{kHz}$ at a sampling frequency of $f_\mathrm{A} = 8\,\text{kHz}$.

The signals shall be played through a soundcard and have a duration of $d = 2\,\mathrm{s}$.  
How many samples are necessary for this?

Display the signals using `display(Audio(x, rate=fa, normalize=False))` and play them with the notebook audio player.

```python
# ##### Subtask d: Denormalized Signals, Sound Generation #####

# BEGIN SOLUTION
f0 = 1000  # Signal frequency
d = 2  # Playback duration in seconds
w = d * fa  # Number of samples
k = np.arange(0, w + 1)  # Coefficient vector of the time index
xsin = np.sin(2 * np.pi * f0 / fa * k)  # Sine wave vector
xsaw = np.mod(f0 / fa * k, 1)  # Sawtooth wave vector

# Playback of the signals
print('Audio: 2s Sine Wave')
display(Audio(xsin, rate=fa, normalize=False))

print('Audio: 2s Sawtooth Wave')
display(Audio(xsaw, rate=fa, normalize=False))
# END SOLUTION
```

The period of the continuous signal

is $T_0 = 1 / f_0$.  
Since the number of samples per period of the time continuous signal is an integer,  
the discrete signal has to have a period of

$$
\begin{array}{rcl}
k_p &=& T_0 / T \\
    &=& f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 400\,\text{Hz} \\
    &=& 20
\end{array}
$$

values.

The period of the time discrete signal appears after

$$
\begin{array}{rcl}
k_p &=& L \cdot T_0 / T \\
    &=& L \cdot f_A / f_0 \\
    &\Downarrow& f_A = 8\,\text{kHz},\ f_0 = 960\,\text{Hz} \\
    &=& L \cdot 8.\overline{3} \\
    &\Downarrow& L = 3 \\
    &=& 25
\end{array}
$$

samples.  
For this, $L=3$ is the number of periods of the continuous signal, after which the periodicity of the sampled signal settles.
