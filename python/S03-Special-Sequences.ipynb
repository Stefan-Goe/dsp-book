## Background Theory

The impulse series is a discrete form of the [Dirac delta function](https://en.wikipedia.org/wiki/Dirac_delta_function) and takes the form:

$$
\delta[k] := 
  \begin{cases}
    1 \quad k = 0 \\
    0 \quad k \neq 0 \\
  \end{cases}
$$

When our time index $k$ is at zero, the function has a value of $1$, otherwise  it has a value of $0$. This is a useful fundamental series for digital signal processing - for example, we will later be used as part of the convolution process of the Fourier Transform.

<div class="alert alert-block alert-info">
<strong>Task 3 (a): Impulse and Impulse sequence</strong><br>
<ul>
<li>Generate and plot an impulse series:

$$x[k]=1.2 \cdot \delta(k − 4), \qquad −5 \leq k \leq 10$$

where

$$
\delta[k] := 
  \begin{cases}
    1 \quad k = 0 \\
    0 \quad k \neq 0 \\
  \end{cases}
$$

We could either create an array of $0$s and a single $1$ to represent the impulse series, or we could define an impulse function and apply it using [`vectorize`](https://numpy.org/doc/stable/reference/generated/numpy.vectorize.html).
</li>
</ul>
</div>

```python
# Let's do some necessary and nice-to-have imports
%matplotlib inline
import matplotlib.pyplot as plt  # plotting
#import seaborn as sns; sns.set() # styling
import numpy as np               # math
import IPython.display as ipd    # for audio playback
```

```python
k = np.arange(-5, 11)                   # initialise an index

def impulse(k):                         # define an impulse function
    if k == 0:
        return 1
    else:
        return 0

x = 1.2 * np.vectorize(impulse)(k-4)  # vectorise lets us apply our impulse function 

plt.figure()
plt.title('$x[k]=1.2 \cdot \delta[k−4]$')
plt.xlabel("$k$")
plt.ylabel("amplitude")
plt.stem(k, x, use_line_collection = True);
```
