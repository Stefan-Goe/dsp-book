Plot the frequency response and the zero-pole plot for the given system using the **Python** libraries **SciPy** and **Matplotlib** (e.g. `scipy.signal.freqz()` and `scipy.signal.tf2zpk()`).

```python
import numpy as np
import matplotlib.pyplot as plt
from helper_functions import lsys2
from scipy.signal import tf2zpk
# BEGIN SOLUTION
# Impulse response
x = np.concatenate(([1], np.zeros(20)))
y = lsys2(x)


# Frequency response
b = np.array([1.0, 1.5, 1.0])
a = np.array([1.0, -1.0, 0.5])

N = 256
w = np.linspace(0, np.pi, N)
z = np.exp(1j * w)

B = np.polyval(b, z**-1)
A = np.polyval(a, z**-1)
H = B / A

mag_db = 20 * np.log10(np.abs(H))

plt.figure()
plt.plot(w / np.pi, mag_db)
plt.grid(True)
plt.axis([0, 1, -35, 25])
plt.xlabel(r'$\Omega/\pi$')
plt.ylabel('Magnitude (dB)')
plt.title("Frequency Response")


# z-plane (poles and zeros)
zeros, poles, gain = tf2zpk(b, a)

plt.figure()
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.scatter(np.real(zeros), np.imag(zeros),
            marker='o', facecolors='none', edgecolors='b', s=80, label='Zeros')

plt.scatter(np.real(poles), np.imag(poles),
            marker='x', color='r', s=80, label='Poles')

phi = np.linspace(0, 2*np.pi, 400)
plt.plot(np.cos(phi), np.sin(phi), 'k--', linewidth=0.8)

plt.xlabel("Real")
plt.ylabel("Imag")
plt.title("z-Plane")
plt.legend()
plt.axis("equal")
plt.grid(True)

plt.show()

# END SOLUTION
```

### Solution: Analysis of a 2nd order system

---

#### a)
The schematic corresponds to the following equation

$$
Y(z) = Y(z) z^{-1} - 0.5 \cdot Y(z) z^{-2}
      + X(z) + 1.5 \cdot X(z) z^{-1} + X(z) z^{-2}
$$

Transformation to the time domain yields the difference equation

$$
y(k) = y(k-1) - 0.5 \cdot y(k-2)
       + x(k) + 1.5 \cdot x(k-1) + x(k-2).
$$

Transforming the transfer function results in the familiar expression:

$$
Y(z)\left(1 - z^{-1} + 0.5 \cdot z^{-2}\right)
=
X(z)\left(1 + 1.5 \cdot z^{-1} + z^{-2}\right)
$$

$$
H(z) = \frac{Y(z)}{X(z)}
     = \frac{1 + 1.5 \cdot z^{-1} + z^{-2}}
            {1 - z^{-1} + 0.5 \cdot z^{-2}}.
$$

---

#### b)
The impulse response can be obtained by exciting the system with

$$
x(k)=
\begin{cases}
1 & \text{for } k=0 \\
0 & \text{otherwise}
\end{cases}
$$

Its first 10 values are:

$$
h(k)=
\left[
1,\ 2.5,\ 3,\ 1.75,\ 0.25,\ -0.625,\ -0.75,\ -0.4375,\ -0.0625,\ 0.1562,\ \dots
\right]
$$

---

#### c)
The coefficient vectors are

$$
\mathbf{b} = \{1,\ 1.5,\ 1\}
$$

and

$$
\mathbf{a} = \{1,\ -1,\ 0.5\}.
$$

The connection between frequency response and zero-pole plot can be demonstrated by well-known relationships. Zeros on the unit circle yield a break-in of the frequency response, while the poles are responsible for the maximum.
