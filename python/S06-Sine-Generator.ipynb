# Sine Generator

In this exercise a sine generator for a fixed frequency shall be synthesized, which
is realized without using any tables or polynomial approximations.
The basis of this generator is a 2nd order system which, once excited,
produces the values of a sine signal.

For the synthesis process the following correspondence of the Z-transform shall be used:

$$
h(k)=\sin!\big(\Omega_0 k\big)
$$

$$
H(z) = Z{h(k)} = \frac{z ,\sin(\Omega_0)}{z^2 - 2 z \cdot \cos(\Omega_0) + 1}
$$

This correspondence can be interpreted as the **impulse response** $h(k)$ of a system
$H(z)=Y(z)/X(z)$.

### d)

Now the effects of quantizing the coefficients of the system shall be investigated.

Therefore replace the expression $2 \cdot \cos(\Omega_0)$ in the coefficient vector `a` by $2.1 \cdot \cos(\Omega_0)$.

Display the zero–pole plot and the impulse response of the system using Python.

````python
# BEGIN SOLUTION
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import dlti, dimpulse, tf2zpk

# Normalized dimensionless angular frequency
Omega0 = 2 * np.pi * 1 / 32

# Sampled sine signal
k = np.arange(0, 100)
x = np.sin(Omega0 * k)

# ##### Subtask b: Sine Generator #####
b = [0, np.sin(Omega0), 0]
a = [1, -2 * np.cos(Omega0), 1]

zeros, poles, gain = tf2zpk(b, a)

# z-Plane Plot
plt.figure()
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)

# Draw the unit circle
theta = np.linspace(0, 2*np.pi, 500)
plt.plot(np.cos(theta), np.sin(theta), 'k--', label='Unit Circle')

# Plot zeros and poles
plt.scatter(np.real(zeros), np.imag(zeros), marker='o', facecolors='none', edgecolors='b', label='Zeros')
plt.scatter(np.real(poles), np.imag(poles), marker='x', color='r', label='Poles')

plt.gca().set_aspect('equal')  # Ensure circle is not distorted
plt.grid(True)
plt.xlabel('Real Part')
plt.ylabel('Imaginary Part')
plt.title('z-Plane')
plt.show()


# Impulse response
system = dlti(b, a)
t, h = dimpulse(system, n=100)
h = np.squeeze(h)

plt.figure()
plt.stem(t, h)
plt.xlabel('k')
plt.ylabel('h(k)')
plt.title('Impulse Response')
plt.plot(k, x, 'r-', label='Sampled Sine Signal')
plt.legend()
plt.show()

# ##### Subtask c: Forward Branch #####
b = [1, 0, 0]
a = [1, -2 * np.cos(Omega0), 1]

# Graphical output
system = dlti(b, a)
t, h = dimpulse(system, n=100)
h = np.squeeze(h)

plt.figure()
plt.stem(t, h)
plt.xlabel('k')
plt.ylabel('h(k)')
plt.title('Impulse Response')
plt.show()

# ##### Subtask d: Instability #####
b = [0, np.sin(Omega0), 0]
a = [1, -2.1 * np.cos(Omega0), 1]

# Graphical output
# Poles and zeros in the z-plane
plt.figure()
zeros, poles, gain = tf2zpk(b, a)

theta = np.linspace(0, 2*np.pi, 500)
plt.plot(np.cos(theta), np.sin(theta), 'k--', label='Unit Circle')

plt.scatter(np.real(zeros), np.imag(zeros), marker='o', label='Zeros')
plt.scatter(np.real(poles), np.imag(poles), marker='x', label='Poles')

plt.gca().set_aspect('equal')  # Ensure circle is not distorted
plt.xlabel('Real Part')
plt.ylabel('Imaginary Part')
plt.title('z-Plane')
plt.grid()
plt.show()

# Impulse response
system = dlti(b, a)
t, h = dimpulse(system, n=100)
h = np.squeeze(h)

plt.figure()
plt.stem(t, h)
plt.xlabel('k')
plt.ylabel('h(k)')
plt.title('Impulse Response')
plt.show()
# END SOLUTION
````

## Solution: Sine Generator

#### a)

$$
H(z) = \frac{Y(z)}{X(z)} = \frac{z^{-1} \sin(\Omega_0)}{1 - 2 z^{-1} \cos(\Omega_0) + z^{-2}}
$$
