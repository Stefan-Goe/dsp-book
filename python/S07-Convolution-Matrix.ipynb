### c) Commutativity of convolution
The convolution operator is **commutative**, i.e., you can interchange **signal** and **system**.  

Therefore, it is possible to build the convolution matrix either with the **short impulse response h**  
or the **long signal x**.

Decide which of the two possible convolution matrices leads to a more efficient computation.

```python
import numpy as np
# BEGIN SOLUTION
# ##### Task a: Write own convmtx function #####
def lconvmtx(h, n):
    """
    Generate a convolution matrix for a given impulse response vector.

    Parameters:
    - h: numpy array or list
        The impulse response vector.
    - n: int
        The number of columns in the convolution matrix.

    Returns:
    - H: numpy array
        The convolution matrix.

    Example:
    >>> h = [1, 2, 3]
    >>> n = 4
    >>> lconvmtx(h, n)
    array([[1., 0., 0., 0.],
           [2., 1., 0., 0.],
           [3., 2., 1., 0.]])
    """
    h = np.asarray(h).flatten()            # Ensure h is 1D
    H = np.zeros((len(h) + n - 1, n))      # Create zero matrix of correct size
    for l in range(n):
        H[l:l+len(h), l] = h               # Fill with shifted impulse response
    return H


# ##### Task b: Test convolution matrix #####
h = np.array([1, 2, 3, 4, 5])              # Impulse response of the system
x = np.sin(2 * np.pi * 1/10 * np.arange(20))  # Test signal

yhx = np.convolve(h, x)                    # Direct convolution

Hh = lconvmtx(h, len(x))                   # Convolution using convolution matrix
yHhx = Hh @ x                              # Matrix multiplication
difference = yhx - yHhx                    # Compare results
print("Difference:", difference)


# ##### Task c: Swap system <-> signal #####
print("1st variant: Convolution matrix formed with h")
Hh = lconvmtx(h, len(x))
Hhsize = Hh.shape                          # Size of Hh (24 x 20)
Hhsizeprod = np.prod(Hhsize)               # Total number of entries
convHhx = Hh @ x
print("Hh size:", Hhsize)
print("Hh entries:", Hhsizeprod)

print("2nd variant: Convolution matrix formed with x")
Hx = lconvmtx(x, len(h))
Hxsize = Hx.shape                          # Size of Hx (24 x 5)
Hxsizeprod = np.prod(Hxsize)               # Total number of entries
yHxx = Hx @ h
print("Hx size:", Hxsize)
print("Hx entries:", Hxsizeprod)

# END SOLUTION
```

### c) Efficiency of different convolution matrices

Let  

- $n_x$ = length of vector $\mathbf{x}$  
- $n_h$ = length of vector $\mathbf{h}$  

---

If the convolution matrix is built from $\mathbf{h}$, it has dimensions  

$$
(n_h + n_x - 1) \times n_x
$$  

and therefore  

$$
n_1 = n_x n_h + n_x^2 - n_x
$$  

entries.

---

If the convolution matrix is built from $\mathbf{x}$, it has dimensions  

$$
(n_x + n_h - 1) \times n_h
$$  

and therefore  

$$
n_2 = n_x n_h + n_h^2 - n_h
$$  

entries.

---

For the case $n_x > n_h$ and $n_h > 0$, it holds:  

$$
\begin{aligned}
n_1 &> n_2 \\
n_x n_h + n_x^2 - n_x &> n_x n_h + n_h^2 - n_h \\
n_x^2 - n_x &> n_h^2 - n_h
\end{aligned}
$$

In order to minimize the number of entries in the convolution matrix, the matrix should be built from the **longer sequence**.
