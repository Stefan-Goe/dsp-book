# Exercise 2.14: Correlation Series

Beside the moments of exercise~\ref{dsv2a11e} there are some more  
characteristical parameters for stochastic processes, i.e. *the  
autocorrelation, cross-correlation and the auto-covariance* resp.  
*cross-covariance*.

The common property of these functions is: they are values for  
the dependency of the observed processes and their random  
variables. In this exercise the autocorrelation series is observed  
exemplarily. The following stochastic processes are expected to be  
*ergodic*, see section (2.6). A main problem is the  
estimation of the theoretical autocorrelation series (2.6.6):

$$
r_{XX}(\kappa) = E\{X^\ast(k)\, X(k+\kappa)\}
$$

based on a finite number $N$ of samples.

In Python, the autocorrelation series can be estimated using functions like `numpy.correlate()` or `scipy.signal.correlate()`.  
Calling `r = numpy.correlate(x, x, mode='full')` creates a non-normalized autocorrelation series from vector `x` using the formula  

$$
\hat{r}_{XX}(\kappa) = \sum_{k=0}^{N-|\kappa|-1} x(k)\,x^{\ast}(k+\kappa)
$$

In contrast to eq. (2.6.6), the autocorrelation series is defined conjugate even using Python. Therefore the results are always conjugated.

Calling a normalized autocorrelation can be implemented by dividing by different normalization factors. The following normalization options correspond to:

- biased:
$$
\hat{r}_{XX}(\kappa) = \frac{1}{N} \sum_{k=0}^{N-|\kappa|-1} x(k) \, x^*(k+\kappa)
$$

- unbiased
$$
\hat{r}'_{XX}(\kappa) = \frac{1}{N - |\kappa|} \sum_{k=0}^{N - |\kappa| - 1} x(k) \, x^*(k+\kappa)
$$

- coeff  
normalizes the correlation series $\hat{r}''_{XX}(\kappa)$ to have a value of one at $\kappa=0$

**a)**  
Create a white Gaussian noise sequence of length \(N = 8000\).  
Estimate the autocorrelation series using `numpy.correlate()`  
and plot the results.  
Use all three different normalization options: `biased`, `unbiased`, and `coeff`.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import correlate
from scipy.fftpack import fft
from helper_functions import ldtft
# BEGIN SOLUTION
n = np.random.randn(800)
kappa = np.arange(-799, 800)

s_biased = correlate(n, n, 'full') / len(n)
s_unbiased = correlate(n, n, 'full') / (len(n) - np.abs(np.arange(-len(n) + 1, len(n))))
s_coeff = correlate(n, n, 'full') / np.max(correlate(n, n, 'full'))

# Plotting
# Autocorrelation Sequence: Biased
plt.figure()
plt.stem(kappa, s_biased)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Biased')
plt.show()

# Autocorrelation Sequence: Unbiased
plt.figure()
plt.stem(kappa, s_unbiased)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Unbiased')
plt.show()

# Autocorrelation Sequence: Normalized at time-lag Zero
plt.figure()
plt.stem(kappa, s_coeff)
plt.grid()
plt.xlabel('kappa')
plt.ylabel('Amplitude')
plt.title('Autocorrelation Sequence: Normalized at time-lag Zero')
plt.show()
# END SOLUTION
```

**b)**  
The Fourier transform of the autocorrelation sequence is the (periodic) power spectral density.  
Use your Python function `ldtft()` from the file `helper_functions.py`  
and plot the estimate of the power spectral density.

```python
# BEGIN SOLUTION
n = np.random.randn(200)
s = correlate(n, n, 'full') / len(n)
S, W = ldtft(s, 512)

# Plotting
plt.figure()
plt.semilogy(W / np.pi, np.abs(S))  # normalized frequency
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel(r'$|S_{xx}(e^{j\Omega})|$')
plt.title('Power Spectral Density')
plt.show()
# END SOLUTION
```
