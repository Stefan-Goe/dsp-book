# Remez Design (Chebyshev-Approximation)

By designing FIR filters using Chebyshev approximation, it was shown that this approximation delivers a great benefit because it utilizes the whole tolerance scheme. Such a design method (applying a Chebyshev approximation) for FIR systems is implemented in Python’s `scipy.signal.remez()` function (Parks–McClellan algorithm). The approximation takes place in the pass and stop bands.

The call of `b = remez(m, F, G)` designs an \(m\)-th order filter, whose specification is given by vectors `F` and `G`. `F` includes a pair of frequencies (between 0 and 1) to set the pass or stop bands and `G` is the frequency response (amplitude) corresponding to `F`.

**Example:**
```python
from scipy.signal import remez

F = [0, 0.3, 0.5, 1]
G = [1, 1, 0, 0]

## Tasks

Calculate a filter with the given frequency response. Experiment with different orders and count the number of extreme values within the approximation interval.

**b)**  
The call of `b = remez(m, F, G, W)` weights a desired frequency response, where frequency bands with a lower weight are less accurately approximated than frequency bands with a higher weight. The number of weights in vector `W` is half the number of entries in `F` or `G` (for each frequency band one weight factor).

The frequency band between 0.0 and 0.3 (pass-band) doesn’t have to approximate the desired frequency response very closely. A higher weight should be used for the frequency band between 0.5 and 1.  
Calculate the filter again and compare the frequency response with the filter designed in task a).

**c)**  
Design the low-pass – high-pass combination as specified below, where the passbands should have a (by factor 4) higher weight than the stopbands:

| Band        | Frequency Range |
|-------------|----------------|
| Passband 1  | 0 – 0.2        |
| Stopband 1  | 0.26 – 0.44    |
| Passband 2  | 0.5 – 0.7      |
| Stopband 2  | 0.76 – 1       |

Filter order $m$: 128

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import remez, freqz
# BEGIN SOLUTION
# #####  Part a: Design Filters  ######################################
f = [0, 0.3, 0.5, 1.0]  # Frequencies of passband and stopband
g = [1, 0]              # Desired frequency response
w = [1,1]
v_m = [9, 10]           # Vector with filter orders

f_plot = [f[0], f[1], f[2], f[3]]
g_plot = [g[0], g[0], g[1], g[1]]

for m in v_m:
    b = remez(m + 1, f, g, weight=w, fs=2.0)
    W, H = freqz(b, worN=512)  # Frequency response calculation

    # # Graphical output # #
    string = f'Desired/Actual Response - Filter Order: {m}'
    plt.figure()
    plt.plot(f_plot, g_plot, label='Desired Response')
    plt.plot(W / np.pi, np.abs(H), label='Actual Response')
    plt.axhline(y=2 - np.max(np.abs(H)), xmin=0, xmax=0.3, linestyle='--')
    plt.axhline(y=np.max(np.abs(H)), xmin=0, xmax=0.5, linestyle='--')
    plt.axhline(y=np.max(np.abs(H[400:])), xmin=0.5, xmax=1, linestyle='--')
    plt.axvline(x=0.3, ymin=2 - np.max(np.abs(H)), ymax=0, linestyle='--')
    plt.axvline(x=0.5, ymin=np.max(np.abs(H)), ymax=np.max(np.abs(H[400:])), linestyle='--')
    plt.grid()
    plt.xlabel('Omega/pi')
    plt.ylabel('|H(exp(j*Omega))|')
    plt.title(string)
    plt.legend()

# #####  Part b: Weight the Frequency Response  ##################################
f = [0, 0.3, 0.5, 1.0]  # Frequencies of passband and stopband
g = [1, 0]              # Desired frequency response
w = [0.5, 2]            # Weighting of passband and stopband
v_m = [9, 10]           # Vector with filter orders

f_plot = [f[0], f[1], f[2], f[3]]
g_plot = [g[0], g[0], g[1], g[1]]

for m in v_m:
    b = remez(m + 1, f, g, weight=w, fs=2.0)
    W, H = freqz(b, worN=512)
    
    # # Graphical output # #
    string = f'Desired/Actual Response - Filter Order: {m}'
    plt.figure()
    plt.plot(f_plot, g_plot, label='Desired Response')
    plt.plot(W / np.pi, np.abs(H), label='Actual Response')
    plt.axhline(y=2 - np.max(np.abs(H)), xmin=0, xmax=0.3, linestyle='--')
    plt.axhline(y=np.max(np.abs(H)), xmin=0, xmax=0.5, linestyle='--')
    plt.axhline(y=np.max(np.abs(H[400:])), xmin=0.5, xmax=1, linestyle='--')
    plt.axvline(x=0.3, ymin=2 - np.max(np.abs(H)), ymax=0, linestyle='--')
    plt.axvline(x=0.5, ymin=np.max(np.abs(H)), ymax=np.max(np.abs(H[400:])), linestyle='--')
    plt.grid()
    plt.xlabel('Omega/pi')
    plt.ylabel('|H(exp(j*Omega))|')
    plt.title(string)
    plt.legend()

# #####  Part c: Low-pass Band-pass Combination  #########################
f = [0, 0.2, 0.26, 0.44, 0.5, 0.7, 0.76, 1]  # Vector of corner frequencies
g = [1,0,1,0]                 # Desired frequency response
w = [1, 4, 1, 4]              # Vector with weights
m = 128                       # Filter order

b = remez(m + 1, f, g, weight=w, fs=2.0)
W, H = freqz(b, worN=512)

# # Graphical output # #
plt.figure()
plt.plot(W / np.pi, 20 * np.log10(np.abs(H)))
plt.grid()
plt.xlabel('Omega/pi')
plt.ylabel('|H(exp(j*Omega))| in dB')
plt.title('LP-BP Combination')

plt.show()
# END SOLUTION
````
