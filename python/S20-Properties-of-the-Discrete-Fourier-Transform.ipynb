x_1(k) = x((k+\lambda) \bmod M)

$X_1(n) = \text{DFT of } x_1(k)$ and $X(n)$?


**d)** Calculate the discrete Fourier transform $X_2(n)$ of  

$$
x_2(k) = x(k) \cdot \exp\Big(-j \, 2\pi \, 3 k / 16\Big)
$$

What is the relation between $X_2(n)$ and $X(n)$?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
# #####  Part a: Compute DFT of x(k)  #####
kk = np.arange(16)
x = kk + 1j * (7 - kk)
xr = np.real(x)
xi = np.imag(x)

X = np.fft.fft(x)
Xr = np.real(X)
Xi = np.imag(X)

# # Graphical Output # #
# Real part
plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

# Imaginary part
plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part b: DFT of Re{x(k)} and Im{x(k)}  #####
# # Graphical Output # #
plt.figure()
plt.stem(kk, np.real(np.fft.fft(xr)))
plt.title('Re(FFT(Re(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(np.fft.fft(xr)))
plt.title('Im(FFT(Re(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(np.fft.fft(1j * xi)))
plt.title('Re(FFT(j*Im(x(k))))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(np.fft.fft(1j * xi)))
plt.title('Im(FFT(j*Im(x(k))))')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part c: Cyclic shift  #####
xtmp = np.concatenate([x, x])
x1_04 = xtmp[kk + 5]
X1_04 = np.fft.fft(x1_04)
x1_08 = xtmp[kk + 9]
X1_08 = np.fft.fft(x1_08)

plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X1_04))
plt.title('Re(X1(n)), lambda=4')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X1_04))
plt.title('Im(X1(n)), lambda=4')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X1_08))
plt.title('Re(X1(n)), lambda=8')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X1_08))
plt.title('Im(X1(n)), lambda=8')
plt.xlabel('n')
plt.grid(True)
plt.show()
input('....Press RETURN for the next part')

# #####  Part d: Multiplication with exp(j*...)  #####
x2 = x * np.exp(-1j * 2 * np.pi * 3 * kk / 16)
X2 = np.fft.fft(x2)

plt.figure()
plt.stem(kk, Xr)
plt.title('Re(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, Xi)
plt.title('Im(X(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.real(X2))
plt.title('Re(X2(n))')
plt.xlabel('n')
plt.grid(True)

plt.figure()
plt.stem(kk, np.imag(X2))
plt.title('Im(X2(n))')
plt.xlabel('n')
plt.grid(True)
plt.show()
# END SOLUTION
```

# Solution: Properties of the DFT

## Part (a)

For a better illustration of the properties of the discrete Fourier transform (DFT), a periodic signal with 16 periods and not a finite signal with length 16 (like given in the exercise question) is considered. The DFT of $x(k)$ (real and imaginary part) is plotted beside the figures of exercise part (b).

## Part (b)

The DFT of a real (imaginary) signal is conjugate even (odd), that means the real part is even (odd) with respect to the frequency $0$ and the imaginary part is odd (even). To understand the symmetry properties, it has to be taken into account that the spectra are continuously periodic in both the positive and negative directions. From table 6.1, it can be seen how to calculate $\Re X(n)$ (or $\Im X(n)$) from $X(n)$.

## Part (c)

According to table 6.1, a displacement in the time domain by $\lambda$ samples corresponds to a multiplication of $X(n)$ with $\exp(j\,2\pi n\lambda/N)$ in the frequency domain. For $N=16$ and $\lambda =4$ resp. $\lambda =8$ we obtain:

$$
X_1(n) = j^n X(n) \quad \text{resp.} \quad X_1(n) = (-1)^n X(n)
$$

## Part (d)

According to table 6.1, it is obtained that:

$$
\text{DFT}\{ \exp(-j\,2\pi i k/N) x(k) \} = X((n+i))_N
$$

The notation $((i))_N$ describes a modulo-$N$ indexing. The multiplication of $x(k)$ with $\exp(-j\,2\pi\,3k/16)$ leads to a frequency offset of $3 f_A/16$, i.e.

$$
X_2(n) = X((n+3))_{16}
$$
