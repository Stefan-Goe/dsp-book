# Exercise: DFT of Real-Valued Series

By using a discrete Fourier Transform (DFT) to transform a real-valued series, the result contains redundancy: it is conjugate-even.

This property can be used to transform a real-valued series $x_1(k)$ of length $2N$ using an $N$-point FFT.  

**Task:**

- Write a *Python function* that exploits this property.
- Write a small script to demonstrate the correctness of this approach.

```python
import numpy as np
from helper_functions import lfft
# BEGIN SOLUTION
N = 2**10
x = np.random.randn(N)
X1 = np.fft.fft(x)
X2 = lfft(x)
difference = np.mean(np.abs(X1 - X2) ** 2)
print(difference)
# END SOLUTION
```

The Python function `lfft` is implemented in `helper_functions.py`.  
It realizes the algorithm (described in paragraph 7.1) to transform a real-valued series $x_1(k)$ of length $2N$ using an $N$-point FFT.
