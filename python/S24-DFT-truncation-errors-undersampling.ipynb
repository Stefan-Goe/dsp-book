### Exercise: DFT: truncation error / sub-sampling

Calculate the discrete Fourier transform  

$$
X_i(n) = \mathrm{DFT}\{x_i(k)\}
$$  

of the following series:

\begin{aligned}
x_1(k) &= \sin(k \pi / 8), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63 \\
x_2(k) &= \sin(k \pi / 2), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63 \\
x_3(k) &= \sin(3 k \pi / 2), &\text{for } k = 0\ldots 7 \text{ as well as } k = 0\ldots 63
\end{aligned}

Use Python's `numpy.fft.fft(x)` function.

Plot the time domain signal $x_i(k)$, as well as the absolute values of the spectra $|X_i(n)|$ for both ranges of time indices.

In which cases is the spectrum of the associated continuous sine-signal $x_i(t)$ reproduced exactly?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
kk08 = np.arange(8)
kk64 = np.arange(64)
kk100 = np.arange(0, 8, 8/100)

x1_08 = np.sin(kk08 * np.pi / 8)
x1_64 = np.sin(kk64 * np.pi / 8)
x2_08 = np.sin(kk08 * np.pi / 2)
x2_64 = np.sin(kk64 * np.pi / 2)
x3_08 = np.sin(kk08 * np.pi * 3 / 2)
x3_64 = np.sin(kk64 * np.pi * 3 / 2)

# Plot x1(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi / 8), ':')
plt.stem(kk08, x1_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x1(k)=sin(k*pi/8), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x1_08)))
plt.axis([0, 7, -1, 6])
plt.title('abs(X1(n)), n=0...7')

# Plot x1(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x1_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x1(k)=sin(k*pi/8), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x1_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X1(n)), n=0...63')

input('....Press RETURN to continue: ')

# Plot x2(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi / 2), ':')
plt.stem(kk08, x2_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x2(k)=sin(k*pi/2), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x2_08)))
plt.axis([0, 7, -1, 6])
plt.title('abs(X2(n)), n=0...7')

# Plot x2(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x2_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x2(k)=sin(k*pi/2), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x2_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X2(n)), n=0...63')

input('....Press RETURN to continue: ')

# Plot x3(k) for k = 0...7 and its DFT
plt.figure()
plt.plot(kk100, np.sin(kk100 * np.pi * 3 / 2), ':')
plt.stem(kk08, x3_08)
plt.axis([0, 8, -1.2, 1.2])
plt.title('x3(k)=sin(k*pi*3/2), k=0...7')

plt.figure()
plt.stem(kk08, np.abs(np.fft.fft(x3_08)))
plt.axis([0, 8, -1, 6])
plt.title('abs(X3(n)), n=0...7')

# Plot x3(k) for k = 0...63 and its DFT
plt.figure()
plt.stem(kk64, x3_64)
plt.axis([0, 64, -1.2, 1.2])
plt.title('x3(k)=sin(k*pi*3/2), k=0...63')

plt.figure()
plt.stem(kk64, np.abs(np.fft.fft(x3_64)))
plt.axis([0, 64, -1, 40])
plt.title('abs(X3(n)), n=0...63')

plt.show()

# END SOLUTION
```

### Solution: DFT: truncation error / sub-sampling

$\tilde{x}_\nu(k)$ is the periodic continuation of the series $x_\nu(k)$. The spectrum of the corresponding continuous sinusoidal signal $x_\nu(t)$ with period $T$ is exactly described by $\mathrm{DFT}\{\tilde{x}_\nu(k)\}$, if

1. $\tilde{x}_\nu(k)$ is identical with the sampled continuous sinusoidal signal $x_\nu(t)|_{t=kT_A}$  
2. AND the sampling theorem is fulfilled, i.e. the sampling rate $1/T_A$ is higher than double the frequency of the continuous sinusoidal signal $x_\nu(t)$.  
   In other words: $T_A < T/2$

---

**Signal $x_1(k)$:** In both cases, the sampling theorem is fulfilled (the continuous sinusoidal signal is plotted dotted). For $k=0..7$, condition (1) is however violated, so that $X_1(n)$ shows spectral components at many frequencies  (instead of only two). For $k=0..63$, both conditions are fulfilled. The spectrum of the continuous sinusoidal signal is obtained.

**Signal $x_3(k)$:** The sampling theorem is violated here.  
Even though the frequency of the continuous sinusoidal signal
is higher than $x_2(t)$, the sampling rate is not increased
accordingly. Because $x_3(k)$ corresponds with $x_2(k)$ until
the sign, the FFT yields exactly the same results as the signal
$x_2(k)$. The spectrum $X_3(n)$ does not describe the spectrum of
the continuous sinusoidal signal $x_3(t)$.
