### Exercise: Comparison of window functions

Plot the coefficients $f(k)$ and the absolute value of the normalized spectra $F(e^{j\Omega}) / F(e^{j0})$ (in dB) of a rectangular, Hann-, Hamming-, Blackman-, and a Dolph-Chebyshev window ($48.6\ \mathrm{dB}$ and $76.1\ \mathrm{dB}$ stop-band attenuation)
 
with $N=63$ coefficients.

Compare the stop band attenuations and the spectral resolution obtained with windowing.

---

**Python hints:**  

The window functions of length $N$ can be imported from `scipy.signal`:

```python
from scipy.signal import boxcar, hanning, hamming, blackman, chebwin

# Example windows
rect = boxcar(N)            
hann = hanning(N)           
hamm = hamming(N)          
black = blackman(N)         


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import boxcar, hamming, blackman, chebwin
# BEGIN SOLUTION
N = 63
NFFT = 2**9
f_re = boxcar(N)
aF_re = np.abs(np.fft.fft(f_re, NFFT) / np.sum(f_re))
f_hn = np.hanning(N)
aF_hn = np.abs(np.fft.fft(f_hn, NFFT) / np.sum(f_hn))
f_hm = hamming(N)
aF_hm = np.abs(np.fft.fft(f_hm, NFFT) / np.sum(f_hm))
f_bl = blackman(N)
aF_bl = np.abs(np.fft.fft(f_bl, NFFT) / np.sum(f_bl))
a_d1 = 48.6
f_d1 = chebwin(N, at=a_d1)
aF_d1 = np.abs(np.fft.fft(f_d1, NFFT) / np.sum(f_d1))
a_d2 = 76.1
f_d2 = chebwin(N, at=a_d2)
aF_d2 = np.abs(np.fft.fft(f_d2, NFFT) / np.sum(f_d2))
kk = np.arange(N)
fT = np.arange(0, 1, 1/NFFT)

# Calculation of a_min (fsT:=Omega_s/(2*pi); Stopband: n_s+1<=n<=NFFT+1-n_s)
fsT = 1/N
n_s = int(round(NFFT * fsT))
a_re = 20 * np.log10(np.max(aF_re[n_s+1:NFFT+1-n_s]))
fsT = 2/N
n_s = int(round(NFFT * fsT))
a_hn = 20 * np.log10(np.max(aF_hn[n_s+1:NFFT+1-n_s]))
fsT = 2/N
n_s = int(round(NFFT * fsT))
a_hm = 20 * np.log10(np.max(aF_hm[n_s+1:NFFT+1-n_s]))
fsT = 3/N
n_s = int(round(NFFT * fsT))
a_bl = 20 * np.log10(np.max(aF_bl[n_s+1:NFFT+1-n_s]))

# Graphical Output
# RECTANGLE Window
plt.figure()
plt.bar(kk, f_re)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Rectangle Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()

plt.figure()
plt.plot(fT, 20 * np.log10(aF_re + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_re+13, f'a_min = {a_re:.1f} dB')
plt.plot([0, 1.9], [a_re, a_re], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# HANNING Window
plt.figure()
plt.bar(kk, f_hn)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Hanning Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_hn + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_hn+15, f'a_min = {a_hn:.1f} dB')
plt.plot([0, 1.9], [a_hn, a_hn], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# HAMMING Window
plt.figure()
plt.bar(kk, f_hm)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Hamming Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_hm + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_hm+15, f'a_min = {a_hm:.1f} dB')
plt.plot([0, 1.9], [a_hm, a_hm], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# BLACKMAN Window
plt.figure()
plt.bar(kk, f_bl)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title('Blackman Window')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, a_bl+15, f'a_min = {a_bl:.1f} dB')
plt.plot([0, 1.9], [a_bl, a_bl], '--')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# DOLPH-CHEBYSHEV-1
plt.figure()
plt.bar(kk, f_d1)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title(f'Dolph-Cheb., {a_d1} dB')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()


plt.figure()
plt.plot(fT, 20 * np.log10(aF_d1 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -150, 10])
plt.text(.3, -a_d1+15, f'a_min = {a_d1:.1f} dB')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()


input('....Press RETURN to continue: ')

# DOLPH-CHEBYSHEV-2
plt.figure()
plt.bar(kk, f_d2)
plt.axis([-.5, N-.5, 0, 1.1])
plt.title(f'Dolph-Cheb., {a_d2} dB')
plt.ylabel('f(k)')
plt.xlabel('k')
plt.show()

# Normalized Spectral Function
plt.figure()
plt.plot(fT, 20 * np.log10(aF_d2 + np.finfo(float).eps))
plt.axis([0, 1, -150, 10])
plt.text(.3, -a_d2+15, f'a_min = {a_d2:.1f} dB')
plt.title('|Normalized Spectral Function|')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()
# END SOLUTION
````

# Comparison of Window Functions

The following figures illustrate the coefficients $f(k)$ on the left-hand side and the absolute value of the normalized spectrum 

$$
20 \log_{10} \left| \frac{F(e^{j\Omega})}{F(e^{j0})} \right|
$$ 

in dB on the right-hand side, for different windows with $N=63$ coefficients.

The **rectangular window** reaches a minimal stop band attenuation of only $a_\text{min} = 13.3\,\text{dB}$, but has a very narrow pass band $|\Omega| \le \Omega_s = 2\pi/N \approx 0.1$, which gives it very good spectral resolution.

With the **Hann window**, the oscillations in the stop band are clearly attenuated ($a_\text{min} = 31.5\,\text{dB}$), mitigating leakage effects. However, this comes at the cost of spectral resolution, because the pass band is twice as wide compared to the rectangular window: $\Omega_s = 4\pi/N \approx 0.2$.

The **Hamming window** minimizes the first sidelobe of the stop band and maintains almost constant attenuation over the entire stop band ($a_\text{min} = 42.5\,\text{dB}$). The pass band is the same as for the Hann window: $\Omega_s = 4\pi/N \approx 0.2$. Note that the first and last coefficients ($f(0)$ and $f(N-1)$) are not zero.

The last traditional window, the **Blackman window**, achieves an even higher stop band attenuation ($58.2\,\text{dB}$), but again with a lower spectral resolution: $\Omega_s = 6\pi/N \approx 0.3$.

**Figures: Blackman**

The **Dolph-Chebyshev window** has a uniform approximation in the stop band, i.e., the minimal attenuation is constant for all frequencies $\Omega_s \le \Omega \le \pi$. Unlike traditional windows, the window length $N$ does **not** define the stop band cutoff frequency $\Omega_s$. If either the value of $a_\text{min}$ or $\Omega_s$ is chosen, the other value is fixed, and the window function can be obtained.

<img src="./pngs/ldsv7a04_dt7611.png" width="250"> <img src="./pngs/ldsv7a04_dt7612.png" width="250">  

According to equation (7.4.28), we get values $\Omega_s \approx 4\pi/N \approx 0.2$ and $6\pi/N \approx 0.3$ for $N=63$ with $a_\text{min} = 48.6$ dB and $76.1$ dB, respectively. At the same $N$ and $\Omega_s$, the window reaches a higher minimal stop band attenuation, about $6\,\text{dB}$ higher than the corresponding Hamming window and $18\,\text{dB}$ higher than the Blackman window.

For any fixed window length $N$, the following table summarizes the relationship between the reachable minimal stop band attenuation $a_\text{min}$ and the stop band cutoff frequency $\Omega_s$:

| Window Function | Rectangular | Hann | Hamming | Blackman | Dolph-Chebyshev  | Dolph-Chebyshev |
|-----------------|------------|------|---------|----------|-----------------------------|-----------------------------|
| Minimal Stop Band Attenuation $a_\text{min}$ | $13.3\,\text{dB}$ | $31.5\,\text{dB}$ | $42.5\,\text{dB}$ | $58.2\,\text{dB}$ | $48.6\,\text{dB}$ | $76.1\,\text{dB}$ |
| Pass Band |$\Omega$| $\le$ $\Omega_s$ | $2\pi/N$ | $4\pi/N$ | $4\pi/N$ | $6\pi/N$ | $\approx 4\pi/N$ | $\approx 6\pi/N$ |
| Notes | Narrowest pass band, poor stop band | Oscillations attenuated, moderate resolution | Minimizes first sidelobe, moderate resolution | High attenuation, lower resolution | Uniform stop band, stop band set by $a_\text{min}$ or $\Omega_s$ | Uniform stop band, stop band set by $a_\text{min}$ or $\Omega_s$ |

**Note:** For traditional windows, the $a_\text{min}$ values are **independent of the window length $N$**. Increasing $N$ only compresses the sidebands together — their maxima remain unchanged. This is not the case for the Dolph-Chebyshev window.
