### Exercise — Side Band Attenuation of the Blackman Window

Plot the absolute value of the normalized spectrum  
$$
\frac{F(\mathrm{e}^{j\Omega})}{F(\mathrm{e}^{j0})}
$$
(in dB) of a Blackman window with  
$N = 32$, $N = 64$ resp.~$N = 128$ coefficients.  

Determine (for these values of $N$) the ratio  
of the maximum value and the second greatest value.

---

Which effect can be observed here as well as with all other  
traditional window functions?

```python
import numpy as np
import matplotlib.pyplot as plt
# BEGIN SOLUTION
NFFT = 2**10
f_bl032 = np.blackman(32)
aF_bl032 = np.abs(np.fft.fft(f_bl032, NFFT) / np.sum(f_bl032))
f_bl064 = np.blackman(64)
aF_bl064 = np.abs(np.fft.fft(f_bl064, NFFT) / np.sum(f_bl064))
f_bl128 = np.blackman(128)
aF_bl128 = np.abs(np.fft.fft(f_bl128, NFFT) / np.sum(f_bl128))
fT = np.arange(0, 1, 1/NFFT)

# Calculation of minimum stopband attenuation a_min:
# Os := Omega_s/(2*pi);  n_s: stopband range indices: n_s+1<=n<=NFFT+1-n_s
Os = 3/32
n_s = round(NFFT * Os)
a032 = 20 * np.log10(np.max(aF_bl032[n_s+1:NFFT+1-n_s]))
Os = 3/64
n_s = round(NFFT * Os)
a064 = 20 * np.log10(np.max(aF_bl064[n_s+1:NFFT+1-n_s]))
Os = 3/128
n_s = round(NFFT * Os)
a128 = 20 * np.log10(np.max(aF_bl128[n_s+1:NFFT+1-n_s]))

# Graphical Output
plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl032 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a032, f'a_min = {a032:.1f} dB')
plt.plot([0, 0.39], [a032, a032], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=32')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')

plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl064 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a064, f'a_min = {a064:.1f} dB')
plt.plot([0, 0.39], [a064, a064], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=64')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')

plt.figure()
plt.plot(fT, 20 * np.log10(aF_bl128 + np.finfo(float).eps))
#plt.hold(True)
plt.axis([0, 1, -180, 10])
plt.text(0.41, a128, f'a_min = {a128:.1f} dB')
plt.plot([0, 0.39], [a128, a128], '--')
plt.title('|Normalized Spectral Function| of the Blackman Window:  N=128')
plt.ylabel('|F(exp(j*Om.))/F(exp(j*0))| in dB')
plt.xlabel('Omega/2pi = f T')
plt.show()
# END SOLUTION
```

### Blackman Window

The equation below determines the Blackman window:

$$
f^{BL}(k) =
\begin{cases}
0.42 - 0.5 \cdot \cos\left(\frac{2\pi k}{N - 1}\right) + 0.08 \cdot \cos\left(\frac{4\pi k}{N - 1}\right), & 0 \le k \le N - 1 \\
0, & \text{else}
\end{cases}
$$

With $a_{min} \approx 58\,\text{dB}$, it shows, in comparison to all other traditional windows,  
the highest stop band attenuation  
(rectangular: $13.3\,\text{dB}$, Hann: $31.5\,\text{dB}$, Hamming: $42.5\,\text{dB}$).  

On the other hand, the strong attenuation must be paid off with a wide passband  
$\Omega_s = \frac{6\pi}{N}$ (rectangular: $\frac{2\pi}{N}$, Hann and Hamming: $\frac{4\pi}{N}$).

We confirm:
1. The stop band cut-off frequency is inversely proportional to the length of the window:  
   $\Omega_s \sim \frac{1}{N}$.
2. The minimum stop band attenuation only changes insignificantly with the window length:  
   $a_{min} \approx \text{const}_N$.

**Remark:**  
For all traditional windows (rectangular, Hann, Hamming, Blackman),  
the *side band attenuation does not depend on the length of the window* $N$.  
An increase in window length only leads to a compression of the side bands  
and can be used to improve frequency resolution.
